# Урок 21. Кипение. Удельная теплота парообразования. Зависимость температуры кипения от давления

## 1. Что происходит при кипении

Рассмотрим нагревание воды в колбе. Сначала на стенках и дне появляются пузырьки: это
выделяется при нагревании воздух, растворённый в воде. В каждый пузырёк испаряется вода,
поэтому он заполнен насыщенным паром. Пузырьки растут, отрываются и всплывают, но, попадая в верхние, ещё холодные слои, пар в них конденсируется, и пузырьки схлопываются.
Этим объясняется шум, возникающий перед закипанием.

Когда вся вода прогревается до 100 °C, пузырьки перестают схлопываться: они всплывают,
лопаются на поверхности, и пар выходит наружу. Данный процесс и представляет собой кипение.

```{admonition} Кипение
:class: important
**Кипение** представляет собой парообразование, происходящее по всему объёму жидкости
при определённой температуре. Температуру, при которой жидкость кипит, называют
**температурой кипения**.

Пока жидкость кипит, её температура **не меняется**, какое бы количество теплоты
ни подводилось. Кипение возможно, когда давление насыщенного пара внутри пузырьков
достигает внешнего давления.
```

| Вещество | $t_{\text{кип}}$, °C | Вещество | $t_{\text{кип}}$, °C |
|---|---:|---|---:|
| Водород | −253 | Спирт | 78 |
| Азот | −196 | Вода | 100 |
| Кислород | −183 | Ртуть | 357 |
| Эфир | 35 | Свинец | 1740 |
| | | Медь | 2567 |
| | | Железо | 2750 |

Значения в таблице учебника (§ 21) приведены при нормальном атмосферном давлении. Из неё
видно, почему спиртовым термометром нельзя измерить температуру кипятка (Марон, № 225)
и почему в кипящей воде закипают пробирки с эфиром и спиртом, но не с водой (Марон, № 224):
для кипения жидкость должна получать теплоту, а вода при 100 °C от воды при 100 °C
её не получает.

## 2. Температура кипения зависит от давления

Кипение начинается, когда давление пара в пузырьках сравнивается с внешним. При уменьшении
внешнего давления пузырькам легче расти, и жидкость закипает при меньшей температуре.
На вершине Эльбруса вода кипит около 83 °C, а в шприце с водой при резком оттягивании
поршня закипает при комнатной температуре (Марон, № 264). При увеличении давления кипение
наступает позже: в скороварке под клапаном вода кипит при 110–120 °C, в автоклавах,
описанных в учебнике, при 135 °C, поэтому в них стерилизуют инструменты.

В упражнении 18 учебника рассматривается спирт, закипевший в закрытом сосуде при 50 °C, и требуется определить, что можно сказать о давлении. Давление ниже атмосферного, так как при нормальном давлении спирт кипит при 78 °C.

```{admonition} Куда идёт огонь под кипящей кастрюлей
:class: note
Девочка увеличила подачу газа вчетверо, чтобы картофель сварился быстрее (учебник,
упражнение 18). Быстрее картофель не сварится: температура кипящей воды останется равной
100 °C, а вся дополнительная теплота расходуется на более быстрое выкипание воды.
Единственным способом повысить температуру варки является повышение давления, для чего служит скороварка.
```

## 3. Удельная теплота парообразования

При кипении температура остаётся постоянной, тогда как теплота продолжает поступать. Она
расходуется на отрыв молекул друг от друга: расстояния между ними при переходе в пар
возрастают в десятки раз, и на это требуется энергия, значительно большая, чем при
плавлении.

```{admonition} Удельная теплота парообразования
:class: important
Количество теплоты, необходимое для превращения жидкости массой 1 кг в пар при температуре
кипения, называют **удельной теплотой парообразования**. Обозначается буквой $L$, единицей
измерения служит Дж/кг.

$$Q = L\,m.$$

При конденсации пара массой $m$ **выделяется** такое же количество теплоты.
```

| Вещество | $L$, 10⁶ Дж/кг | Вещество | $L$, 10⁶ Дж/кг |
|---|---:|---|---:|
| Вода | 2,3 | Ртуть | 0,3 |
| Аммиак (жидкий) | 1,4 | Эфир | 0,4 |
| Спирт | 0,9 | Воздух (жидкий) | 0,2 |

У воды $L = 2{,}3 \cdot 10^6$ Дж/кг, что в семь раз больше удельной теплоты плавления льда
и в пять с половиной раз больше количества теплоты, необходимого для нагревания той же воды
от 0 до 100 °C. По этой причине чайник закипает за пять минут, а выкипает за полчаса.

Полный график для льда, взятого при −40 °C, теперь можно довести до конца.

````python
from phys import sim
sim.heating_curve(boil=True)
````

Вторая полка, соответствующая кипению, почти в семь раз длиннее первой. Из подписей внизу
видно, что нагрев воды от 0 до 100 °C требует 420 кДж на килограмм, тогда как её выкипание
требует 2300 кДж.

Пример учебника (§ 22). Определим количество теплоты, необходимое для того, чтобы 3 кг эфира
при 20 °C обратить в пар.

````python
c_ef, L_ef, t_kip = 2350, 0.4e6, 35
m, t0 = 3, 20
Q1 = c_ef * m * (t_kip - t0)
Q2 = L_ef * m
print(f"нагрев до кипения {Q1:.0f} Дж, парообразование {Q2:.0f} Дж, всего {(Q1 + Q2) / 1e6:.1f} МДж")
````

Ответ учебника составляет $1{,}3 \cdot 10^6$ Дж, что совпадает с полученным. На нагрев расходуется менее десятой части теплоты, а основная её часть идёт на само превращение в пар.

## 4. Почему ожог паром страшнее кипятка

В задаче Марона № 223 требуется объяснить, почему ожоги стоградусным паром сильнее, чем кипятком той же температуры.
Пар, коснувшись кожи, конденсируется и отдаёт $L m$ (2,3 МДж на килограмм), и только после этого начинает остывать как вода. Кипяток отдаёт лишь теплоту остывания. Следовательно, на грамм вещества пар приносит почти в десять раз больше энергии, чем кипяток.

```{admonition} Опыт 1. Термометр в кипящей воде (демонстрация из рабочей программы)
:class: opyt
**Что нужно:** кастрюля с водой, термометр до 110 °C (лабораторный или кулинарный), плита.

за термометром ещё 3 минуты.

**Результат:** температура растёт, затем останавливается на 100 °C (несколько ниже, если
измерения проводятся не на уровне моря) и не изменяется при любом увеличении огня. Меняется
только скорость выкипания.
```


```{admonition} Опыт 3. Кипение при комнатной температуре (Марон, № 264)
:class: opyt
Наберите в медицинский шприц немного тёплой воды, закройте носик пальцем и резко потяните

```

```{admonition} Осторожно
:class: tb
Все опыты с кипятком проводятся под наблюдением взрослого. Термометр держат за верхнюю
часть, не касаясь дна кастрюли. Пар от кипящей воды обжигает сильнее самой воды, поэтому
```

```{admonition} Копнуть глубже: заморозить воду, откачивая пар (Савченко, 5.10.5)
:class: dropdown
В колбе находится вода при 0 °C. Пар над ней непрерывно откачивают. Испарение отбирает
у воды теплоту, и она начинает замерзать. Требуется определить, какая часть воды испарится
к моменту, когда остальная замёрзнет.

Пусть испарилось $m_1$, замёрзло $m_2$. Испарение забрало $L m_1$, а кристаллизация выделила $\lambda m_2$, и в теплоизолированной колбе эти величины равны: $L m_1 = \lambda m_2$. Тогда
доля испарившейся воды

$$\frac{m_1}{m_1 + m_2} = \frac{\lambda}{L + \lambda} = \frac{3{,}4}{23 + 3{,}4} \approx 0{,}13.$$

Следовательно, испаряется около 13 % воды, а остальные 87 % превращаются в лёд. Подобным
образом получают лёд без холодильника, за счёт испарения, и на этом же принципе работает обычный холодильник, только испаряется в нём не вода.
```
