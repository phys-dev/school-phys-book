# Урок 21. Кипение. Удельная теплота парообразования. Зависимость температуры кипения от давления

## 1. Что происходит при кипении

Нагреваем воду в колбе и следим за ней. Сначала на стенках и дне появляются пузырьки:
это воздух, растворённый в воде, выделяется при нагревании. В каждый пузырёк испаряется
вода, и он заполнен насыщенным паром. Пузырьки растут, отрываются и всплывают — но,
попав в верхние, ещё холодные слои, пар в них конденсируется, пузырьки схлопываются.
Отсюда шум перед закипанием.

Когда вся вода прогреется до 100 °C, пузырьки перестают схлопываться: они всплывают,
лопаются на поверхности, и пар выходит наружу. Это кипение.

```{admonition} Кипение
:class: important
**Кипение** — парообразование, происходящее по всему объёму жидкости при определённой
температуре. Температуру, при которой жидкость кипит, называют **температурой кипения**.

Пока жидкость кипит, её температура **не меняется**, сколько бы теплоты ни подводили.
Кипение возможно, когда давление насыщенного пара внутри пузырьков достигает внешнего
давления.
```

| Вещество | $t_{\text{кип}}$, °C | Вещество | $t_{\text{кип}}$, °C |
|---|---:|---|---:|
| Водород | −253 | Спирт | 78 |
| Азот | −196 | Вода | 100 |
| Кислород | −183 | Ртуть | 357 |
| Эфир | 35 | Свинец | 1740 |
| | | Медь | 2567 |
| | | Железо | 2750 |

Таблица учебника (§ 21) — при нормальном атмосферном давлении. Из неё видно, почему
спиртовым термометром нельзя измерить температуру кипятка (Марон, № 225) и почему в кипящей
воде закипят пробирки с эфиром и спиртом, но не с водой (Марон, № 224): чтобы жидкость
кипела, ей нужно получать теплоту, а вода при 100 °C от воды при 100 °C её не получит.

## 2. Температура кипения зависит от давления

Кипение начинается, когда давление пара в пузырьках сравнивается с внешним. Уменьшите
внешнее давление — пузырькам легче расти, и жидкость закипит при меньшей температуре.
На вершине Эльбруса вода кипит около 83 °C, в шприце с водой при резком оттягивании поршня —
при комнатной температуре (Марон, № 264). Увеличьте давление — кипение отодвинется:
в скороварке под клапаном вода кипит при 110–120 °C, в автоклавах учебника — при 135 °C,
поэтому в них стерилизуют инструменты.

Упражнение 18 учебника задаёт вопрос: спирт в закрытом сосуде закипел при 50 °C. Что
можно сказать о давлении? Оно ниже атмосферного — при нормальном давлении спирт кипит
при 78 °C.

```{admonition} Куда идёт огонь под кипящей кастрюлей
:class: note
Девочка увеличила подачу газа вчетверо, чтобы картофель сварился быстрее (учебник,
упражнение 18). Быстрее не сварится: температура кипящей воды останется 100 °C, вся лишняя
теплота уйдёт на то, чтобы вода быстрее выкипала. Единственный способ повысить температуру
варки — поднять давление, то есть взять скороварку.
```

## 3. Удельная теплота парообразования

При кипении температура стоит, а теплота поступает. На что она уходит? На отрыв молекул
друг от друга: расстояния между ними при переходе в пар растут в десятки раз, и на это
нужна энергия — гораздо большая, чем при плавлении.

```{admonition} Удельная теплота парообразования
:class: important
Количество теплоты, необходимое для превращения жидкости массой 1 кг в пар при температуре
кипения, называют **удельной теплотой парообразования**. Обозначение $L$, единица — Дж/кг.

$$Q = L\,m.$$

При конденсации пара массой $m$ **выделяется** столько же теплоты.
```

| Вещество | $L$, 10⁶ Дж/кг | Вещество | $L$, 10⁶ Дж/кг |
|---|---:|---|---:|
| Вода | 2,3 | Ртуть | 0,8 |
| Аммиак (жидкий) | 1,4 | Эфир | 0,4 |
| Спирт | 0,9 | Воздух (жидкий) | 0,2 |

У воды $L = 2{,}3 \cdot 10^6$ Дж/кг — в семь раз больше удельной теплоты плавления льда
и в пять с половиной раз больше, чем нужно, чтобы нагреть ту же воду от 0 до 100 °C.
Вот почему чайник закипает за пять минут, а выкипает за полчаса.

Полный график для льда, взятого при −40 °C, теперь можно довести до конца.

````python
from phys import sim
sim.heating_curve(boil=True)
````

Вторая полка — кипение — почти в семь раз длиннее первой. Сравните подписи внизу: нагрев
воды от 0 до 100 °C стоит 420 кДж на килограмм, её выкипание — 2300 кДж.

Пример учебника (§ 22). Сколько теплоты нужно, чтобы 3 кг эфира при 20 °C обратить в пар?

````python
c_ef, L_ef, t_kip = 2350, 0.4e6, 35
m, t0 = 3, 20
Q1 = c_ef * m * (t_kip - t0)
Q2 = L_ef * m
print(f"нагрев до кипения: {Q1:.0f} Дж; парообразование: {Q2:.0f} Дж; всего {(Q1 + Q2) / 1e6:.1f} МДж")
````

Ответ учебника — $1{,}3 \cdot 10^6$ Дж, сходится. Заметьте, что на нагрев ушло меньше десятой
части; главное — само превращение в пар.

## 4. Почему ожог паром страшнее кипятка

Марон, № 223: почему ожоги стоградусным паром сильнее, чем кипятком той же температуры?
Пар, коснувшись кожи, конденсируется и отдаёт $L m$ — по 2,3 МДж на килограмм — и только
после этого начинает остывать как вода. Кипяток отдаёт лишь теплоту остывания. На грамм
вещества пар приносит в двадцать с лишним раз больше энергии.

```{admonition} Копнуть глубже: заморозить воду, откачивая пар (Савченко, 5.10.5)
:class: dropdown
В колбе вода при 0 °C. Пар над ней непрерывно откачивают. Испарение отбирает у воды
теплоту, и она начинает замерзать. Какая часть воды испарится к моменту, когда остальная
замёрзнет?

Пусть испарилось $m_1$, замёрзло $m_2$. Испарение забрало $L m_1$, кристаллизация выделила
$\lambda m_2$; в теплоизолированной колбе они равны: $L m_1 = \lambda m_2$. Тогда доля
испарившейся воды

$$\frac{m_1}{m_1 + m_2} = \frac{\lambda}{L + \lambda} = \frac{3{,}4}{23 + 3{,}4} \approx 0{,}13.$$

Испаряется около 13 % воды, остальные 87 % превращаются в лёд. Так получают лёд без
холодильника — испарением; на этом же принципе работает и обычный холодильник, только
испаряется в нём не вода.
```
