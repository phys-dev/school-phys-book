## 2. Три задачи, которые нужно уметь решать

**Задача А. Комбинированная, по этапам.** Лёд массой 2 кг при −10 °C превратить в пар
при 100 °C. Сколько теплоты нужно?

Строится график процесса, и каждый участок рассчитывается отдельно, а на горизонтальных участках температура не изменяется.

````python
m = 2
etapy = [("нагрев льда от −10 до 0 °C", 2100 * m * 10),
         ("плавление льда", 3.4e5 * m),
         ("нагрев воды от 0 до 100 °C", 4200 * m * 100),
         ("превращение воды в пар", 2.3e6 * m)]
itogo = 0
for nazv, Q in etapy:
    itogo += Q
    print(f"{nazv:32s} {Q / 1e3:8.0f} кДж")
print(f"{'всего':32s} {itogo / 1e3:8.0f} кДж = {itogo / 1e6:.2f} МДж")
````

Типичной ошибкой является пропуск одного из этапов или расчёт нагрева льда с теплоёмкостью
воды.

**Задача Б. Тепловой баланс с плавлением.** В 500 г воды при 40 °C бросили 100 г льда
при 0 °C. Какая температура установится?

Вода отдаёт количество теплоты $c\,m_{\text{в}}(40 - t)$. Лёд получает теплоту в два этапа:
сначала он плавится, на что расходуется $\lambda m_{\text{л}}$, затем талая вода нагревается
от 0 до $t$, на что расходуется $c\,m_{\text{л}}\,t$.


````python
c, lam = 4200, 3.4e5
m_v, t_v, m_l = 0.5, 40, 0.1
# c·m_v·(t_v − t) = λ·m_l + c·m_l·t  →  t = (c·m_v·t_v − λ·m_l) / (c·(m_v + m_l))
t = (c * m_v * t_v - lam * m_l) / (c * (m_v + m_l))
print(f"установится {t:.1f} °C")
print(f"проверка: вода отдала {c * m_v * (t_v - t):.0f} Дж, лёд получил {lam * m_l + c * m_l * t:.0f} Дж")
````

Предварительно следует убедиться, что теплоты воды достаточно для плавления всего льда:
$4200 \cdot 0{,}5 \cdot 40 = 84$ кДж против $3{,}4 \cdot 10^5 \cdot 0{,}1 = 34$ кДж, и теплоты достаточно. В противном случае часть льда осталась бы нерасплавленной, и ответом
было бы 0 °C.

**Задача В. Топливо и КПД.** Сколько дров нужно сжечь, чтобы нагреть 50 л воды от 10 до 60 °C,
если в воду попадает 30 % выделяющейся теплоты?

````python
Q_vode = 4200 * 50 * 50
Q_topliva = Q_vode / 0.30
m_drov = Q_topliva / 1.0e7
print(f"воде нужно {Q_vode / 1e6:.2f} МДж, сжечь придётся {Q_topliva / 1e6:.1f} МДж, или {m_drov:.1f} кг дров")
````

## 4. Типичные ошибки

* Разность температур переводят в кельвины, прибавляя 273. Этого делать не нужно, разность одинакова в обеих шкалах.
* Путают $c$, $q$, $\lambda$, $L$: размерность каждой из них содержит Дж/кг, но $c$ дополнительно относится к одному градусу.
* Теплоту плавления рассчитывают по формуле с $\Delta t$. При плавлении $\Delta t = 0$,
  и применяется формула $Q = \lambda m$.
* В уравнении баланса забывают, что лёд после таяния становится водой и **нагревается дальше**.
* КПД вычисляют как $A / Q_2$ или выражают в долях, тогда как требуется ответ в процентах.
* Массу подставляют в граммах, тогда как все формулы записаны для килограммов.
