## 2. Темы докладов

Тем восемь, группа выбирает одну. К каждой указаны опорные источники; докладчикам стоит
добавить один свой пример, найденный дома или в интернете.

1. **Утепление дома.** Стены с воздушным зазором, минеральная вата, пенопласт, двойные
   и тройные окна. Почему в утеплителях главное — не вещество, а поры и воздух в них.
   *Источники:* учебник, § 7 (теплопроводность газов, пористые материалы), упражнение 5;
   Марон, № 25, 32.

2. **Термос, термокружка, холодильник.** Как одно устройство перекрывает все три пути
   теплопередачи. Почему испаритель в холодильнике сверху. *Источники:* учебник, § 9
   («Как устроен термос»); Марон, № 35.

3. **Одежда и тепло.** Пух, шерсть, мех, современные мембраны; почему светлое носят летом
   и почему несколько тонких слоёв греют лучше одного толстого. *Источники:* учебник, § 7
   и § 9, упражнение 7; Марон, № 33.

4. **Как животные сберегают тепло.** Шерсть и перья как воздушная подушка, подкожный жир,
   уменьшенные уши и лапы у северных животных, тетерева в снегу. *Источники:* учебник,
   «Это любопытно. Приспособление животных к различным температурным условиям» (после § 7);
   Марон, № 33, 34.

5. **Снег как одеяло.** Почему озимые не вымерзают под снегом, картофель в яме не замерзает,
   а кусты малины пригибают к земле. *Источники:* учебник, упражнение 5 (пруд под снегом);
   Марон, № 27, 28, 30, 32.

6. **Городские «острова тепла».** Почему в центре города теплее, чем на окраине; тёмный
   асфальт и светлые крыши; роль зелени и ветра. *Источники:* учебник, «Это любопытно.
   Городские „острова тепла“» (после § 9); Марон, № 46.

7. **Теплица и парниковый эффект.** Стекло пропускает солнечное излучение и задерживает
   излучение почвы; то же делают водяной пар и углекислый газ в атмосфере. Костры и дым
   от заморозков. *Источники:* учебник, § 9; Марон, № 47, 48.

8. **Энергосбережение в быту.** Радиаторы внизу, отражающий экран за радиатором, тёплые
   полы, закрытые крышкой кастрюли, кофе с молоком. Что из этого экономит энергию и за счёт
   чего. *Источники:* учебник, § 8; Марон, № 29, 36, 44, 45.

## 3. Что должно быть в докладе

* **Явление или устройство** — о чём речь, одной фразой.
* **Вид теплопередачи**, который оно ограничивает или использует, и объяснение через
  свойства вещества: поры с воздухом, блестящая поверхность, невозможность потоков в твёрдом
  теле и так далее.
* **Одно число или сравнение.** Например, во сколько раз воздух проводит тепло хуже кирпича
  (данные из урока 10), или на сколько градусов теплее под снегом, чем над ним.
* **Предмет или картинка.** Кусок пенопласта, термос, пуховая перчатка — что угодно, что можно
  показать классу.
* **Вопрос слушателям** в конце — один, на понимание.

## 5. Числа для докладов

Данные из урока 10 можно использовать прямо в выступлении.

```python
teploprovodnost = {"медь": 400, "сталь": 50, "стекло": 1.0, "кирпич": 0.7, "вода": 0.6,
                   "дерево": 0.15, "шерсть": 0.04, "пенопласт": 0.04, "воздух": 0.026}

def vo_skolko_raz(a, b):
    return teploprovodnost[a] / teploprovodnost[b]

print(f"кирпичная стена проводит тепло хуже стальной в {vo_skolko_raz('сталь', 'кирпич'):.0f} раз")
print(f"слой пенопласта той же толщины — ещё в {vo_skolko_raz('кирпич', 'пенопласт'):.0f} раз хуже кирпича")
print(f"а сам пенопласт лишь в {vo_skolko_raz('пенопласт', 'воздух'):.1f} раза уступает воздуху, который в нём и заперт")
```

:::{admonition} Копнуть глубже: сколько тепла уходит через окно
:class: dropdown
Строители описывают окно **коэффициентом теплопередачи** $U$: это мощность, уходящая через
один квадратный метр при разности температур в один градус. Формула из старших классов
проста: $P = U \cdot S \cdot \Delta t$.

Важно, что у одинарного стекла $U$ определяется не самим стеклом, а тонкими слоями
воздуха у его поверхностей: без них потери были бы в десятки раз больше. Ориентировочные
значения: одинарное стекло около 5,8 Вт/(м²·К), двухкамерный стеклопакет около 2,8,
энергосберегающий с отражающим покрытием около 1,1.

```python
S = 2.0      # площадь окна, м²
dt = 20      # разность температур внутри и снаружи, °C
for okno, U in [("одинарное стекло", 5.8), ("двухкамерный стеклопакет", 2.8),
                ("энергосберегающий стеклопакет", 1.1)]:
    P = U * S * dt
    print(f"{okno:32s} теряет {P:5.0f} Вт — за сутки {P * 24 / 1000:5.1f} кВт·ч")
```

Разница между первым и третьим окном — больше двух киловатт-часов в сутки с одного окна.
Стекло в них почти одно и то же; вся экономия — в воздухе между стёклами и в тонком
отражающем слое, то есть в теплопроводности и излучении.
:::
