# Урок 11. Урок-конференция «Практическое использование тепловых свойств веществ и материалов в целях энергосбережения»

Урок отведён рабочей программой под конференцию: класс делится на группы, каждая готовит
короткий доклад и отвечает на вопросы. Ниже приведены план урока, темы, источники
и критерии оценки. Подготовка к конференции начинается с урока 10.

## 1. Как устроен урок

| Этап | Время | Что происходит |

Конференция посвящена одному вопросу: **какой вид теплопередачи ограничивает данный
материал или приём, и за счёт какого свойства вещества это достигается**. Каждый доклад
должен содержать явный ответ на него.

## 2. Темы докладов

Предлагается восемь тем, из которых группа выбирает одну. К каждой теме указаны опорные источники, а докладчикам следует добавить один собственный пример, найденный дома или в интернете.

1. **Утепление дома.** Стены с воздушным зазором, минеральная вата, пенопласт, двойные
   и тройные окна. Почему в утеплителях главную роль играет не вещество, а поры
   и содержащийся в них воздух.
   *Источники:* учебник, § 7 (теплопроводность газов, пористые материалы), упражнение 5, а также Марон, № 25, 32.

2. **Термос, термокружка, холодильник.** Как одно устройство перекрывает все три пути
   теплопередачи. Почему испаритель в холодильнике сверху. *Источники:* учебник, § 9 («Как устроен термос»), а также Марон, № 35.

3. **Одежда и тепло.** Пух, шерсть, мех и современные мембраны, а также вопросы о том, почему светлое носят летом и почему несколько тонких слоёв греют лучше одного толстого. *Источники:* учебник, § 7 и § 9, упражнение 7, а также Марон, № 33.

4. **Как животные сберегают тепло.** Шерсть и перья как воздушная подушка, подкожный жир,
   уменьшенные уши и лапы у северных животных, тетерева в снегу. *Источники:* учебник,
   «Это любопытно. Приспособление животных к различным температурным условиям» (после § 7), а также Марон, № 33, 34.

5. **Снег как одеяло.** Почему озимые не вымерзают под снегом, картофель в яме не замерзает,
   а кусты малины пригибают к земле. *Источники:* учебник, упражнение 5 (пруд под снегом), а также Марон, № 27, 28, 30, 32.

6. **Городские «острова тепла».** Причины того, что в центре города теплее, чем на окраине, роль тёмного асфальта и светлых крыш, зелени и ветра. *Источники:* учебник, «Это любопытно.
   Городские „острова тепла“» (после § 9), а также Марон, № 46.

7. **Теплица и парниковый эффект.** Стекло пропускает солнечное излучение и задерживает излучение почвы, и то же делают водяной пар и углекислый газ в атмосфере. Костры и дым
   от заморозков. *Источники:* учебник, § 9, а также Марон, № 47, 48.

8. **Энергосбережение в быту.** Радиаторы внизу, отражающий экран за радиатором, тёплые
   полы, закрытые крышкой кастрюли, кофе с молоком. Что из этого экономит энергию и за счёт
   чего. *Источники:* учебник, § 8, а также Марон, № 29, 36, 44, 45.

## 3. Что должно быть в докладе

* **Явление или устройство**: предмет доклада, сформулированный одной фразой.
* **Вид теплопередачи**, который оно ограничивает или использует, и объяснение через
  свойства вещества: поры с воздухом, блестящая поверхность, невозможность потоков в твёрдом
  теле и так далее.
* **Одно число или сравнение.** Например, во сколько раз воздух проводит тепло хуже кирпича
  (данные из урока 10), или на сколько градусов теплее под снегом, чем над ним.
* **Предмет или картинка.** Кусок пенопласта, термос, пуховая перчатка или любой другой
  предмет, который можно показать классу.
* **Вопрос слушателям** в конце доклада: один вопрос на понимание.

## 5. Числа для докладов

Приведённые ниже данные из урока 10 можно использовать непосредственно в выступлении.

```python
teploprovodnost = {"медь": 400, "сталь": 50, "стекло": 1.0, "кирпич": 0.7, "вода": 0.6,
                   "дерево": 0.15, "шерсть": 0.04, "пенопласт": 0.04, "воздух": 0.026}

def vo_skolko_raz(a, b):
    return teploprovodnost[a] / teploprovodnost[b]

print(f"кирпичная стена проводит тепло хуже стальной в {vo_skolko_raz('сталь', 'кирпич'):.0f} раз")
print(f"слой пенопласта той же толщины проводит тепло ещё в {vo_skolko_raz('кирпич', 'пенопласт'):.0f} раз хуже кирпича")
print(f"а сам пенопласт лишь в {vo_skolko_raz('пенопласт', 'воздух'):.1f} раза уступает воздуху, который в нём и заперт")
```

:::{admonition} Копнуть глубже: сколько тепла уходит через окно
:class: dropdown
В строительстве окно характеризуют **коэффициентом теплопередачи** $U$, который
представляет собой мощность, уходящую через один квадратный метр при разности температур
в один градус. Соответствующая формула из курса старших классов имеет простой вид:
$P = U \cdot S \cdot \Delta t$.

У одинарного стекла $U$ определяется не самим стеклом, а тонкими слоями воздуха у его поверхностей, без которых потери были бы в десятки раз больше.
Ориентировочные значения составляют: одинарное стекло около 5,8 Вт/(м²·К), однокамерный
стеклопакет около 2,8, энергосберегающий с отражающим покрытием около 1,1.

```python
S = 2.0      # площадь окна, м²
dt = 20      # разность температур внутри и снаружи, °C
for okno, U in [("одинарное стекло", 5.8), ("однокамерный стеклопакет", 2.8),
                ("энергосберегающий стеклопакет", 1.1)]:
    P = U * S * dt
    print(f"{okno:32s} теряет {P:5.0f} Вт, или {P * 24 / 1000:5.1f} кВт·ч за сутки")
```

Разница между первым и третьим окном составляет более четырёх киловатт-часов в сутки
с одного окна. Стекло в них практически одно и то же, а вся экономия обеспечивается воздухом между стёклами и тонким отражающим слоем, подавляющими теплопроводность и излучение.
:::

## 6. Критерии оценки доклада

| Критерий | 2 балла | 1 балл | 0 баллов |
|---|---|---|---|
| Назван вид теплопередачи и объяснён через свойства вещества | верно и полно | верно, но без объяснения | неверно |
| Приведено число или сравнение | есть и уместно | есть, но не связано с темой | нет |
| Есть предмет, картинка или опыт | есть | только словами | — |
| Ответы на вопросы класса | уверенные и верные | частично | — |
| Регламент 5 минут | выдержан | превышен до минуты | — |

Отметка «5» выставляется за 9–10 баллов, «4» за 7–8 баллов, «3» за 5–6 баллов.
