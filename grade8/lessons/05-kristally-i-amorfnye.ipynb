# Урок 5. Кристаллические и аморфные тела

Речь пойдёт о двух видах твёрдых тел — кристаллических и аморфных. Мы разберёмся,
что такое кристаллическая решётка, почему стекло правильнее считать застывшей
жидкостью и отчего одно и то же вещество застывает то кристаллом, то стеклом.

## 1. Два вида твёрдых тел

Молекулы твёрдого тела сидят на своих местах и колеблются — это мы уже знаем.
Но «сидят на местах» можно понимать двояко: **в строгом порядке** или **как попало**.

Оба варианта встречаются в природе, и это два разных вида твёрдого состояния.

```{admonition} Кристаллическое и аморфное состояния
:class: important
**Кристаллическое состояние** — молекулы и атомы расположены в строго определённом
порядке, образуя повторяющуюся структуру: **кристаллическую решётку**. Места,
где находятся частицы, называют **узлами** решётки.

**Аморфное состояние** — кристаллической решётки нет, расположение частиц
беспорядочное. Порядок сохраняется только среди ближайших соседей и с расстоянием
исчезает.
```

Кристаллическое строение имеют поваренная соль, лёд, алмаз, сахар, железо и
большинство металлов. В аморфном состоянии находятся опал, янтарь, смолы, воск,
обычное стекло.

## 2. Ближний и дальний порядок

Разницу удобно описать двумя словами.

* **Дальний порядок** (кристалл): зная, где стоит одна частица, можно предсказать,
  где стоят частицы за тысячи узлов от неё. Решётка повторяет сама себя.
* **Ближний порядок** (аморфное тело): у каждой частицы примерно то же число соседей
  на примерно том же расстоянии — но чуть дальше правильность теряется.

Отсюда и ровные грани у кристалла соли: он растёт, повторяя одну и ту же ячейку,
и внешняя форма выдаёт внутреннее устройство. У куска
стекла или янтаря никакой правильной формы нет — раскалывается он как попало.

## 3. Откуда берётся то или другое

Самое интересное: **одно и то же вещество** может застыть и кристаллом, и аморфным
телом. Всё решает то, **как быстро его охлаждали**.

Когда расплав остывает медленно, у молекул есть время найти удобные места
и выстроиться в решётку. Если охладить быстро, молекулы застывают там, где их
застигло, — и беспорядок расплава сохраняется в твёрдом теле.

Убедитесь сами. Сначала охладите расплав медленно — ползунок влево до упора.
Потом нажмите «Заново» и охладите быстро.

````python
from phys import sim
sim.crystallization()
````

Синим отмечены частицы, у которых правильное окружение — примерно шесть соседей,
как и положено в плотной упаковке. Оранжевым — те, у кого порядок нарушен.
При медленном охлаждении почти всё поле становится синим и появляется решётка;
при быстром остаётся много оранжевого — это и есть застывший беспорядок.

```{admonition} Так делают настоящее стекло
:class: note
Обычное оконное стекло получают именно быстрым охлаждением расплава — у молекул
кремнезёма не остаётся времени построить решётку. А если тот же расплав охлаждать
очень медленно, получится не стекло, а кристаллический кварц.

Существуют даже **металлические стёкла**: если расплавленный металл вылить на
холодный вращающийся барабан, он остынет за тысячные доли секунды и застынет
без решётки. Такие сплавы прочнее обычных металлов, и их применяют
в сердечниках трансформаторов.
```

## 4. Аморфные тела текут

У аморфных тел есть свойство, которое сближает их с жидкостями: **они текут**,
только очень медленно. Молекулы в них не закреплены в узлах, поэтому под действием
силы постепенно перемещаются.

Поэтому физики говорят об аморфном теле как об **очень густой жидкости**,
застывшая настолько, что её течение незаметно.

```{admonition} Чем ещё отличаются кристаллы и аморфные тела
:class: important
| | Кристаллическое тело | Аморфное тело |
|---|---|---|
| Расположение частиц | строгий порядок — решётка | беспорядок, порядок только ближний |
| Внешняя форма | правильные грани и углы | случайная |
| Как плавится | при строго определённой температуре | постепенно размягчается |
| Течёт ли | нет | да, очень медленно |
| Примеры | соль, лёд, алмаз, металлы | стекло, янтарь, смола, воск, пластилин |
```

Про температуру плавления мы поговорим подробно через несколько уроков.
Пока запомните разницу: лёд тает **при 0 °C** и ни при какой другой температуре,
а воск или стекло сначала становятся мягкими, потом всё более текучими —
и назвать одну температуру плавления для них нельзя.

## 6. Численный опыт: как выглядит порядок в числах

Отличить кристалл от аморфного тела можно, не глядя на картинку, — по числам.
Посмотрим на **расстояния до соседей** каждой частицы.

````python
import numpy as np
import matplotlib.pyplot as plt
from phys import style

style.use()
rng = np.random.default_rng(5)

# кристалл: узлы правильной треугольной решётки с небольшими тепловыми колебаниями
shag = 1.0
uzly = [(i * shag + (j % 2) * shag / 2, j * shag * 0.87) for i in range(14) for j in range(14)]
kristall = np.array(uzly) + rng.normal(0, 0.05, (len(uzly), 2))

# аморфное тело: то же число частиц, но расставленных беспорядочно
amorfnoe = rng.uniform(0, 12, (len(uzly), 2))

def rasstoyaniya(tochki):
    """Все расстояния между парами частиц."""
    raznost = tochki[:, None, :] - tochki[None, :, :]
    d = np.sqrt((raznost ** 2).sum(axis=2))
    return d[np.triu_indices(len(tochki), k=1)]

fig, ax = style.axes("Расстояние между частицами", "Сколько таких пар",
                     "Кристалл помнит порядок далеко, аморфное тело — нет")
ax.hist(rasstoyaniya(kristall), bins=120, range=(0, 5), alpha=0.75, label="кристалл")
ax.hist(rasstoyaniya(amorfnoe), bins=120, range=(0, 5), alpha=0.6, label="аморфное тело")
ax.legend()
plt.show()
````

У кристалла — резкие пики: расстояния между частицами принимают только определённые
значения, и так до самого края тела. Это и есть **дальний порядок**. У аморфного тела
пиков нет: расстояния всевозможные.

По такой картинке физики и определяют строение вещества: просвечивают образец
рентгеном и смотрят на пики.
