# Урок 5. Кристаллические и аморфные тела

Твёрдые тела бывают двух сортов, кристаллические и аморфные. Сегодня будет видно,
что такое кристаллическая решётка, почему стекло правильнее считать застывшей
жидкостью и отчего одно вещество застывает то кристаллом, то стеклом.

## 1. Два вида твёрдых тел

Молекулы твёрдого тела сидят на своих местах и колеблются, это уже известно.
Но «сидят на местах» можно понимать двояко. **В строгом порядке** или **как попало**.

Оба варианта встречаются в природе, и это два разных вида твёрдого состояния.

```{admonition} Кристаллическое и аморфное состояния
:class: important
**Кристаллическое состояние** — молекулы и атомы расположены в строго определённом
порядке, образуя повторяющуюся структуру: **кристаллическую решётку**. Места,
где находятся частицы, называют **узлами** решётки.

**Аморфное состояние** — кристаллической решётки нет, расположение частиц
беспорядочное. Порядок сохраняется только среди ближайших соседей и с расстоянием
исчезает.
```

Кристаллическое строение имеют поваренная соль, лёд, алмаз, сахар, железо и
большинство металлов. В аморфном состоянии находятся опал, янтарь, смолы, воск,
обычное стекло.

## 2. Ближний и дальний порядок

Разницу удобно описать двумя словами.

* **Дальний порядок** (кристалл): зная, где стоит одна частица, можно предсказать,
  где стоят частицы за тысячи узлов от неё. Решётка повторяет сама себя.
* **Ближний порядок** (аморфное тело): у каждой частицы примерно то же число соседей
  на примерно том же расстоянии, но чуть дальше правильность теряется.

Отсюда и ровные грани у кристалла соли. Он растёт, повторяя одну и ту же ячейку,
и внешняя форма выдаёт внутреннее устройство. А кусок стекла или янтаря правильной
формы не имеет и раскалывается как попало.

## 3. Откуда берётся то или другое

Теперь самое интересное. **Одно и то же вещество** застывает и кристаллом,
и аморфным телом. Решает **скорость охлаждения**.

Если расплав остывает медленно, у молекул есть время найти удобные места и выстроиться
в решётку. При быстром охлаждении молекулы застывают там, где их застигло, и беспорядок
расплава остаётся в твёрдом теле навсегда.

Убедитесь сами. Сначала охладите расплав медленно, сдвинув ползунок влево до упора.
Потом нажмите «Заново» и охладите быстро.

````python
from phys import sim
sim.crystallization()
````

Синие частицы стоят правильно, соседей у них около шести, как и положено в плотной
упаковке. Оранжевые обозначают нарушенный порядок.

Охлаждайте медленно, и поле почти сплошь синеет, проступает решётка. Охладите быстро,
и оранжевого останется много. Это и есть застывший беспорядок.

```{admonition} Так делают настоящее стекло
:class: note
Обычное оконное стекло получают именно быстрым охлаждением расплава — у молекул
кремнезёма не остаётся времени построить решётку. А если тот же расплав охлаждать
очень медленно, получится не стекло, а кристаллический кварц.

Существуют даже **металлические стёкла**: если расплавленный металл вылить на
холодный вращающийся барабан, он остынет за тысячные доли секунды и застынет
без решётки. Такие сплавы прочнее обычных металлов, и их применяют
в сердечниках трансформаторов.
```

## 4. Аморфные тела текут

Одно свойство сближает аморфные тела с жидкостями. **Они текут**, только очень
медленно. Молекулы в них не закреплены в узлах и под нагрузкой понемногу перемещаются.

Потому физики и говорят об аморфном теле как об **очень густой жидкости**, застывшей
настолько, что течение её незаметно.

```{admonition} Чем ещё отличаются кристаллы и аморфные тела
:class: important
| | Кристаллическое тело | Аморфное тело |
|---|---|---|
| Расположение частиц | строгий порядок — решётка | беспорядок, порядок только ближний |
| Внешняя форма | правильные грани и углы | случайная |
| Как плавится | при строго определённой температуре | постепенно размягчается |
| Течёт ли | нет | да, очень медленно |
| Примеры | соль, лёд, алмаз, металлы | стекло, янтарь, смола, воск, пластилин |
```

Разговор о температуре плавления впереди, через несколько уроков. Пока запомните
разницу. Лёд тает **при 0 °C** и ни при какой другой температуре. А воск или стекло
сначала размягчаются, потом текут всё легче, и одной температуры плавления
для них не назвать.

## 6. Численный опыт: как выглядит порядок в числах

Отличить кристалл от аморфного тела можно и не глядя на картинку, по одним числам.
Посмотрим на **расстояния до соседей** каждой частицы.

````python
import numpy as np
import matplotlib.pyplot as plt
from phys import style

style.use()
rng = np.random.default_rng(5)

# кристалл: узлы правильной треугольной решётки с небольшими тепловыми колебаниями
shag = 1.0
uzly = [(i * shag + (j % 2) * shag / 2, j * shag * 0.87) for i in range(14) for j in range(14)]
kristall = np.array(uzly) + rng.normal(0, 0.05, (len(uzly), 2))

# аморфное тело: то же число частиц, но расставленных беспорядочно
amorfnoe = rng.uniform(0, 12, (len(uzly), 2))

def rasstoyaniya(tochki):
    """Все расстояния между парами частиц."""
    raznost = tochki[:, None, :] - tochki[None, :, :]
    d = np.sqrt((raznost ** 2).sum(axis=2))
    return d[np.triu_indices(len(tochki), k=1)]

fig, ax = style.axes("Расстояние между частицами", "Сколько таких пар",
                     "Кристалл помнит порядок далеко, аморфное тело — нет")
ax.hist(rasstoyaniya(kristall), bins=120, range=(0, 5), alpha=0.75, label="кристалл")
ax.hist(rasstoyaniya(amorfnoe), bins=120, range=(0, 5), alpha=0.6, label="аморфное тело")
ax.legend()
plt.show()
````

У кристалла пики резкие. Расстояния между частицами принимают только определённые
значения, и так до самого края тела. Это и есть **дальний порядок**. У аморфного тела
пиков нет, расстояния всевозможные.

:::{admonition} Копнуть глубже: почему решётка получается именно такой
:class: dropdown
Молекулы при кристаллизации не выбирают расположение случайно. Они укладываются так,
чтобы занять как можно меньше места, ведь притяжение стягивает их вместе.

Задачу можно решить на плоскости, школьной геометрией. Разложим одинаковые круги
двумя способами и посчитаем, какую долю плоскости они закроют.

**Квадратная укладка.** Центры кругов в вершинах квадратов со стороной $2R$.
На каждый квадрат приходится ровно один круг: четыре четвертинки по углам. Доля
занятой площади равна

$$\frac{\pi R^2}{(2R)^2} = \frac{\pi}{4} \approx 0{,}785 .$$

**Гексагональная укладка.** Каждый следующий ряд сдвинут на полшага и лежит
в углублениях предыдущего, так что у каждого круга шесть соседей. Ячейкой служит
ромб со стороной $2R$ и углом 60°, его площадь $(2R)^2 \sin 60^\circ = 2\sqrt{3}R^2$,
и на неё тоже приходится один круг:

$$\frac{\pi R^2}{2\sqrt{3}R^2} = \frac{\pi}{2\sqrt{3}} \approx 0{,}907 .$$

````python
import numpy as np

kvadratnaya = np.pi / 4
geksagonalnaya = np.pi / (2 * np.sqrt(3))

print(f"квадратная укладка:     {100 * kvadratnaya:.1f} % площади")
print(f"гексагональная укладка: {100 * geksagonalnaya:.1f} % площади")
print(f"выигрыш в плотности:    {100 * (geksagonalnaya / kvadratnaya - 1):.1f} %")
````

Гексагональная укладка плотнее на пятнадцать процентов, и природа выбирает именно её.
Посмотрите на модель кристаллизации ещё раз: у каждой частицы там оказывается
по шесть соседей, что мы и брали за признак правильного окружения.

Кстати, шесть соседей на плоскости — причина того, что снежинки шестиугольные.
Молекулы воды в кристалле льда укладываются в шестиугольную сетку, и внешняя форма
снежинки повторяет её симметрию.

В пространстве задача решается сложнее. Иоганн Кеплер предположил в 1611 году,
что плотнее всего шары укладываются как апельсины на прилавке, занимая 74 % объёма.
Строгое доказательство этой гипотезы получили лишь в 1998 году, и оно потребовало
перебора на компьютере.
:::

По такой картинке физики и определяют строение вещества: просвечивают образец
рентгеном и смотрят на пики.
