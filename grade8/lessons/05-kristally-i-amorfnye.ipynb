# Урок 5. Кристаллические и аморфные тела

Твёрдые тела делятся на два вида: кристаллические и аморфные. В настоящем уроке
выясняется, что представляет собой кристаллическая решётка, почему стекло правильнее
считать застывшей жидкостью и отчего одно и то же вещество застывает то кристаллом,
то стеклом.

## 1. Два вида твёрдых тел

Ранее было установлено, что молекулы твёрдого тела находятся на своих местах
и колеблются около них. При этом выражение «находятся на своих местах» допускает
два понимания: частицы могут располагаться **в строгом порядке** или **беспорядочно**.

Оба варианта встречаются в природе и соответствуют двум различным видам твёрдого
состояния.

```{admonition} Кристаллическое и аморфное состояния
:class: important
В **кристаллическом состоянии** молекулы и атомы расположены в строго определённом
порядке, образуя повторяющуюся структуру: **кристаллическую решётку**. Места,
где находятся частицы, называют **узлами** решётки.

В **аморфном состоянии** кристаллическая решётка отсутствует, расположение частиц
беспорядочное. Порядок сохраняется только среди ближайших соседей и с расстоянием
исчезает.
```

Кристаллическое строение имеют поваренная соль, лёд, алмаз, сахар, железо и
большинство металлов. В аморфном состоянии находятся опал, янтарь, смолы, воск,
обычное стекло.

## 2. Ближний и дальний порядок

Различие между двумя видами удобно описывается двумя терминами.

* **Дальний порядок** (кристалл): зная положение одной частицы, можно предсказать,
  где находятся частицы за тысячи узлов от неё, поскольку решётка повторяет саму себя.
* **Ближний порядок** (аморфное тело): каждая частица имеет примерно одинаковое число
  соседей на примерно одинаковом расстоянии, однако на несколько большем расстоянии
  правильность расположения теряется.

Этим объясняются ровные грани кристалла соли. Кристалл растёт, повторяя одну и ту же
ячейку, и внешняя форма отражает внутреннее устройство. Кусок стекла или янтаря,
напротив, правильной формы не имеет и раскалывается произвольным образом.

## 3. Откуда берётся то или другое

**Одно и то же вещество** может застывать как кристаллом, так и аморфным телом.
Результат определяется **скоростью охлаждения**.

Если расплав остывает медленно, молекулы успевают найти удобные места и выстроиться
в решётку. При быстром охлаждении молекулы застывают там, где их застало охлаждение,
и беспорядок расплава сохраняется в твёрдом теле.

Данный вывод можно проверить на модели. Сначала следует охладить расплав медленно,
сдвинув ползунок влево до упора, затем нажать «Заново» и охладить его быстро.


````python
from phys import sim
sim.crystallization()
````

Синие частицы расположены правильно: число соседей у них около шести, как и должно
быть в плотной упаковке. Оранжевые частицы обозначают нарушенный порядок.

При медленном охлаждении поле почти полностью становится синим и проступает решётка.
При быстром охлаждении оранжевых частиц остаётся много, то есть беспорядок расплава
застывает.

```{admonition} Так получают настоящее стекло
:class: note
Обычное оконное стекло получают охлаждением расплава, при котором у молекул
кремнезёма не остаётся времени построить решётку. Если же тот же расплав охлаждать
очень медленно, получится не стекло, а кристалл.

Существуют также **металлические стёкла**: если расплавленный металл вылить на
холодный вращающийся барабан, он остынет за тысячные доли секунды и застынет
без решётки. Такие сплавы очень прочны и легко перемагничиваются, поэтому из них
изготавливают сердечники трансформаторов.
```

## 4. Аморфные тела текут

Одно свойство сближает аморфные тела с жидкостями: **они текут**, хотя и очень
медленно. Молекулы в них не закреплены в узлах и под нагрузкой понемногу перемещаются.

По этой причине аморфное тело рассматривают как **очень густую жидкость**, застывшую
настолько, что её течение незаметно.

```{admonition} Чем ещё отличаются кристаллы и аморфные тела
:class: important
| | Кристаллическое тело | Аморфное тело |
|---|---|---|
| Расположение частиц | строгий порядок (решётка) | беспорядок, порядок только ближний |
| Внешняя форма | правильные грани и углы | случайная |
| Как плавится | при строго определённой температуре | постепенно размягчается |
| Течёт ли | нет | да, очень медленно |
| Примеры | соль, лёд, алмаз, металлы | стекло, янтарь, смола, воск, пластилин |
```

Температура плавления подробно рассматривается через несколько уроков. Пока
достаточно запомнить различие: лёд тает **при 0 °C** и ни при какой другой
температуре, тогда как воск или стекло сначала размягчаются, затем текут всё легче,
и единой температуры плавления для них указать нельзя.

## 6. Численный опыт: как выглядит порядок в числах

Отличить кристалл от аморфного тела можно и без изображения, по одним лишь числам.
Рассмотрим **расстояния до соседей** для каждой частицы.


````python
import numpy as np
import matplotlib.pyplot as plt
from phys import style

style.use()
rng = np.random.default_rng(5)

# кристалл: узлы правильной треугольной решётки с небольшими тепловыми колебаниями
shag = 1.0
uzly = [(i * shag + (j % 2) * shag / 2, j * shag * 0.87) for i in range(14) for j in range(14)]
kristall = np.array(uzly) + rng.normal(0, 0.05, (len(uzly), 2))

# аморфное тело: то же число частиц, но расставленных беспорядочно
amorfnoe = rng.uniform(0, 12, (len(uzly), 2))

def rasstoyaniya(tochki):
    """Все расстояния между парами частиц."""
    raznost = tochki[:, None, :] - tochki[None, :, :]
    d = np.sqrt((raznost ** 2).sum(axis=2))
    return d[np.triu_indices(len(tochki), k=1)]

fig, ax = style.axes("Расстояние между частицами", "Сколько таких пар",
                     "Кристалл сохраняет порядок на большом расстоянии, аморфное тело не сохраняет")
ax.hist(rasstoyaniya(kristall), bins=120, range=(0, 5), alpha=0.75, label="кристалл")
ax.hist(rasstoyaniya(amorfnoe), bins=120, range=(0, 5), alpha=0.6, label="аморфное тело")
ax.legend()
plt.show()
````

У кристалла пики резкие. Расстояния между частицами принимают только определённые
значения, и так вплоть до края тела, что и называется **дальним порядком**. У аморфного
тела пиков нет, расстояния принимают всевозможные значения.

Именно по такой картине определяют строение вещества: образец просвечивают
рентгеновским излучением и анализируют положение пиков.

:::{admonition} Копнуть глубже: почему решётка получается именно такой
:class: dropdown
Молекулы при кристаллизации не выбирают расположение случайно. Они укладываются так,
чтобы занять как можно меньше места, поскольку притяжение стягивает их вместе.

Данную задачу можно решить на плоскости средствами школьной геометрии. Разложим
одинаковые круги двумя способами и подсчитаем, какую долю плоскости они закроют.

**Квадратная укладка.** Центры кругов находятся в вершинах квадратов со стороной $2R$.
На каждый квадрат приходится ровно один круг: четыре четверти по углам. Доля
занятой площади равна

$$\frac{\pi R^2}{(2R)^2} = \frac{\pi}{4} \approx 0{,}785 .$$

**Гексагональная укладка.** Каждый следующий ряд сдвинут на полшага и лежит
в углублениях предыдущего, так что у каждого круга шесть соседей. Ячейкой служит
ромб со стороной $2R$ и углом 60°, его площадь $(2R)^2 \sin 60^\circ = 2\sqrt{3}R^2$,
и на неё также приходится один круг:

$$\frac{\pi R^2}{2\sqrt{3}R^2} = \frac{\pi}{2\sqrt{3}} \approx 0{,}907 .$$


````python
import numpy as np

kvadratnaya = np.pi / 4
geksagonalnaya = np.pi / (2 * np.sqrt(3))

print(f"квадратная укладка:     {100 * kvadratnaya:.1f} % площади")
print(f"гексагональная укладка: {100 * geksagonalnaya:.1f} % площади")
print(f"выигрыш в плотности:    {100 * (geksagonalnaya / kvadratnaya - 1):.1f} %")
````

Гексагональная укладка плотнее на пятнадцать процентов, и природа выбирает именно её.
В модели кристаллизации у каждой частицы оказывается по шесть соседей, что и было
принято за признак правильного окружения.

Следует отметить, что шесть соседей на плоскости служат причиной шестиугольной формы
снежинок. Молекулы воды в кристалле льда укладываются в шестиугольную сетку, и внешняя
форма снежинки повторяет её симметрию.

В пространстве задача решается сложнее. Иоганн Кеплер предположил в 1611 году,
что плотнее всего шары укладываются как апельсины на прилавке, занимая 74 % объёма.
Строгое доказательство этой гипотезы было получено лишь в 1998 году, и оно потребовало
перебора на компьютере.
