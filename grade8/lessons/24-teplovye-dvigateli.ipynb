## 1. Работа газа и пара при расширении

Опыт из учебника (§ 23). В пробирку наливают воду, закрывают пробкой и нагревают. Вода
закипает, давление пара растёт, и пробка вылетает: пар совершил работу, приподняв пробку
и толкнув её. Внутренняя энергия пара при этом уменьшилась — пар охладился.

Так работает любой тепловой двигатель: нагретый газ расширяется и толкает что-нибудь.

```{admonition} Тепловой двигатель
:class: important
**Тепловым двигателем** называют машину, в которой внутренняя энергия топлива превращается
в механическую энергию.

Топливо сгорает — нагревает газ (**рабочее тело**) — газ расширяется и совершает работу.
Затем отработавший газ выбрасывают и цикл повторяется.
```

Все тепловые двигатели — паровая машина, паровая турбина, двигатель внутреннего сгорания,
реактивный двигатель — устроены по этой схеме. Различаются лишь способы нагреть газ и снять
с него работу.

## 2. Двигатель внутреннего сгорания

```{admonition} Определение
:class: important
Двигатель, в котором топливо сжигается **внутри цилиндра**, а образовавшийся газ расширяется
там же, называют **двигателем внутреннего сгорания**.
```

Устройство простейшего одноцилиндрового двигателя (рис. 48 учебника): цилиндр, в нём
поршень, соединённый шатуном с коленчатым валом. В верхней части цилиндра два клапана —
впускной и выпускной — и свеча зажигания. Работа идёт в четыре такта, каждый — один ход
поршня, то есть полоборота вала.

````python
from phys import sim
sim.engine()
````

1. **Впуск.** Поршень идёт вниз, давление в цилиндре падает, через открытый впускной клапан
   входит горючая смесь бензина с воздухом (в дизеле — чистый воздух).
2. **Сжатие.** Оба клапана закрыты, поршень идёт вверх и сжимает смесь в 6–7 раз. Смесь
   при этом нагревается — вспомните урок 9.
3. **Рабочий ход.** В верхней точке искра поджигает смесь. Газы, нагретые до 1500–2000 °C,
   расширяются и толкают поршень вниз. Это единственный такт, в котором совершается работа.
4. **Выпуск.** Открывается выпускной клапан, поршень идёт вверх и выталкивает отработавшие
   газы.

Один рабочий цикл — четыре хода поршня и два оборота вала. Три такта из четырёх поршень
движется по инерции — за счёт тяжёлого маховика на валу или за счёт других цилиндров,
у которых в этот момент рабочий ход. Поэтому в автомобильных двигателях цилиндров четыре
и больше, а маховик у многоцилиндрового двигателя меньше (Марон, № 280).

```{admonition} Дизель
:class: note
В дизельном двигателе искры нет. Сжимается чистый воздух, и сжимается сильно — в 17 раз;
при этом он нагревается до 600–700 °C. В конце сжатия в цилиндр впрыскивают топливо,
и оно вспыхивает само. Отсюда ответ на задачу Марона № 292: бензиновую смесь нельзя сжимать
так сильно — она воспламенится раньше искры, не вовремя.
```

## 3. Паровая турбина

В турбине нет поршня. Пар из котла под большим давлением вырывается из сопел и бьёт
в лопатки, закреплённые на диске. Диск вращается — и вместе с ним вал. Пар расширяется,
охлаждается и уходит в конденсатор, где превращается в воду и возвращается в котёл.

Турбина использует **кинетическую** энергию струи пара (Марон, № 283). Она равномерно
вращается, не имеет возвратно-поступательных частей, и потому её можно делать очень
большой и мощной. Именно паровые турбины стоят на тепловых и атомных электростанциях:
там всё равно, чем греть воду — углём, газом или ядерным топливом; дальше работает
одна и та же турбина.

## 4. Из истории

Первую паровую машину для заводских нужд построил в 1765 году **Иван Иванович Ползунов**
на Барнаульском заводе. Усовершенствовал паровую машину **Джеймс Уатт**: его двигатель
1780-х годов стал двигателем промышленной революции, а его имя носит единица мощности.
Двигатель внутреннего сгорания появился век спустя, и к концу XIX века на нём поехал
первый автомобиль.

За две тысячи лет до этого Герон Александрийский описал шар, вращающийся под действием
вырывающегося из него пара, — первую реактивную турбину. Но тогда она осталась игрушкой:
не было ни нужды в двигателях, ни металлов, способных выдержать давление.

## 5. Численный опыт: сколько бензина ест автомобиль

Проверим, много ли энергии в бензине, на понятной задаче: сколько топлива нужно легковому
автомобилю на 100 км.

````python
F = 550            # сила сопротивления движению на ровной дороге, Н
put = 100e3        # путь, м
kpd = 0.25         # КПД бензинового двигателя
q = 4.6e7          # удельная теплота сгорания бензина, Дж/кг
rho = 720          # плотность бензина, кг/м³

A = F * put                       # полезная работа
Q = A / kpd                       # энергия, которую пришлось получить от топлива
m = Q / q
print(f"полезная работа {A / 1e6:.0f} МДж, энергия топлива {Q / 1e6:.0f} МДж")
print(f"бензина: {m:.1f} кг = {m / rho * 1000:.1f} л на 100 км")
````

Шесть с половиной литров на сотню — как у настоящей машины. Из них три четверти улетают в выхлопную
трубу и уходят в радиатор охлаждения. О том, куда уходит энергия и как считать КПД, —
следующий урок.
