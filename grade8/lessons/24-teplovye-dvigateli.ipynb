## 1. Работа газа и пара при расширении

Рассмотрим опыт из учебника (§ 23). В пробирку наливают воду, закрывают пробкой и нагревают.
Вода закипает, давление пара растёт, и пробка вылетает: пар совершил работу, приподняв
пробку и толкнув её. Внутренняя энергия пара при этом уменьшилась, пар охладился.

Аналогичным образом работает любой тепловой двигатель: нагретый газ расширяется и толкает
некоторое тело.

```{admonition} Тепловой двигатель
:class: important
**Тепловым двигателем** называют машину, в которой внутренняя энергия топлива превращается
в механическую энергию.

Топливо сгорает и нагревает газ (**рабочее тело**), газ расширяется и совершает работу.
Затем отработавший газ выбрасывают, и цикл повторяется.
```

Все тепловые двигатели (паровая машина, паровая турбина, двигатель внутреннего сгорания,
реактивный двигатель) устроены по данной схеме. Различаются лишь способы нагрева газа
и получения от него работы.

## 2. Двигатель внутреннего сгорания

```{admonition} Определение
:class: important
Двигатель, в котором топливо сжигается **внутри цилиндра**, а образовавшийся газ расширяется
там же, называют **двигателем внутреннего сгорания**.
```

Устройство простейшего одноцилиндрового двигателя (рис. 48 учебника) следующее: цилиндр,
в нём поршень, соединённый шатуном с коленчатым валом. В верхней части цилиндра расположены
два клапана, впускной и выпускной, и свеча зажигания. Работа идёт в четыре такта, каждый из которых представляет собой один ход поршня (пол-оборота вала).

````python
from phys import sim
sim.engine()
````

1. **Впуск.** Поршень идёт вниз, давление в цилиндре падает, через открытый впускной клапан
   входит горючая смесь бензина с воздухом (в дизеле чистый воздух).
2. **Сжатие.** Оба клапана закрыты, поршень идёт вверх и сжимает смесь в 6–7 раз. Смесь
   при этом нагревается, как было установлено в уроке 9.
3. **Рабочий ход.** В верхней точке искра поджигает смесь. Газы, нагретые до 1500–2000 °C,
   расширяются и толкают поршень вниз. Это единственный такт, в котором совершается работа.
4. **Выпуск.** Открывается выпускной клапан, поршень идёт вверх и выталкивает отработавшие
   газы.

Один рабочий цикл включает четыре хода поршня и два оборота вала. Три такта из четырёх
поршень движется по инерции, за счёт тяжёлого маховика на валу или за счёт других
цилиндров, в которых в этот момент происходит рабочий ход. Поэтому в автомобильных
двигателях цилиндров четыре и больше, а маховик у многоцилиндрового двигателя меньше
(Марон, № 280).

```{admonition} Дизель
:class: note
В дизельном двигателе искра отсутствует. Сжимается чистый воздух, причём сильно, в 17 раз, и при этом он нагревается до 600–700 °C. В конце сжатия в цилиндр впрыскивают
топливо, и оно вспыхивает самопроизвольно. Отсюда следует ответ на задачу Марона № 292:
бензиновую смесь нельзя сжимать так сильно, она воспламенится раньше искры, не вовремя.
```

## 3. Паровая турбина

В турбине поршень отсутствует. Пар из котла под большим давлением вырывается из сопел
и ударяет в лопатки, закреплённые на диске. Диск вращается, а вместе с ним и вал. Пар
расширяется, охлаждается и уходит в конденсатор, где превращается в воду и возвращается
в котёл.

Турбина использует **кинетическую** энергию струи пара (Марон, № 283). Она равномерно
вращается, не имеет возвратно-поступательных частей, и потому её можно делать очень
большой и мощной. Паровые турбины установлены на тепловых и атомных электростанциях:
способ нагрева воды (углём, газом или ядерным топливом) при этом не имеет значения, далее работает одна и та же турбина.

## 4. Из истории

Первую паровую машину для заводских нужд построил в 1765 году **Иван Иванович Ползунов**
на Барнаульском заводе. Паровую машину усовершенствовал **Джеймс Уатт**: его двигатель
1780-х годов стал двигателем промышленной революции, а его имя носит единица мощности.
Двигатель внутреннего сгорания появился век спустя, и к концу XIX века на нём поехал
первый автомобиль.

За две тысячи лет до этого Герон Александрийский описал шар, вращающийся под действием вырывающегося из него пара, первую реактивную турбину. В то время она осталась игрушкой из-за отсутствия как потребности в двигателях, так и металлов, способных выдержать давление.

## 5. Численный опыт: сколько бензина ест автомобиль

Проверим, много ли энергии содержится в бензине, на наглядной задаче: определим, сколько
топлива требуется легковому автомобилю на 100 км.

````python
F = 550            # сила сопротивления движению на ровной дороге, Н
put = 100e3        # путь, м
kpd = 0.25         # КПД бензинового двигателя
q = 4.6e7          # удельная теплота сгорания бензина, Дж/кг
rho = 720          # плотность бензина, кг/м³

A = F * put                       # полезная работа
Q = A / kpd                       # энергия, которую пришлось получить от топлива
m = Q / q
print(f"полезная работа {A / 1e6:.0f} МДж, энергия топлива {Q / 1e6:.0f} МДж")
print(f"бензина: {m:.1f} кг = {m / rho * 1000:.1f} л на 100 км")
````

Получается шесть с лишним литров на сотню километров, как у настоящей машины. Из них три
четверти уходят в выхлопную трубу и в радиатор охлаждения. Вопрос о том, куда уходит
энергия и как рассчитывать КПД, рассматривается на следующем уроке.
