# Урок 10. Виды теплопередачи

Внутреннюю энергию тела можно изменить теплопередачей. Существует три вида теплопередачи:
**теплопроводность**, **конвекция** и **излучение**. Они различаются тем, как именно
переносится энергия, и потому работают в разных условиях: одни только в веществе, другой —
и через пустоту.

## 1. Теплопроводность

К металлической спице воском прикрепляют несколько гвоздиков и нагревают один её конец.
Гвоздики падают по очереди: сначала ближние к пламени, потом дальние. Спица разогревается
постепенно, и чем дальше её участок от пламени, тем больше времени ему требуется.

Механизм таков. Частицы металла у пламени начинают колебаться сильнее. Взаимодействуя
с соседними частицами, они передают им часть энергии, те — следующим, и так по всей длине.
Сами частицы остаются на своих местах.

```{admonition} Теплопроводность
:class: important
**Теплопроводность** — процесс, при котором энергия передаётся от одного тела к другому
или от одной части тела к другой за счёт теплового движения частиц и их взаимодействия
между собой.

При теплопроводности происходит только передача энергии; **переноса вещества
не происходит**.
```

В модели два стержня из атомов, связанных упругими связями. Левые концы греются, правые
охлаждаются. В верхнем стержне связи жёсткие, как в металле, в нижнем слабые.

холодным почти по всей длине. Атомы при этом никуда не перемещаются — переносится только
энергия их колебаний.

### Разные вещества проводят тепло по-разному

Опыт учебника: два одинаковых стержня, медный и стальной, нагревают с одного конца.
Гвоздики на медном падают заметно раньше. Медь проводит тепло лучше стали, а металлы
в целом лучше всех других веществ.

Жидкости проводят тепло плохо. Если нагревать пробирку с водой сверху, вода у поверхности
закипит, а кусочек льда на дне, придавленный грузиком, не растает. Ещё хуже проводят газы:
поднесите палец сбоку к пламени на расстоянии сантиметра, и тепла вы почти не почувствуете,
хотя между пламенем и пальцем всего лишь воздух. Хуже всего проводит тепло сильно
разреженный газ, то есть вакуум: там почти нечему передавать колебания.

Плохую теплопроводность газов человек использует. Шерсть, пух, мех, войлок, пористые
строительные материалы греют не сами по себе — они удерживают в порах воздух, а воздух
и есть настоящий теплоизолятор. Снег тоже пористый и защищает почву и озимые от промерзания.

````python
import numpy as np
import matplotlib.pyplot as plt
from phys import style

style.use()
# коэффициенты теплопроводности при комнатной температуре, Вт/(м·К); справочные округлённые значения
veshchestva = {"медь": 400, "алюминий": 220, "сталь": 50, "лёд": 2.2, "стекло": 1.0,
               "кирпич": 0.7, "вода": 0.6, "дерево (сосна)": 0.15, "шерсть": 0.04,
               "пенопласт": 0.04, "воздух": 0.026}
fig, ax = style.axes("Теплопроводность, Вт/(м·К), логарифмическая шкала", "",
                     "Во сколько раз одни вещества проводят тепло лучше других", figsize=(7.5, 4.6))
names = list(veshchestva)[::-1]
ax.barh(names, [veshchestva[n] for n in names])
ax.set_xscale("log")
plt.show()
print(f"медь проводит тепло лучше воздуха в {veshchestva['медь'] / veshchestva['воздух']:.0f} раз")
print(f"дерево проводит лучше воздуха всего в {veshchestva['дерево (сосна)'] / veshchestva['воздух']:.0f} раз")
print(f"шерсть и пенопласт — почти как воздух: в них его и держат")
````

Шкала на графике логарифмическая — каждая клетка вправо означает рост в десять раз.
Между медью и воздухом четыре таких клетки, то есть более десяти тысяч раз.

## 2. Конвекция

Если жидкости и газы так плохо проводят тепло, почему быстро нагревается вода в кастрюле
и почему тёплый воздух от радиатора доходит до потолка? Здесь работает другой вид
теплопередачи.

Опыт учебника: в колбу с водой опускают кристаллик красителя и нагревают снизу. Окрашенная
вода поднимается вверх от дна, у поверхности расходится в стороны и по стенкам опускается
вниз. Нагретая снизу вода расширяется, становится легче окружающей и всплывает; на её
место опускается холодная, более плотная.

```{admonition} Конвекция
:class: important
**Конвекция** — вид теплопередачи, при котором энергия переносится потоками жидкости
или газа. В отличие от теплопроводности при конвекции **происходит перенос вещества**.

В твёрдых телах конвекция невозможна: их частицы лишь колеблются около положений
равновесия, и потоку вещества взяться неоткуда.
```

Конвекцию, возникающую сама собой при нагревании снизу, называют **естественной**.
Если же жидкость или газ перемешивают насосом или вентилятором, конвекция
**вынужденная** — так работает система охлаждения двигателя и обычный фен.

На естественной конвекции держится многое. Радиаторы отопления ставят внизу комнаты,
чтобы тёплый воздух поднимался и обходил всё помещение. Испаритель холодильника,
наоборот, помещают вверху: холодный воздух опускается и охлаждает камеру. Тяга в печной
трубе — тоже конвекция: горячие газы поднимаются, а на их место снизу поступает свежий
воздух, необходимый горению.

## 3. Излучение

Как энергия Солнца доходит до Земли? Пространство между ними практически пусто, значит,
ни теплопроводность, ни конвекция здесь не работают. Энергия передаётся **излучением**.

Все нагретые тела испускают излучение, и тем сильнее, чем выше температура. Испуская его,
тело теряет внутреннюю энергию; поглощая — приобретает. Излучение — это электромагнитные
волны, о которых пойдёт речь в 9 классе; передачу энергии этим путём тоже называют
излучением.

Опыт учебника с **теплоприёмником**. Это плоская металлическая коробочка, одна сторона
которой чёрная, другая блестящая. Коробочку соединяют с манометром и подносят сбоку
раскалённый кусок металла. Воздух в коробочке нагревается, и манометр это показывает.
Теплопроводность воздуха здесь ни при чём — она ничтожна; конвекция тоже, ведь коробочка
не над металлом, а сбоку. Остаётся излучение.

Повернём к металлу сначала чёрную сторону, потом блестящую. С чёрной стороной манометр
реагирует заметно сильнее.

```{admonition} Поглощение и испускание излучения
:class: important
Тела с тёмной поверхностью поглощают излучение лучше, чем тела со светлой или блестящей
поверхностью. Тёмная поверхность и испускает излучение сильнее.
```

Отсюда светлая одежда летом и белые стены домов на юге, тёмные крыши солнечных
нагревателей и зеркальные стенки термоса.

```{admonition} Как устроен термос
:class: note
Термос — пример сложного теплообмена, где перекрыты все три пути. Колба у него
с двойными стенками; из пространства между ними выкачан воздух, что почти исключает
теплопроводность и конвекцию. Стенки покрыты блестящим слоем, который отражает излучение.
Остаётся пробка — через неё и уходит тепло, поэтому термос дольше держит температуру,
когда стоит вертикально.
```

Тот же принцип у теплицы. Стекло пропускает солнечное излучение, почва под ним нагревается,
а излучение нагретой почвы стекло пропускает плохо — энергия оказывается в ловушке. Так
же действуют водяной пар и углекислый газ в атмосфере Земли; это явление называют
парниковым эффектом.

```{admonition} Три вида теплопередачи
:class: important
| | Теплопроводность | Конвекция | Излучение |
|---|---|---|---|
| Как передаётся энергия | от частицы к частице при их взаимодействии | потоками жидкости или газа | электромагнитными волнами |
| Переносится ли вещество | нет | да | нет |
| Где возможна | в любом веществе, лучше всего в металлах | только в жидкостях и газах | в любой среде и в пустоте |
| Пример | нагрев ложки в чае | тяга в трубе, отопление комнаты | тепло Солнца, тепло костра |
```
