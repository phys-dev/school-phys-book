# Урок 10. Виды теплопередачи

Внутреннюю энергию тела можно изменить теплопередачей. Существует три вида теплопередачи:
**теплопроводность**, **конвекция** и **излучение**. Они различаются механизмом переноса
энергии и потому реализуются в различных условиях: первые два только в веществе, тогда как
третий возможен и в пустоте.

## 1. Теплопроводность

К металлической спице воском прикрепляют несколько гвоздиков и нагревают один её конец.
Гвоздики падают по очереди: сначала ближние к пламени, затем дальние. Спица разогревается
постепенно, и чем дальше её участок расположен от пламени, тем большее время требуется
для его нагревания.

Механизм данного процесса состоит в следующем. Частицы металла вблизи пламени начинают
колебаться сильнее. Взаимодействуя с соседними частицами, они передают им часть энергии,
те, в свою очередь, передают её следующим, и так по всей длине спицы. Сами частицы при
этом остаются на своих местах.

```{admonition} Теплопроводность
:class: important
**Теплопроводность** представляет собой процесс, при котором энергия передаётся от одного
тела к другому или от одной части тела к другой за счёт теплового движения частиц и их
взаимодействия между собой.

При теплопроводности происходит только передача энергии, **переноса вещества не происходит**.
```

В модели рассматриваются два стержня из атомов, связанных упругими связями. Левые концы
нагреваются, правые охлаждаются. В верхнем стержне связи жёсткие, как в металле, в нижнем
слабые.

````python
from phys import sim
sim.conduction()
````

Нагрев распространяется по верхнему стержню и достигает дальнего конца, тогда как нижний
стержень остаётся холодным почти по всей длине. Атомы при этом не перемещаются, переносится
только энергия их колебаний.

### Разные вещества проводят тепло по-разному

В опыте учебника два одинаковых стержня, медный и стальной, нагревают с одного конца.
Гвоздики на медном стержне падают заметно раньше. Следовательно, медь проводит тепло лучше
стали, а металлы в целом лучше всех других веществ.

Жидкости проводят тепло плохо. Если нагревать пробирку с водой сверху, вода у поверхности
закипает, а кусочек льда на дне, придавленный грузиком, не тает. Ещё хуже проводят тепло
газы: если поднести палец сбоку к пламени на расстоянии сантиметра, тепло почти
не ощущается, хотя между пламенем и пальцем находится лишь воздух. Хуже всего проводит тепло сильно разреженный газ (вакуум), в котором почти нет частиц, способных передавать колебания.

Плохая теплопроводность газов широко используется человеком. Шерсть, пух, мех, войлок,
пористые строительные материалы греют не сами по себе: они удерживают в порах воздух,
который и является настоящим теплоизолятором. Снег также пористый и защищает почву
и озимые от промерзания.

````python
import numpy as np
import matplotlib.pyplot as plt
from phys import style

style.use()
# коэффициенты теплопроводности при комнатной температуре, Вт/(м·К); справочные округлённые значения
veshchestva = {"медь": 400, "алюминий": 220, "сталь": 50, "лёд": 2.2, "стекло": 1.0,
               "кирпич": 0.7, "вода": 0.6, "дерево (сосна)": 0.15, "шерсть": 0.04,
               "пенопласт": 0.04, "воздух": 0.026}
fig, ax = style.axes("Теплопроводность, Вт/(м·К), логарифмическая шкала", "",
                     "Во сколько раз одни вещества проводят тепло лучше других", figsize=(7.5, 4.6))
names = list(veshchestva)[::-1]
ax.barh(names, [veshchestva[n] for n in names])
ax.set_xscale("log")
plt.show()
print(f"медь проводит тепло лучше воздуха в {veshchestva['медь'] / veshchestva['воздух']:.0f} раз")
print(f"дерево проводит лучше воздуха всего в {veshchestva['дерево (сосна)'] / veshchestva['воздух']:.0f} раз")
print(f"шерсть и пенопласт проводят тепло почти как воздух, который в них и удерживается")
````

Шкала на графике логарифмическая: каждая клетка вправо соответствует росту в десять раз.
Между медью и воздухом четыре таких клетки, различие составляет более десяти тысяч раз.

## 2. Конвекция

Поскольку жидкости и газы плохо проводят тепло, возникает вопрос, почему вода в кастрюле
нагревается быстро, а тёплый воздух от радиатора достигает потолка. В данных случаях
действует другой вид теплопередачи.

В опыте учебника в колбу с водой опускают кристаллик красителя и нагревают её снизу.
Окрашенная вода поднимается вверх от дна, у поверхности расходится в стороны и по стенкам
опускается вниз. Нагретая снизу вода расширяется, становится легче окружающей и всплывает, а на её место опускается холодная, более плотная вода.

```{admonition} Конвекция
:class: important
**Конвекция** представляет собой вид теплопередачи, при котором энергия переносится потоками
жидкости или газа. В отличие от теплопроводности при конвекции **происходит перенос
вещества**.

В твёрдых телах конвекция невозможна, так как их частицы лишь колеблются около положений равновесия и поток вещества образоваться не может.
```

Конвекцию, возникающую самопроизвольно при нагревании снизу, называют **естественной**.
Если же жидкость или газ перемешивают насосом или вентилятором, конвекцию называют
**вынужденной**: так работают система охлаждения двигателя и обычный фен.

На естественной конвекции основано действие многих устройств. Радиаторы отопления
располагают внизу комнаты, чтобы тёплый воздух поднимался и обходил всё помещение.
Испаритель холодильника, напротив, помещают вверху: холодный воздух опускается
и охлаждает камеру. Тяга в печной трубе также представляет собой конвекцию: горячие газы
поднимаются, а на их место снизу поступает свежий воздух, необходимый для горения.

## 3. Излучение

Рассмотрим, каким образом энергия Солнца достигает Земли. Пространство между ними
практически пусто, следовательно, ни теплопроводность, ни конвекция здесь невозможны.
Энергия передаётся **излучением**.

Все нагретые тела испускают излучение, причём тем сильнее, чем выше их температура.
Испуская излучение, тело теряет внутреннюю энергию, поглощая его, приобретает. Излучение
представляет собой электромагнитные волны, которые рассматриваются в 9 классе, и передачу энергии данным путём также называют излучением.

Опыт учебника проводится с **теплоприёмником**. Он представляет собой плоскую металлическую
коробочку, одна сторона которой чёрная, другая блестящая. Коробочку соединяют с манометром
и подносят сбоку раскалённый кусок металла. Воздух в коробочке нагревается, что
регистрируется манометром. Ничтожная теплопроводность воздуха здесь не играет роли, а конвекция исключена расположением коробочки не над металлом, а сбоку. Следовательно, энергия передаётся излучением.

Повернём к металлу сначала чёрную сторону, затем блестящую. При чёрной стороне манометр
реагирует заметно сильнее.

```{admonition} Поглощение и испускание излучения
:class: important
Тела с тёмной поверхностью поглощают излучение лучше, чем тела со светлой или блестящей
поверхностью. Тёмная поверхность и испускает излучение сильнее.
```

Этим объясняются светлая одежда летом и белые стены домов на юге, тёмные крыши солнечных
нагревателей и зеркальные стенки термоса.

```{admonition} Как устроен термос
:class: note
Термос служит примером сложного теплообмена, в котором перекрыты все три пути передачи
энергии. Колба термоса имеет двойные стенки, из пространства между которыми выкачан воздух, что почти исключает теплопроводность и конвекцию. Стенки покрыты блестящим слоем, который
отражает излучение. Остаётся пробка, через которую и уходит тепло, поэтому термос дольше
сохраняет температуру, когда стоит вертикально.
```

На том же принципе основана теплица. Стекло пропускает солнечное излучение, почва под ним
нагревается, а излучение нагретой почвы стекло пропускает плохо, и энергия оказывается
в ловушке. Аналогично действуют водяной пар и углекислый газ в атмосфере Земли, и данное явление называют парниковым эффектом.

```{admonition} Три вида теплопередачи
:class: important
| | Теплопроводность | Конвекция | Излучение |
|---|---|---|---|
| Как передаётся энергия | от частицы к частице при их взаимодействии | потоками жидкости или газа | электромагнитными волнами |
| Переносится ли вещество | нет | да | нет |
| Где возможна | в любом веществе, лучше всего в металлах | только в жидкостях и газах | в любой среде и в пустоте |
| Пример | нагрев ложки в чае | тяга в трубе, отопление комнаты | тепло Солнца, тепло костра |
```
