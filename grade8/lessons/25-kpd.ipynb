# Урок 25. КПД теплового двигателя. Тепловые двигатели и защита окружающей среды

## 1. Куда уходит энергия топлива

Двигатель автомобиля сжигает бензин, однако лишь четверть его энергии расходуется
на движение машины. Остальная часть нагревает радиатор и уходит из выхлопной трубы
в виде горячего газа. В настоящем уроке выясняется, почему иначе быть не может.

В любом тепловом двигателе газ, совершив работу, остаётся горячим, то есть более нагретым,
чем окружающий воздух. Чтобы цикл повторился, отработавший газ необходимо выбросить,
а вместе с ним уходит и часть энергии. Кроме того, детали трутся, а стенки цилиндра
нагреваются и охлаждаются водой. Теплоту, которую двигатель получает от сгорания топлива,
обозначают $Q_1$; теплоту, отданную окружающей среде (выхлопу и системе охлаждения),
обозначают $Q_2$. Работа равна их разности:

$$A = Q_1 - Q_2.$$

```{admonition} Коэффициент полезного действия
:class: important
Отношение совершённой двигателем работы к количеству теплоты, выделившемуся при сгорании
топлива, называют **коэффициентом полезного действия** теплового двигателя:

$$\eta = \frac{A}{Q_1} = \frac{Q_1 - Q_2}{Q_1}.$$

Обычно КПД выражают в процентах. Поскольку всегда $Q_2 > 0$, работа меньше $Q_1$, и **КПД
меньше 100 %**.
```

Типичные значения из учебника: бензиновый двигатель 20–25 %, дизель около 35 %, паровые
турбоагрегаты электростанций 40–50 %. Паровоз имел 5–8 %.

## 2. Численный опыт: сравниваем двигатели

Рассмотрим одну и ту же задачу, а именно совершение работы 100 МДж (такое количество
требуется легковому автомобилю на сотню километров), и определим, сколько топлива сожжёт
каждый двигатель.

````python
import matplotlib.pyplot as plt
from phys import style

style.use()
A = 100e6                                    # нужная работа, Дж
q = 4.6e7                                    # теплота сгорания жидкого топлива, Дж/кг
dvigateli = {"паровоз": 0.07, "бензиновый": 0.25, "дизель": 0.35, "турбина ТЭС": 0.45}

fig, ax = style.axes("", "Топлива на 100 МДж работы, кг", "Чем выше КПД, тем меньше топлива на ту же работу")
ax.bar(list(dvigateli), [A / (eta * q) for eta in dvigateli.values()])
plt.show()
for nazv, eta in dvigateli.items():
    Q1 = A / eta
    print(f"{nazv:12s} КПД {eta*100:2.0f} %: топлива {Q1 / q:5.1f} кг, в потери уходит {(Q1 - A) / 1e6:5.0f} МДж")
````

Паровоз сжигал в пять раз больше топлива, чем современный дизель, на ту же работу. Марон
(№ 297) предлагает доказать, что замена паровозов тепловозами с КПД 28 % сэкономила 75 %
топлива: $7/28 = 1/4$, то есть тепловозу требуется четверть паровозного расхода.

## 3. Как считать КПД

Пример учебника (упражнение 21). Рабочее тело получило от нагревателя 35 кДж; холодильнику
отдано 75 % этой энергии. Определим работу и КПД.

````python
Q1 = 35e3
Q2 = 0.75 * Q1
A = Q1 - Q2
print(f"работа A = {A / 1e3:.2f} кДж, КПД = {A / Q1 * 100:.0f} %")
````

Задача решается в обе стороны: зная КПД и работу, можно найти топливо; зная топливо
и работу, можно найти КПД. Существенно не путать $Q_1$ (всю энергию топлива) с $A$
(полезной частью).

**Марон, № 294.** Двигатель совершил работу $4{,}6 \cdot 10^7$ Дж, израсходовав 4 кг бензина.
КПД?

````python
A, m, q = 4.6e7, 4, 4.6e7
print(f"энергия топлива {m * q:.2e} Дж, КПД = {A / (m * q) * 100:.0f} %")
````

Ответ сборника: 25 %.

## 4. Тепловые двигатели и окружающая среда

Тепловые двигатели сжигают ежегодно миллиарды тонн топлива. В учебнике (§ 26) перечислены
последствия: в атмосферу уходят углекислый газ, оксиды серы и азота, угарный газ, сажа;
расходуется кислород; города накрывает смог; растёт парниковый эффект. Чёрный дым
из выхлопной трубы при резком старте представляет собой несгоревшее топливо и сажу
(Марон, № 287): смеси не хватило воздуха.

В учебнике указаны следующие меры:

* повышать КПД, то есть получать ту же работу при меньшем сжигании топлива;
* очищать выхлоп: катализаторы, фильтры сажи, контроль состава смеси;
* переходить на более чистое топливо, такое как природный газ и водород;
* развивать электромобили и электростанции без сжигания: гидро-, атомные, солнечные,
  ветровые.

Масштаб проблемы показывает задача Марона № 293. Саяно-Шушенская ГЭС мощностью
$6{,}4 \cdot 10^6$ кВт заменяет тепловую электростанцию с КПД 37 %. Сколько условного топлива
она экономит за сутки?

````python
P = 6.4e9                 # мощность, Вт
t = 24 * 3600             # сутки, с
kpd = 0.37
q_usl = 2.93e7            # удельная теплота сгорания условного топлива, Дж/кг
A = P * t
m = A / kpd / q_usl
print(f"работа за сутки {A:.2e} Дж, сэкономлено около {m / 1e6:.0f} тыс. тонн условного топлива")
````

Получается пятьдесят тысяч тонн условного топлива в сутки, то есть восемьсот вагонов угля,
которые не сожжены и не выброшены в воздух.

## 5. Задачи

**Задача 1 (учебник, упражнение 21).** Тепловой двигатель с КПД 12 % совершает за цикл
работу 150 Дж. Сколько теплоты он получает за цикл от нагревателя и сколько отдаёт
холодильнику?

````python
from phys import quiz
quiz.reveal("Показать разбор",
            "Q₁ = A / η = 150 / 0,12 = 1250 Дж. Холодильнику отдаётся Q₂ = Q₁ − A = 1100 Дж, то есть почти "
            "девять десятых полученной энергии.")
````

**Задача 2 (Марон, № 296).** Автомобиль на пути 110 км израсходовал 6,9 кг бензина. Средняя
мощность двигателя 13 кВт, средняя скорость 75 км/ч. Найдите КПД.

````python
quiz.numeric("КПД двигателя, %:", answer=22, unit="%", tol=0.05,
             explain="Время в пути 110 / 75 = 1,47 ч = 5280 с. Работа A = 13 000 · 5280 ≈ 6,9·10⁷ Дж. "
                     "Энергия топлива Q₁ = 6,9 · 4,6·10⁷ ≈ 3,2·10⁸ Дж. КПД = 6,9·10⁷ / 3,2·10⁸ ≈ 22 %. "
                     "Ответ сборника: 22 %.")
````

**Задача 3 (Марон, № 298).** Расход условного топлива на тепловых электростанциях снизился
с 645 до 327 г на киловатт-час. На сколько вырос КПД?

````python
quiz.reveal("Показать разбор",
            "Киловатт-час составляет 3,6·10⁶ Дж полезной работы. Раньше на неё сжигали 0,645 кг условного "
            "топлива, то есть 0,645 · 2,93·10⁷ ≈ 1,9·10⁷ Дж: КПД ≈ 19 %. Теперь сжигают 0,327 кг, то есть 9,6·10⁶ Дж: "
            "КПД ≈ 38 %. КПД вырос почти вдвое, примерно на 19 процентных пунктов.")
````

**Задача 4 (Марон, № 284).** Почему для автомобилей с частыми остановками нормы расхода
бензина выше?

````python
quiz.reveal("Показать разбор",
            "При каждой остановке кинетическая энергия машины переходит в тепло тормозов, а при разгоне "
            "её необходимо создавать заново, снова за счёт топлива. Кроме того, двигатель на холостом ходу сжигает "
            "бензин, не совершая полезной работы. В результате средний КПД поездки падает.")
````

около 3,1 кг CO₂.
