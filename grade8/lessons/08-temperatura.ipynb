## 1. Тепловые явления и тепловое равновесие

Нагревание и охлаждение воды, таяние льда, плавление парафина в пламени свечи — всё это
тепловые явления. Учебник даёт им определение: **тепловыми** называют явления, важной
характеристикой которых является температура.

В повседневной жизни температура — мера того, насколько нагрето тело. Но ощущения
ненадёжны. Опустите одну руку в горячую воду, другую в холодную, подержите минуту,
а затем обе руки перенесите в тёплую. Одной руке та же вода покажется холодной, другой —
тёплой. Пройдёт немного времени, и ощущения выровняются.

Этот опыт показывает важнейшее свойство температуры.

```{admonition} Тепловое равновесие
:class: important
Если тела разной температуры привести в соприкосновение, то через некоторое время
температуры всех тел станут одинаковыми. Говорят, что в системе тел установилось
**тепловое равновесие**.
```

На этом свойстве держится всё измерение температуры. Термометр, зажатый под мышкой,
показывает не температуру тела, а **собственную** температуру. Просто через несколько
минут между ним и телом устанавливается равновесие, и температуры совпадают. Отсюда
правило: обычный термометр нельзя вынимать из среды до снятия показаний, иначе он начнёт
приходить в равновесие уже с воздухом.

## 2. Что происходит при установлении равновесия

С молекулярной точки зрения тепловое равновесие означает следующее. Молекулы горячего
тела движутся в среднем быстрее. Соприкасаясь с холодным телом, они при столкновениях
передают часть энергии его молекулам. Горячее тело остывает, холодное нагревается, и так
продолжается, пока средние кинетические энергии молекул не выровняются.

не проходят — переходит только энергия: быстрые молекулы слева раскачивают атомы стенки,
те передают колебания молекулам справа.

````python
from phys import sim
sim.thermal_contact()
````

Следите за графиком. Красная линия опускается, синяя поднимается, и через некоторое
время они сходятся к одному уровню — дальше лишь колеблются около него. Это тепловое
равновесие, увиденное на уровне молекул. Заметьте, что число молекул слева и справа
не меняется: вещество через стенку не переносится.

## 3. Термометр и шкала Цельсия

Термометр содержит **рабочее вещество**, свойства которого зависят от температуры.
В жидкостном термометре это спирт или ртуть: при нагревании жидкость расширяется, столбик
удлиняется, и о температуре судят по его длине.

Шкалу термометра строят по двум опорным точкам. Термометр опускают в тающий лёд и ставят
отметку 0 °C, затем в кипящую воду при нормальном атмосферном давлении и ставят отметку
100 °C. Промежуток делят на сто равных частей — получается **шкала Цельсия**, предложенная
шведским учёным Андерсом Цельсием в 1742 году. Температура по этой шкале обозначается
буквой $t$ и измеряется в градусах Цельсия (°C).

```{admonition} Максимальный термометр
:class: note
В капилляре медицинского термометра сделано сужение. Когда термометр вынимают, ртуть
начинает сжиматься, столбик разрывается в сужении, и ртуть выше сужения остаётся на месте.
Показание сохраняется — термометр фиксирует **максимальную** температуру, и его,
единственного, можно снять с тела до снятия показаний. Перед следующим измерением ртуть
стряхивают обратно.
```

## 4. Что характеризует температура

Термометры появились к середине XVIII века, а ответ на вопрос, что именно они измеряют,
физика дала лишь в середине XIX. Ключ — броуновское движение из первого урока: с ростом
температуры броуновские частицы движутся заметнее, значит, быстрее движутся и молекулы
жидкости. Потому беспорядочное движение молекул и называют **тепловым**.

Чем тепловое движение отличается от механического? Механическое движение совершает
отдельное тело, и в любой момент можно указать его положение и скорость. В тепловом
движении участвует огромное число молекул, и следить за каждой невозможно. Поэтому вводят
характеристики не отдельной частицы, а всей их совокупности. Температура — первая из таких
характеристик.

```{admonition} Температура и энергия молекул
:class: important
При тепловом равновесии средние кинетические энергии молекул соприкасающихся тел
одинаковы. Поэтому **температура характеризует среднюю кинетическую энергию теплового
движения молекул**.

Речь именно о средней энергии огромного числа молекул, а не об энергии одной. Скорости
отдельных молекул всё время меняются при столкновениях; средняя же при данной температуре
остаётся неизменной.
```

Численный опыт покажет, как это выглядит. Возьмём газ и посмотрим на скорости отдельных
молекул при двух температурах.

````python
import numpy as np
import matplotlib.pyplot as plt
from phys import style

style.use()
rng = np.random.default_rng(1)

def skorosti(T, n=20000):
    # каждая составляющая скорости разбросана тем сильнее, чем выше температура
    vx, vy = rng.normal(0, np.sqrt(T), n), rng.normal(0, np.sqrt(T), n)
    return np.sqrt(vx**2 + vy**2)

fig, ax = style.axes("Скорость молекулы, усл. ед.", "Доля молекул",
                     "Одна температура — много разных скоростей")
for T, podpis in [(1.0, "газ при температуре T"), (2.0, "тот же газ при 2T")]:
    v = skorosti(T)
    ax.hist(v, bins=80, density=True, alpha=0.6, label=podpis)
    print(f"{podpis:26s} средняя кинетическая энергия {np.mean(v**2 / 2):.2f}, "
          f"самая быстрая молекула в {v.max() / v.mean():.1f} раза быстрее средней")
ax.legend()
plt.show()
````

В газе одной и той же температуры есть и медленные, и очень быстрые молекулы. Температура
ничего не говорит о конкретной молекуле — только о среднем. Нагрели вдвое, и средняя
кинетическая энергия выросла ровно вдвое, а разброс скоростей остался.

## 5. Абсолютный нуль и шкала Кельвина

Если бы молекулы перестали двигаться, их кинетическая энергия обратилась бы в нуль.
Температуру, при которой в теле прекратилось бы тепловое движение, называют
**абсолютным нулём**. Меньшей температуры в природе не существует. По шкале Цельсия
абсолютный нуль равен $-273{,}15$ °C; при расчётах округляют до $-273$ °C.

От абсолютного нуля отсчитывает температуру **шкала Кельвина**, названная по имени
британского физика Уильяма Томсона, лорда Кельвина. Её используют в научных
исследованиях. Температура по Кельвину обозначается $T$, единица — кельвин (К). Один
кельвин равен одному градусу Цельсия, так что разности температур в обеих шкалах
одинаковы, а связь между ними такова:

$$T = t + 273.$$

````python
def v_kelviny(t): return t + 273
def v_celsii(T): return T - 273

for T in (220, 273, 673):
    print(f"{T} К = {v_celsii(T):5d} °C")
for t in (127, -27, 36.6):
    print(f"{t:6} °C = {v_kelviny(t):5.1f} К")
````

Это упражнение 2 из учебника, решённое кодом. Температура тела здорового человека
по Кельвину — около 310 К.
