## 1. Парообразование и его виды

```{admonition} Определения
:class: important
Процесс превращения жидкости в пар называют **парообразованием**. Парообразование
с поверхности жидкости называют **испарением**. Обратный процесс, превращение пара
в жидкость, называют **конденсацией**.
```

Существует и второй вид парообразования, кипение, которое происходит по всему объёму жидкости, ему посвящён следующий урок. Испарение же происходит при любой температуре:
лужи высыхают и летом, и осенью, а бельё сохнет даже на морозе.

## 2. Испарение с точки зрения МКТ

Молекулы жидкости движутся с различными скоростями, и среди них всегда имеются очень
быстрые. Такая молекула, оказавшись у поверхности, может преодолеть притяжение соседей
и вылететь наружу. Медленные молекулы на это не способны, притяжение возвращает их назад.

Из данного рассуждения вытекают два следствия.

Первое следствие состоит в том, что **жидкость при испарении охлаждается**. Уходят самые быстрые молекулы, поэтому средняя кинетическая энергия оставшихся падает и температура понижается. По этой причине при выходе из воды после купания ощущается прохлада,
особенно сильная на ветру.

Второе следствие состоит в том, что **скорость испарения зависит от температуры** (чем
горячее жидкость, тем больше в ней молекул, способных вырваться), а также от рода жидкости,
площади поверхности и от того, уносится ли пар. Эфир испаряется быстрее воды, вода
в тарелке быстрее, чем в бутылке, а бельё на ветру быстрее, чем в безветренном месте.

Рассмотрим модель. Внизу расположена капля жидкости, молекулы которой притягиваются
друг к другу.

````python
from phys import sim
sim.evaporation()
````

Красными отмечены молекулы пара. В отсутствие ветра они движутся над каплей, и часть
из них возвращается обратно. При включении ветра пар уносится, испарение не прекращается,
а средняя кинетическая энергия молекул жидкости заметно падает. Модель невелика и содержит
всего сотню молекул, поэтому числа колеблются, и оценивать следует тенденцию за полминуты, а не мгновенные значения.

Причину столь сильного ускорения испарения при нагреве показывает численный опыт.
Рассчитаем, какая доля молекул имеет энергию, достаточную для вылета, при различных
температурах.

````python
import numpy as np
import matplotlib.pyplot as plt
from phys import style

style.use()
rng = np.random.default_rng(3)
porog = 3.0                                   # энергия, нужная молекуле, чтобы вырваться

def dolya_bystryh(T, n=400_000):
    vx, vy = rng.normal(0, np.sqrt(T), n), rng.normal(0, np.sqrt(T), n)
    E = (vx**2 + vy**2) / 2
    return np.mean(E > porog)

for T in (0.3, 0.4, 0.5, 0.6):
    print(f"температура {T:.1f}: способны вылететь {dolya_bystryh(T) * 100:6.3f} % молекул")
````

Температура выросла вдвое, тогда как доля быстрых молекул увеличилась в сотни раз.
Следовательно, небольшой нагрев превращает медленное испарение в быстрое.

## 3. Насыщенный и ненасыщенный пар

Рассмотрим процессы в закрытом сосуде. Молекулы вылетают из жидкости, но и пар над ней не пуст: его молекулы, ударяясь о поверхность, возвращаются в жидкость. Вначале
вылетающих молекул больше. Пара становится всё больше, и всё больше молекул возвращается.
Наконец за одно и то же время вылетает столько же молекул, сколько возвращается. Уровень
жидкости перестаёт изменяться, хотя обмен молекулами продолжается непрерывно.

```{admonition} Динамическое равновесие
:class: important
Состояние, при котором число молекул, покидающих жидкость за некоторое время, равно числу
молекул, возвращающихся в неё за то же время, называют **динамическим равновесием**.

Пар, находящийся в динамическом равновесии со своей жидкостью, называют **насыщенным**, а пар, не находящийся в таком равновесии, **ненасыщенным**.
```

В открытом сосуде пар уносится и насыщенным не становится, поэтому лужа высыхает
полностью, тогда как в закрытой бутылке вода не убывает годами.

Твёрдые тела также испаряются, минуя жидкое состояние. Так исчезает иней на морозе,
так «тает» сухой лёд, так распространяется запах нафталина. Данный процесс называют
**сублимацией**.

## 4. Конденсация: энергия возвращается

Испарение отбирает у жидкости энергию. Конденсация её возвращает: пар, превращаясь
в жидкость, **выделяет** ровно столько теплоты, сколько было затрачено на испарение.

Это наблюдается в природе, о чём рассказывается в разделе учебника «Водяной пар
в атмосфере». Ночью земля остывает, водяной пар у поверхности становится насыщенным
и конденсируется, в результате чего выпадает роса. При этом выделяется энергия, которая
препятствует остыванию растений и защищает их от заморозков. Таким же образом образуются
туман и облака: воздух, поднимаясь, охлаждается, и пар в нём конденсируется на пылинках,
служащих центрами конденсации. Запотевшие с мороза очки и пар изо рта в холодный день
представляют собой ту же конденсацию в миниатюре.
