## 1. Парообразование и его виды

```{admonition} Определения
:class: important
Процесс превращения жидкости в пар называют **парообразованием**. Парообразование
с поверхности жидкости называют **испарением**. Обратный процесс, превращение пара
в жидкость, называют **конденсацией**.
```

Есть и второй вид парообразования — кипение, оно идёт по всему объёму жидкости; о нём
следующий урок. Испарение же идёт при любой температуре: лужи высыхают и летом, и осенью,
бельё сохнет даже на морозе.

## 2. Испарение с точки зрения МКТ

Молекулы жидкости движутся с разными скоростями, и среди них всегда есть очень быстрые.
Такая молекула, оказавшись у поверхности, может преодолеть притяжение соседей и вылететь
наружу. Медленным молекулам это не под силу: притяжение возвращает их назад.

Отсюда два следствия, и оба важны.

Первое: **жидкость при испарении охлаждается**. Уходят самые быстрые молекулы, значит,
средняя кинетическая энергия оставшихся падает — а это и есть понижение температуры.
Выходя из воды после купания, вы чувствуете прохладу; на ветру — особенно сильно.

Второе: **скорость испарения зависит от температуры** — чем горячее жидкость, тем больше
в ней молекул, способных вырваться, — а также от рода жидкости, площади поверхности
и от того, уносится ли пар. Эфир испаряется быстрее воды, вода в тарелке — быстрее,
чем в бутылке, а бельё на ветру — быстрее, чем в тихом углу.

Посмотрите на модель. Внизу капля жидкости, молекулы которой притягиваются друг к другу.

````python
from phys import sim
sim.evaporation()
````

Красные молекулы — пар. Без ветра они летают над каплей и часть возвращается обратно.
Включите ветер: пар уносится, испарение не прекращается, а средняя кинетическая энергия
молекул жидкости заметно падает. Модель маленькая, всего сотня молекул, поэтому числа
скачут; смотрите на тенденцию за полминуты, а не на мгновенные значения.

Почему нагрев так сильно ускоряет испарение, показывает численный опыт. Считаем, какая
доля молекул имеет энергию, достаточную для вылета, при разных температурах.

````python
import numpy as np
import matplotlib.pyplot as plt
from phys import style

style.use()
rng = np.random.default_rng(3)
porog = 3.0                                   # энергия, нужная молекуле, чтобы вырваться

def dolya_bystryh(T, n=400_000):
    vx, vy = rng.normal(0, np.sqrt(T), n), rng.normal(0, np.sqrt(T), n)
    E = (vx**2 + vy**2) / 2
    return np.mean(E > porog)

for T in (0.3, 0.4, 0.5, 0.6):
    print(f"температура {T:.1f}: способны вылететь {dolya_bystryh(T) * 100:6.3f} % молекул")
````

Температура выросла вдвое, а доля быстрых молекул — в сотни раз. Так небольшой нагрев
превращает медленное испарение в быстрое.

## 3. Насыщенный и ненасыщенный пар

Что происходит в закрытом сосуде? Молекулы вылетают из жидкости, но и пар над ней
не пуст: его молекулы, ударяясь о поверхность, возвращаются в жидкость. Сначала
вылетающих больше. Пара становится всё больше, и всё больше молекул возвращается. Наконец
за одно и то же время сколько молекул вылетает, столько и возвращается. Уровень жидкости
перестаёт меняться, хотя обмен молекулами идёт непрерывно.

```{admonition} Динамическое равновесие
:class: important
Состояние, при котором число молекул, покидающих жидкость за некоторое время, равно числу
молекул, возвращающихся в неё за то же время, называют **динамическим равновесием**.

Пар, находящийся в динамическом равновесии со своей жидкостью, называют **насыщенным**;
не находящийся — **ненасыщенным**.
```

В открытом сосуде пар уносится и насыщенным не становится — потому лужа и высыхает
до конца, а в закрытой бутылке вода не убывает годами.

Твёрдые тела тоже испаряются, минуя жидкое состояние. Так исчезает иней на морозе, так
«тает» сухой лёд, так пахнет нафталин. Этот процесс называют **сублимацией**.

## 4. Конденсация: энергия возвращается

Испарение отбирает у жидкости энергию. Конденсация её возвращает: пар, превращаясь
в жидкость, **выделяет** ровно столько теплоты, сколько было затрачено на испарение.

Это видно в природе, о чём рассказывает учебник в разделе «Водяной пар в атмосфере».
Ночью земля остывает, водяной пар у поверхности становится насыщенным и конденсируется —
выпадает роса. При этом выделяется энергия, которая мешает растениям остыть и защищает
их от заморозков. Так же образуются туман и облака: воздух, поднимаясь, охлаждается,
и пар в нём конденсируется на пылинках — центрах конденсации. А запотевшие с мороза очки
и туман изо рта в холодный день — та же конденсация в миниатюре.
