## 1. Плавление и отвердевание

Рассмотрим опыт, в котором в пробирку помещают около 80 г растолчённого льда при −40 °C
и начинают нагревать, следя за термометром. Температура растёт, при 0 °C появляется вода,
после чего наблюдается особенность: лёд тает, пробирку продолжают нагревать, а термометр
показывает 0 °C, пока не растает последний кусочек льда. Лишь после этого температура
вновь начинает расти.

```{admonition} Определения
:class: important
Переход вещества из твёрдого состояния в жидкое называют **плавлением**, а температуру, при которой вещество плавится, **температурой плавления**.

Переход из жидкого состояния в твёрдое называют **отвердеванием** или **кристаллизацией**, а температуру, при которой вещество отвердевает, **температурой отвердевания (кристаллизации)**.

Кристаллическое вещество плавится и отвердевает при **одной и той же** температуре.
```

| Вещество | $t_{\text{пл}}$, °C | Вещество | $t_{\text{пл}}$, °C | Вещество | $t_{\text{пл}}$, °C |
|---|---:|---|---:|---|---:|
| Водород | −259 | Натрий | 98 | Медь | 1085 |
| Кислород | −219 | Олово | 232 | Чугун | 1200 |
| Азот | −210 | Свинец | 327 | Кремний | 1414 |
| Спирт | −114 | Цинк | 420 | Сталь | 1500 |
| Ртуть | −39 | Алюминий | 660 | Железо | 1539 |
| Лёд | 0 | Серебро | 962 | Платина | 1772 |
| Цезий | 29 | Латунь | 1000 | Осмий | 3045 |
| Калий | 63 | Золото | 1064 | Вольфрам | 3387 |

В таблице приведены значения из учебника (§ 15). Из неё видно, почему нить лампы
изготавливают из вольфрама, почему ртутным термометром нельзя измерять сибирский мороз
и почему цезий можно расплавить в ладони.

## 2. График плавления и отвердевания

Опыт со льдом описывается графиком, который приведён в учебнике на рис. 31. Построим
данный график самостоятельно: используется нагреватель постоянной мощности, лёд при
−40 °C, и отслеживается температура.

````python
from phys import sim
sim.heating_curve()
````

На графике выделяются три участка. На первом **лёд нагревается**, и температура растёт.
На втором **лёд тает**: температура остаётся равной 0 °C, хотя теплота поступает
по-прежнему. На третьем **вода нагревается**, и температура снова растёт, но медленнее, чем росла у льда, так как удельная теплоёмкость воды вдвое больше.

При остывании те же процессы повторяются в обратном порядке: вода остывает до 0 °C, затем
кристаллизуется при постоянной температуре, выделяя теплоту, и только после этого остывает
лёд.

```{admonition} Куда уходит теплота на «полке»
:class: important
Пока лёд тает, температура постоянна, и кинетическая энергия молекул не растёт.
Вся подводимая энергия идёт на **разрушение кристаллической решётки**: молекулы отрываются
от своих узлов, и растёт потенциальная энергия их взаимодействия. Внутренняя энергия
воды при 0 °C больше внутренней энергии льда при 0 °C ровно на эту величину.

При кристаллизации решётка строится заново, и та же энергия **выделяется**.
```

Данное рассуждение составляет содержание задания 1 упражнения 13 учебника: при плавлении
цинка кинетическая энергия его частиц не меняется, тогда как потенциальная растёт.

## 3. Удельная теплота плавления

Полка на графике тем длиннее, чем больше масса льда. Теплота, необходимая для плавления,
пропорциональна массе, а коэффициент пропорциональности зависит от вещества.

```{admonition} Удельная теплота плавления
:class: important
Физическую величину, показывающую, какое количество теплоты необходимо сообщить
кристаллическому телу массой 1 кг, взятому при температуре плавления, чтобы полностью
превратить его в жидкость, называют **удельной теплотой плавления**. Обозначается
буквой $\lambda$ (лямбда), единицей измерения служит Дж/кг.

Количество теплоты, необходимое для плавления тела массой $m$ при температуре плавления:

$$Q = \lambda\,m.$$

Такое же количество теплоты выделяется при кристаллизации той же массы.
```

| Вещество | $\lambda$, 10⁵ Дж/кг | Вещество | $\lambda$, 10⁵ Дж/кг |
|---|---:|---|---:|
| Алюминий | 3,9 | Сталь | 0,84 |
| Лёд | 3,4 | Золото | 0,67 |
| Железо | 2,7 | Водород | 0,59 |
| Медь | 2,1 | Олово | 0,59 |
| Парафин | 1,5 | Свинец | 0,25 |
| Спирт | 1,1 | Кислород | 0,14 |
| Серебро | 0,87 | Ртуть | 0,12 |

Для льда $\lambda = 3{,}4 \cdot 10^5$ Дж/кг, и для плавления килограмма льда требуется 340 кДж. Для сравнения, при сообщении тех же 340 кДж килограмм воды нагрелся бы
на 80 градусов. Следовательно, на плавление льда затрачивается почти столько же теплоты,
сколько на доведение талой воды до кипения.

````python
import matplotlib.pyplot as plt
from phys import style

style.use()
m = 1.0
etapy = {"нагреть лёд\nот −40 до 0 °C": 2100 * m * 40,
         "растопить лёд\nпри 0 °C": 3.4e5 * m,
         "нагреть воду\nот 0 до 100 °C": 4200 * m * 100}
fig, ax = style.axes("", "Количество теплоты, кДж", "Что стоит дороже: 1 кг воды, три этапа")
ax.bar(list(etapy), [v / 1000 for v in etapy.values()])
plt.show()
for k, v in etapy.items():
    print(f"{k.replace(chr(10), ' '):28s} {v / 1000:5.0f} кДж")
````

## 4. Расчёт с нагреванием и плавлением

Если тело взято не при температуре плавления, количество теплоты рассчитывают по этапам:
сначала нагревание до температуры плавления, затем само плавление.

Рассмотрим пример из учебника (§ 17). Определим количество теплоты, необходимое для
плавления свинца массой 0,5 кг, взятого при 27 °C.


````python
c_pb, lam_pb, t_pl = 140, 0.25e5, 327
m, t0 = 0.5, 27
Q1 = c_pb * m * (t_pl - t0)        # нагреть до температуры плавления
Q2 = lam_pb * m                    # расплавить
print(f"нагрев:    Q₁ = 140 · 0,5 · ({t_pl} − {t0}) = {Q1:.0f} Дж")
print(f"плавление: Q₂ = 0,25·10⁵ · 0,5 = {Q2:.0f} Дж")
print(f"всего:     Q = {Q1 + Q2:.0f} Дж = {(Q1 + Q2) / 1000:.1f} кДж")
````

```{admonition} Копнуть глубже: растопит ли солнце метровый снег за день (Савченко, 5.10.2)

смогут ли солнечные лучи на экваторе растопить за день снежный покров толщиной 1 м?

$100 \cdot 43\,200 = 4{,}3 \cdot 10^6$ Дж на квадратный метр. Данной энергии достаточно для
```
