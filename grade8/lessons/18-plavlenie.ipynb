## 1. Плавление и отвердевание

Положим в пробирку около 80 г растолчённого льда при −40 °C и начнём нагревать, следя
за термометром. Температура растёт, при 0 °C появляется вода, и дальше — странность:
лёд тает, пробирку греют, а термометр показывает 0 °C, пока не растает последний кусочек.
Только потом температура снова идёт вверх.

```{admonition} Определения
:class: important
Переход вещества из твёрдого состояния в жидкое называют **плавлением**; температуру,
при которой вещество плавится, — **температурой плавления**.

Переход из жидкого состояния в твёрдое называют **отвердеванием** или
**кристаллизацией**; температуру, при которой вещество отвердевает, — **температурой
отвердевания (кристаллизации)**.

Кристаллическое вещество плавится и отвердевает при **одной и той же** температуре.
```

| Вещество | $t_{\text{пл}}$, °C | Вещество | $t_{\text{пл}}$, °C | Вещество | $t_{\text{пл}}$, °C |
|---|---:|---|---:|---|---:|
| Водород | −259 | Натрий | 98 | Медь | 1085 |
| Кислород | −219 | Олово | 232 | Чугун | 1200 |
| Азот | −210 | Свинец | 327 | Кремний | 1414 |
| Спирт | −114 | Цинк | 420 | Сталь | 1500 |
| Ртуть | −39 | Алюминий | 660 | Железо | 1539 |
| Лёд | 0 | Серебро | 962 | Платина | 1772 |
| Цезий | 29 | Латунь | 1000 | Осмий | 3045 |
| Калий | 63 | Золото | 1064 | Вольфрам | 3387 |

Таблица учебника (§ 15). Из неё видно, почему нить лампы делают из вольфрама, почему
ртутным термометром нельзя мерить сибирский мороз и почему цезий можно расплавить
в ладони.

## 2. График плавления и отвердевания

Опыт со льдом описывается графиком, который учебник приводит на рис. 31. Соберём его
сами: нагреватель постоянной мощности, лёд при −40 °C, следим за температурой.

````python
from phys import sim
sim.heating_curve()
````

Три участка. **Лёд нагревается** — температура растёт. **Лёд тает** — температура стоит
на 0 °C, хотя теплота поступает всё так же. **Вода нагревается** — температура снова
растёт, но медленнее, чем росла у льда: удельная теплоёмкость воды вдвое больше.

При остывании всё повторяется в обратном порядке: вода остывает до 0 °C, затем
кристаллизуется при постоянной температуре, выделяя теплоту, и только потом остывает лёд.

```{admonition} Куда уходит теплота на «полке»
:class: important
Пока лёд тает, кинетическая энергия молекул не растёт — температура постоянна. Вся
подводимая энергия идёт на **разрушение кристаллической решётки**: молекулы отрываются
от своих узлов, и растёт потенциальная энергия их взаимодействия. Внутренняя энергия
воды при 0 °C больше внутренней энергии льда при 0 °C ровно на эту величину.

При кристаллизации решётка строится заново, и та же энергия **выделяется**.
```

Это задание 1 упражнения 13 учебника: при плавлении цинка кинетическая энергия его частиц
не меняется, потенциальная растёт.

## 3. Удельная теплота плавления

Полка на графике тем длиннее, чем больше льда. Теплота, нужная на плавление,
пропорциональна массе, а коэффициент зависит от вещества.

```{admonition} Удельная теплота плавления
:class: important
Физическую величину, показывающую, какое количество теплоты необходимо сообщить
кристаллическому телу массой 1 кг, взятому при температуре плавления, чтобы полностью
превратить его в жидкость, называют **удельной теплотой плавления**. Обозначение
$\lambda$ (лямбда), единица — Дж/кг.

Количество теплоты, необходимое для плавления тела массой $m$ при температуре плавления:

$$Q = \lambda\,m.$$

Столько же теплоты выделяется при кристаллизации той же массы.
```

| Вещество | $\lambda$, 10⁵ Дж/кг | Вещество | $\lambda$, 10⁵ Дж/кг |
|---|---:|---|---:|
| Алюминий | 3,9 | Сталь | 0,84 |
| Лёд | 3,4 | Золото | 0,67 |
| Железо | 2,7 | Водород | 0,59 |
| Медь | 2,1 | Олово | 0,59 |
| Парафин | 1,5 | Свинец | 0,25 |
| Спирт | 1,1 | Кислород | 0,14 |
| Серебро | 0,87 | Ртуть | 0,12 |

У льда $\lambda = 3{,}4 \cdot 10^5$ Дж/кг: чтобы растопить килограмм льда, нужно 340 кДж.
Сравните с нагревом воды — на те же 340 кДж килограмм воды нагрелся бы на 80 градусов.
Растопить лёд стоит почти столько же, сколько довести талую воду до кипения.

````python
import matplotlib.pyplot as plt
from phys import style

style.use()
m = 1.0
etapy = {"нагреть лёд\nот −40 до 0 °C": 2100 * m * 40,
         "растопить лёд\nпри 0 °C": 3.4e5 * m,
         "нагреть воду\nот 0 до 100 °C": 4200 * m * 100}
fig, ax = style.axes("", "Количество теплоты, кДж", "Что стоит дороже: 1 кг воды, три этапа")
ax.bar(list(etapy), [v / 1000 for v in etapy.values()])
plt.show()
for k, v in etapy.items():
    print(f"{k.replace(chr(10), ' '):28s} {v / 1000:5.0f} кДж")
````

## 4. Расчёт с нагреванием и плавлением

Если тело взято не при температуре плавления, теплоту считают по этапам: сначала нагрев
до температуры плавления, потом само плавление.

Пример учебника (§ 17). Сколько теплоты нужно, чтобы расплавить свинец массой 0,5 кг,
взятый при 27 °C?

````python
c_pb, lam_pb, t_pl = 140, 0.25e5, 327
m, t0 = 0.5, 27
Q1 = c_pb * m * (t_pl - t0)        # нагреть до температуры плавления
Q2 = lam_pb * m                    # расплавить
print(f"нагрев:    Q₁ = 140 · 0,5 · ({t_pl} − {t0}) = {Q1:.0f} Дж")
print(f"плавление: Q₂ = 0,25·10⁵ · 0,5 = {Q2:.0f} Дж")
print(f"всего:     Q = {Q1 + Q2:.0f} Дж = {(Q1 + Q2) / 1000:.1f} кДж")
````

```{admonition} Копнуть глубже: растопит ли солнце метровый снег за день (Савченко, 5.10.2)
:class: dropdown
Задача из [задачника Савченко](https://inp.nsk.su/~telnov/mech/zadachniki/savchenko.pdf), § 5.10:
смогут ли солнечные лучи на экваторе растопить за день снежный покров толщиной 1 м?
Поток солнечной энергии до 1 кВт/м², но снег отражает 90 % света.

Оценка. Поглощается 100 Вт/м². За 12 часов это $100 \cdot 43\,200 = 4{,}3 \cdot 10^6$ Дж
на квадратный метр. Расплавить можно $4{,}3 \cdot 10^6 / 3{,}4 \cdot 10^5 \approx 13$ кг льда —
слой воды около 13 мм, или снега в несколько раз толще, 5–10 см. Метровый покров и близко
не растает. Вот почему снег в горах лежит до лета даже под ярким солнцем: спасает не только
холод, но и белизна — почти весь свет отражается.
```

```{admonition} Копнуть глубже: температура плавления и давление
:class: dropdown
Температура плавления зависит от давления. У льда она с ростом давления **понижается** —
редкое свойство. Опыт из учебника: на брусок льда надевают проволочную петлю с грузом, и
проволока медленно проходит сквозь брусок, а лёд над ней смерзается обратно. Под проволокой
давление велико, лёд тает при температуре ниже нуля; вода вытекает наверх, где давления
нет, и вновь замерзает.

Это упражнение 13, задание 4 учебника. На том же явлении держится скольжение конька:
под узким лезвием давление огромно, лёд подтаивает, и конёк едет по плёнке воды.
```
