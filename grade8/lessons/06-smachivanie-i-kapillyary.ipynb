## 1. Молекулы разных веществ

До сих пор речь шла о притяжении молекул **одного** вещества. Но молекулы разных
веществ тоже тянутся друг к другу. Из соревнования этих двух притяжений и вырастает
вся сегодняшняя тема.

Вот опыт из учебника. Стеклянную пластину кладут на воду и тянут вверх за пружину.
Пластина отрывается и оказывается **покрытой водой**. Разрыв прошёл не между водой
и стеклом, а внутри самой воды.

**Вывод:** притяжение молекул воды друг к другу **слабее**, чем притяжение молекул
воды к молекулам стекла.

Замените воду ртутью, и стекло останется сухим. У ртути всё наоборот, её молекулы
держатся друг за друга крепче, чем за стекло.

```{admonition} Смачивание и несмачивание
:class: important
Если притяжение молекул жидкости к молекулам поверхности **больше**, чем притяжение
молекул жидкости друг к другу, — жидкость **смачивает** поверхность и растекается
по ней.

Если **меньше** — жидкость **не смачивает** поверхность и собирается в каплю.
```

Вода смачивает стекло, дерево, ткань, бумагу. Вода не смачивает жирные поверхности,
воск, перья водоплавающих птиц. Ртуть не смачивает почти ничего: ни стекло, ни кожу, ни дерево.

## 2. Модель: капля на поверхности

В модели молекулы жидкости притягиваются друг к другу, как и прежде. Ползунок задаёт,
насколько сильно они притягиваются **к поверхности**. Больше ничего не меняется.

````python
from phys import sim
sim.wetting()
````

Слева ползунка поверхность вроде жирной или воскованной, и капля стоит шариком.
Справа стоит чистое стекло, по которому жидкость растекается плёнкой.

Переход резкий. Стоит притяжению к поверхности перевесить, и капля расползается.

## 3. Поверхностное натяжение

Почему капля, которая не смачивает поверхность, принимает форму шарика? Здесь работает ещё одно явление, **поверхностное натяжение**.

Возьмите молекулу **внутри** жидкости. Соседи окружают её со всех сторон и тянут
во все стороны одинаково, так что силы уравновешены. Теперь возьмите молекулу
**на поверхности**. Снизу и с боков соседи есть, а сверху жидкости нет, и тянуть
её вверх нечему. Равнодействующая направлена внутрь жидкости.

```{admonition} Откуда берётся поверхностное натяжение
:class: important
Молекулы на поверхности жидкости втягиваются внутрь, и поверхность стремится
**сократиться** — стать как можно меньше по площади.

А наименьшую площадь поверхности при данном объёме имеет шар. Поэтому свободная
капля принимает форму шарика.
```

Поверхностное натяжение стоит за множеством привычных вещей. Капля росы собирается
в шарик, мыльный пузырь получается сферическим, лёгкая иголка лежит на воде и не тонет,
а водомерки бегают по пруду, не проваливаясь.

Прочность этой «плёнки» описывают особой величиной, **поверхностным натяжением**
$\sigma$.
Для воды при комнатной температуре $\sigma = 0{,}073$ Н/м. У мыльного раствора оно в два-три раза меньше, поэтому мыло «разрушает» плёнку
и иголка тонет.

## 4. Форма поверхности у стенок и капилляры

Вблизи стенок сосуда поверхность жидкости искривлена, и вид искривления зависит
от смачивания:

* если жидкость **смачивает** стенки, поверхность у них **вогнутая**, как вода в стакане;
* если **не смачивает**, поверхность **выпуклая**, как ртуть в стеклянной трубке.

Если сосуд узкий, искривлённой оказывается вся поверхность жидкости. Такую узкую
трубку называют **капилляром** (от латинского *capillus* — волос).

```{admonition} Капиллярные явления
:class: important
В узких трубках смачивающие жидкости **поднимаются выше** уровня в широких сосудах,
а несмачивающие — **опускаются ниже**. Это и есть капиллярные явления.

Высота подъёма зависит от рода жидкости и **обратно пропорциональна радиусу**
трубки: чем тоньше капилляр, тем выше поднимается жидкость.
```

В модели ниже стоят пять трубок разного радиуса в одном сосуде. Ползунок меняет краевой
угол, то есть насколько хорошо жидкость смачивает стенки.

````python
sim.capillary()
````

Высота подъёма считается по формуле

$$h = \frac{2\sigma\cos\theta}{\rho g r},$$

где $\sigma$ — поверхностное натяжение, $\theta$ — краевой угол, $\rho$ — плотность
жидкости, $r$ — радиус трубки. Заучивать её не нужно. Важно другое:
**$h$ обратно пропорционально $r$**. Уменьшили радиус вдвое, и подъём вырос вдвое.

````python
import numpy as np
import matplotlib.pyplot as plt
from phys import style

style.use()
sigma = 0.073      # поверхностное натяжение воды, Н/м
rho = 1000         # плотность воды, кг/м³
g = 9.8

def podyom(radius_mm, ugol_grad=0):
    """Высота подъёма воды в капилляре радиусом radius_mm, мм."""
    r = radius_mm / 1000
    return 2 * sigma * np.cos(np.radians(ugol_grad)) / (rho * g * r) * 1000

radiusy = np.linspace(0.02, 2, 300)
fig, ax = style.axes("Радиус трубки, мм", "Высота подъёма, мм",
                     "Чем тоньше капилляр, тем выше поднимается вода")
ax.plot(radiusy, podyom(radiusy))
ax.set_ylim(0, 800)
plt.show()

for r in [0.01, 0.05, 0.1, 0.5, 1.0]:
    print(f"радиус {r:5.2f} мм  →  подъём {podyom(r):7.1f} мм")
````

Так вода и поднимается по стеблю растения или по бумажной салфетке, ведь внутри них
множество тончайших каналов.

## 5. Где это встречается

* **Растения** получают воду из почвы по капиллярам, без всякого насоса.
* **Почву рыхлят**, чтобы разрушить капилляры и уменьшить испарение влаги:
  по целым капиллярам вода поднимается к поверхности и испаряется.
* **Полотенце** впитывает воду, потому что ткань пронизана капиллярами.
* **Гидроизоляцию** кладут в фундамент, чтобы влага не поднималась по капиллярам
  кирпича и бетона в стены.
* **Нефть** смачивает перья птиц и вытесняет из-под них воздух. Пух намокает,
  птица теряет и способность летать, и защиту от холода. Поэтому аварии танкеров
  так губительны для водоплавающих.

## 6. Опыт своими руками

```{admonition} Опыт 1. Измеряем подъём воды в капиллярах
:class: opyt
**Что нужно:** бумажные полоски (салфетка, фильтровальная бумага, газета), стакан
с водой, линейка, часы, пищевой краситель или чернила.

**Что делать:**

1. Нарежьте полоски шириной 2 см из разных сортов бумаги.
2. Подкрасьте воду, чтобы граница подъёма была видна.
3. Опустите полоски в воду на 1 см, закрепив верхние концы, чтобы не падали.
4. Каждую минуту отмечайте карандашом, до какой высоты поднялась вода.
5. Через 10 минут измерьте окончательную высоту для каждой бумаги.

**Что записать:** таблицу «время — высота» для каждого сорта бумаги.

**Что обсудить:** у какой бумаги вода поднялась выше? О чём это говорит
о толщине её капилляров?
```

По высоте подъёма можно прикинуть, какой толщины каналы внутри вашей бумаги.

````python
# ↓ результаты вашего опыта
h_izmerennaya = 0.045     # высота подъёма воды, м (например, 45 мм)

sigma, rho, g = 0.073, 1000, 9.8
r = 2 * sigma / (rho * g * h_izmerennaya)     # выражаем радиус из формулы подъёма

print(f"Вода поднялась на {h_izmerennaya * 1000:.0f} мм")
print(f"Оценка радиуса капилляров: {r * 1e6:.0f} мкм = {r * 1000:.3f} мм")
print(f"Это примерно в {r / 3e-10:.0e} раз больше молекулы воды")
````

```{admonition} Насколько можно верить этой оценке
:class: note
Настоящие поры бумаги в несколько раз тоньше, чем показывает наш расчёт.
Формула описывает **гладкую прямую трубку**, а в бумаге вода идёт по извилистым
каналам между волокнами, цепляясь за них и испаряясь по дороге. Поэтому она
поднимается ниже, чем «должна», — и наш расчёт даёт завышенный радиус.

Такую величину называют **эффективным радиусом**: это радиус той идеальной трубки,
которая вела бы себя так же, как наш запутанный клубок каналов. В физике так
поступают часто, когда точное описание слишком сложно, и всегда честно оговаривают,
что величина эффективная.
```
