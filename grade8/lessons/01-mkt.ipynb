## 1. Откуда взялась мысль о молекулах

Разломим кусочек мела пополам. Затем ещё раз, и ещё.

Долго ли так можно продолжать? Этот вопрос поставил Демокрит две с половиной тысячи лет
назад. Возможных ответов он видел два. Либо деление продолжается неограниченно, либо
однажды остаётся частица, которую разделить нельзя. Такую частицу он назвал **атомом**
(по-гречески *atomos* означает «неделимый»).

Проверить эту догадку Демокрит не мог: подходящих опытов не существовало. Две тысячи лет
она оставалась чистым умозрением.

Наукой учение о молекулах стало только во второй половине XIX века, когда его наконец
подтвердили опытами. Называют его **молекулярно-кинетической теорией**, сокращённо МКТ.
Слово «кинетическая» происходит от греческого *kinetikos*, «приводящий в движение»,
и указывает на главную роль движения частиц в этой теории.

```{admonition} Как устроена физическая теория
:class: note
В обычной речи «теория» значит «предположение». В физике наоборот. Теория здесь —
это утверждение, из которого следуют проверяемые предсказания, и все они до сих пор
сбывались. Не сбудется хоть одно — теорию придётся менять.

Поэтому рядом с каждым положением МКТ дальше будет стоять опыт, который его
подтверждает.
```

## 2. Три положения МКТ

```{admonition} Основные положения молекулярно-кинетической теории
:class: important
1. **Все вещества состоят из частиц** (молекул, атомов), между которыми существуют промежутки.
2. **Все частицы вещества находятся в непрерывном хаотическом** (беспорядочном) **движении**.
3. **Частицы вещества взаимодействуют друг с другом** силами притяжения и отталкивания.
```

Три утверждения.

Из них следует, почему нагретый воздух поднимается вверх, почему сахар растворяется
в чае, почему сталь прочна, а вода текуча, почему зимой разрывает водопроводные трубы,
почему запах из кухни доходит до дальней комнаты и почему бельё высыхает на морозе.
Весь курс физики 8 класса представляет собой развёрнутое следствие этих трёх положений.

## 3. Опытные подтверждения

### Первое положение: вещество состоит из частиц с промежутками между ними

Молекула не видна ни невооружённым глазом, ни в оптический микроскоп, поскольку её
размеры для этого слишком малы. Судить о молекулах приходится **косвенно**, по явлениям,
которые без них объяснения не получают.

| Опыт или явление | Что из него следует |
|---|---|
| **Тепловое расширение**: нагретое тело увеличивается в объёме | Между частицами есть промежутки, которые могут меняться |
| **Сжатие газа** в шприце с закрытым отверстием | В газе промежутки между молекулами велики, их можно уменьшить |
| **Растворение**: 100 мл спирта + 100 мл воды дают не 200, а около 195 мл | Молекулы одного вещества занимают промежутки между молекулами другого |
| **Диффузия**: капля чернил окрашивает всю воду | Молекулы проникают в промежутки между чужими молекулами |

На опыте со спиртом и водой стоит остановиться подробнее. Пять миллилитров объёма
исчезают. Вещество при этом не испаряется и не вытекает: масса смеси в точности равна
сумме масс исходных жидкостей, что проверяется взвешиванием. Уменьшается только объём: молекулы упаковываются плотнее, чем были упакованы
до смешивания.

Наглядная аналогия — стакан пшена, высыпанный в стакан гороха. Смеси получится меньше
двух стаканов, поскольку мелкие зёрна занимают промежутки между крупными.

### Второе положение: частицы непрерывно и хаотически движутся

Здесь два главных доказательства — **диффузия** и **броуновское движение**.

**Диффузия** — самопроизвольное взаимное проникновение веществ друг в друга.
Размешивать чернила не требуется, они расходятся по воде сами. Объяснение здесь
единственное: молекулы обоих веществ непрерывно движутся и проникают в промежутки
между чужими молекулами.

В модели ниже за перегородкой заперты два газа. Уберите её и посмотрите, что будет.
Потом попробуйте то же самое при другой температуре.

Законы механики этого не запрещают.

Каждая молекула движется по сосуду независимо от остальных, и вероятность оказаться

одновременно «орлом» окажутся сто молекул, составляет около $10^{-30}$: ожидание
такого события заняло бы больше времени, чем существует Вселенная.

Диффузия необратима не потому, что обратный процесс запрещён. Он лишь неизмеримо
менее вероятен.

### Броуновское движение

В 1827 году шотландский ботаник Роберт Броун наблюдал в микроскоп пыльцу, взвешенную
в воде. Частицы вели себя необычно: они не плыли по течению, а совершали беспорядочные
скачки, вычерчивая ломаные линии, и движение это не прекращалось.

Сначала Броун счёл его признаком жизни. Затем он растёр в порошок камень и рассмотрел
взвесь в том же микроскопе. Картина оказалась той же самой, хотя камень живым
не был.

Объяснение появилось лишь через восемьдесят лет. В 1905 году **Альберт Эйнштейн**
показал происхождение этого движения, а **Жан Перрен** подтвердил его расчёты опытом.

Видимую частицу со всех сторон бомбардируют невидимые молекулы воды. Ударов очень много,
и в среднем они компенсируют друг друга, но именно в среднем, а не в каждый отдельный
момент. Достаточно, чтобы слева удары оказались чуть сильнее, чем справа, и частица
сместится.

````python
sim.brownian()
````

Красная линия — след частицы. Положение отмечается через равные промежутки времени,
точно так же зарисовывал свои наблюдения Перрен.

По двум признакам броуновское движение узнают сразу. Оно **никогда не прекращается**, сколько бы долго за частицей ни наблюдали. И оно тем **заметнее, чем выше температура**
и чем **мельче сама частица**.

Так и получилось прямое доказательство. Тепловое движение молекул существует
и не затухает.

```{admonition} Почему броуновское движение видно только у мелких частиц
:class: dropdown
Число ударов о частицу пропорционально её площади поверхности, то есть растёт как $R^2$.
А масса частицы растёт как $R^3$. Разница между числом ударов слева и справа —
величина случайная, и она составляет примерно корень из полного числа ударов $N$,
то есть $\sqrt{N}$.

Получается, что ускорение, которое частица получает от «лишних» ударов, ведёт себя как
$a \sim \sqrt{N}/m \sim R/R^3 = R^{-2}$. Увеличили частицу в 10 раз — дрожание ослабло
в 100 раз. Поэтому пылинка размером в микрометр пляшет, а песчинка в миллиметр
уже неподвижна.
```

### Третье положение: частицы притягиваются и отталкиваются

Не будь притяжения между молекулами, любое вещество мгновенно разлетелось бы. Ни
твёрдых тел, ни жидкостей просто не существовало бы. Значит, притяжение есть.

Опыты, которые это показывают:

* два свинцовых цилиндра со свежесрезанными и прижатыми друг к другу гладкими торцами
  слипаются так, что выдерживают гирю;
* капля воды не распадается, а собирается в шарик — молекулы стягивают её поверхность;
* чтобы разорвать проволоку, нужна сила, и тем большая, чем толще проволока.

А отталкивание? Оно включается, когда молекулы сходятся слишком близко. Попробуйте сжать воду в закрытом шприце, и у вас ничего не получится, хотя воздух
в том же шприце сжимается легко.

```{admonition} Итог: одна таблица на весь урок
:class: important
| Положение МКТ | Опыт-подтверждение |
|---|---|
| 1. Частицы и промежутки между ними | тепловое расширение, сжатие газа, смешивание спирта с водой |
| 2. Непрерывное хаотическое движение | диффузия, броуновское движение |
| 3. Притяжение и отталкивание | слипание свинцовых цилиндров, прочность тел, несжимаемость жидкости |
```

## 4. Численный опыт: почему след частицы выглядит именно так

Объяснение Эйнштейна допускает прямую проверку. Зададим в модели одно-единственное
правило: на каждом шаге частица получает толчок в случайном направлении. Ни молекул, ни воды, ни температуры в модель не заложено, одна лишь случайность.

Если результат воспроизведёт картину, которую Броун видел в микроскоп, объяснение
следует признать верным.

Разбирать код не обязательно, достаточно смотреть на рисунки и выводы под ними.

````python
import numpy as np
import matplotlib.pyplot as plt
from phys import style

style.use()
rng = np.random.default_rng(8)          # одно и то же «случайное» — для повторяемости опыта

def blujdanie(shagov, chastic=1):
    """Случайное блуждание: каждый шаг — толчок в случайную сторону."""
    ugly = rng.uniform(0, 2 * np.pi, size=(chastic, shagov))
    x = np.cumsum(np.cos(ugly), axis=1)
    y = np.cumsum(np.sin(ugly), axis=1)
    return x, y

fig, ax = plt.subplots(figsize=(5.6, 5.2))
for nomer in range(4):
    x, y = blujdanie(400)
    ax.plot(x[0], y[0], linewidth=1.2, alpha=0.9)
ax.plot(0, 0, "ko", markersize=6)
ax.set_title("Четыре броуновские частицы, вышедшие из одной точки")
ax.set_xlabel("x, условные единицы")
ax.set_ylabel("y, условные единицы")
ax.set_aspect("equal")
plt.show()
````

Результат воспроизводит зарисовки Перрена, хотя в модель заложена одна лишь
случайность.

Следующий вопрос: **как далеко** частица уходит за время $t$?

Ответ кажется очевидным. Частица движется непрерывно, значит, за вдвое большее время
должна уйти вдвое дальше. Проверим это на двух тысячах частиц, измерив среднее
расстояние от точки старта.

````python
x, y = blujdanie(500, chastic=2000)
rasstoyanie = np.sqrt(x**2 + y**2)        # расстояние каждой частицы от старта
srednee = rasstoyanie.mean(axis=0)        # усредняем по всем 2000 частицам
shagi = np.arange(1, 501)

fig, ax = style.axes("Число шагов (время)", "Среднее расстояние от старта",
                     "Частица удаляется всё медленнее")
ax.plot(shagi, srednee, label="результат моделирования")
ax.plot(shagi, 0.886 * np.sqrt(shagi), "--", label="корень из числа шагов")
ax.legend()
plt.show()

for shag in (100, 200, 400):
    print(f"после {shag:3d} шагов частица в среднем удалилась на {srednee[shag - 1]:.1f}")
````

Посмотрите на числа. Время выросло вчетверо, со 100 шагов до 400, а расстояние увеличилось лишь вдвое.
Результат этот неожидан и важен:

```{admonition} Главный вывод
:class: important
При случайном блуждании расстояние от старта растёт **не как время, а как корень
из времени**. Чтобы уйти вдвое дальше, нужно вчетверо больше времени.
```

Отсюда становится понятной и медлительность диффузии.

Молекулы воды движутся со скоростью в сотни метров в секунду. Такой скорости хватило бы,
чтобы пересечь стакан тысячи раз за секунду. Тем не менее чернила расходятся минутами.
Причина в том, что прямолинейно молекула не движется практически никогда: пролетев около
десятой доли микрометра, она сталкивается с соседней молекулой и меняет направление.
Столкновения происходят миллиарды раз в секунду, и каждое сбивает частицу с прежнего
пути. Продвижение вперёд оказывается несравнимо меньше пройденного пути.

```{admonition} Копнуть глубже: почему смещение растёт именно как корень
:class: dropdown
Численный опыт показал закон, но не доказал его. Между тем доказательство укладывается
в несколько строк и требует только формулы квадрата суммы.

Пусть частица движется по прямой и каждый шаг сдвигается на расстояние $a$ вправо
или влево, причём то и другое равновероятно. Обозначим направление $i$-го шага
через $\xi_i$, где $\xi_i = +1$ при шаге вправо и $\xi_i = -1$ при шаге влево.
После $N$ шагов частица окажется в точке

$$x = a\,(\xi_1 + \xi_2 + \ldots + \xi_N).$$

Само по себе $x$ нам ничего не даст: вправо и влево частица уходит одинаково часто,
и среднее значение $x$ равно нулю. Поэтому и берут **квадрат** смещения, который
всегда положителен. Возведём в квадрат по формуле квадрата суммы:

$$x^2 = a^2\,\bigl(\underbrace{\xi_1^2 + \xi_2^2 + \ldots + \xi_N^2}_{\text{квадраты}}
 + \underbrace{2\xi_1\xi_2 + 2\xi_1\xi_3 + \ldots}_{\text{попарные произведения}}\bigr).$$

Дальше всё решают два наблюдения.

**Квадраты.** Каждое $\xi_i$ равно $+1$ или $-1$, а квадрат и того и другого равен
единице. Значит, квадратов ровно $N$ штук и сумма их равна $N$.

**Попарные произведения.** Каждое произведение $\xi_i \xi_j$ тоже равно $+1$ или $-1$:
плюс, если оба шага в одну сторону, минус, если в разные. Направления шагов
не зависят друг от друга, поэтому оба случая встречаются одинаково часто, и при
усреднении по многим частицам каждое такое слагаемое даёт ноль.

Остаются только квадраты:

$$\langle x^2 \rangle = a^2 N.$$

Число шагов пропорционально времени, $N \sim t$, откуда

$$\sqrt{\langle x^2 \rangle} = a\sqrt{N} \sim \sqrt{t}.$$

Вот и весь вывод. Обратите внимание, где именно спрятана физика: в утверждении,
что направления шагов независимы. Молекула действительно не помнит, куда летела
до столкновения, и потому попарные произведения гасятся. Если бы память сохранялась —
скажем, частицу сносило бы течением, — слагаемые не гасились бы, и смещение росло бы
пропорционально времени, а не корню из него.
```

Доказательство можно проверить прямо здесь. Запустим одномерное блуждание: частица
шагает по прямой вправо или влево, и посмотрим, действительно ли средний квадрат
смещения равен числу шагов.

````python
rng2 = np.random.default_rng(3)

print("шагов N     <x²> измеренное    N (предсказание)   √<x²>    √N")
for N in (100, 400, 1600, 6400):
    shagi = rng2.choice([-1, 1], size=(20000, N))   # 20 000 частиц по N шагов
    x = shagi.sum(axis=1)                            # где оказалась каждая
    sredniy_kvadrat = (x**2).mean()
    print(f"{N:7d} {sredniy_kvadrat:16.0f} {N:18d} {np.sqrt(sredniy_kvadrat):9.1f} {np.sqrt(N):6.1f}")
````

Совпадение с точностью до случайного разброса. Увеличение числа шагов вчетверо
удваивает смещение, и никакие другие законы этому набору случайных толчков
не подходят.

```{admonition} Копнуть глубже: как из этого получили число Авогадро
:class: dropdown
Эйнштейн в 1905 году получил формулу для среднего квадрата смещения броуновской частицы:

$$\langle x^2 \rangle = \frac{RT}{N_A} \cdot \frac{t}{3\pi \eta r},$$

где $\eta$ — вязкость жидкости, $r$ — радиус частицы, $R$ — универсальная газовая
постоянная. Всё в правой части, кроме $N_A$, можно измерить. Значит, наблюдая под
микроскопом дрожание частички и измеряя её смещения, **можно сосчитать число Авогадро** —
узнать, сколько молекул в моле вещества, ни разу не увидев ни одной молекулы.

Именно это и проделал Жан Перрен в 1908 году, получив $N_A \approx 6{,}5 \cdot 10^{23}$
(современное значение $6{,}02 \cdot 10^{23}$). За эту работу он получил Нобелевскую
премию 1926 года, а спор о том, существуют ли атомы на самом деле, был окончательно
закрыт.
```

## 5. Физика рядом: вакуум в ускорителе

Ускоритель заряженных частиц — это труба, по которой движется пучок электронов
или протонов; свёрнута она в кольцо или вытянута в линию, для дальнейшего неважно.
Существенно другое: воздуха внутри быть не должно. Столкновения с молекулами
остаточного газа рассеивают пучок, и он гибнет, не выполнив своей задачи. Поэтому газ
из трубы откачивают до **сверхвысокого вакуума**.

Насколько высокого, показывает конкретный пример.

В новосибирском Академгородке Институт ядерной физики строит **СКИФ** (Сибирский
кольцевой источник фотонов). Давление в его основном накопительном кольце должно
составлять $2 \cdot 10^{-9}$ Торр; при таком вакууме время жизни пучка максимально.

```{admonition} Что такое Торр
:class: note
Это привычные вам миллиметры ртутного столба из 7 класса. Один Торр — давление
столбика ртути высотой 1 мм. Нормальное атмосферное давление равно 760 Торр,
то есть 760 мм рт. ст.
```

Вопрос: сколько молекул остаётся в таком «пустом» кубическом сантиметре?

Для расчёта достаточно уже известного, новой физики не потребуется.

Опытом установлено, что в кубическом сантиметре воздуха при обычных условиях содержится
примерно $2{,}7 \cdot 10^{19}$ молекул. Число молекул в единице объёма пропорционально
давлению: во сколько раз понижено давление, во столько же раз меньше остаётся молекул.

````python
n_vozduh = 2.7e19        # молекул в 1 см³ воздуха при обычных условиях
p_atm = 760              # атмосферное давление, Торр
p_skif = 2e-9            # вакуум в накопительном кольце СКИФ, Торр

vo_skolko_raz = p_atm / p_skif        # во столько раз откачали воздух
n_skif = n_vozduh / vo_skolko_raz     # во столько же раз меньше стало молекул

print(f"Воздух откачали в {vo_skolko_raz:.1e} раз")
print(f"В 1 см³ воздуха:        {n_vozduh:.1e} молекул")
print(f"В 1 см³ кольца СКИФ:    {n_skif:.1e} молекул")
````

Даже лучшая современная вакуумная техника оставляет в кубическом сантиметре около
семидесяти миллионов молекул.

Полностью удалить вещество не удаётся: сколько ни откачивай, часть молекул остаётся.
Это ещё один довод в пользу первого положения МКТ. Будь вещество сплошным, его можно
было бы разрежать неограниченно, деля оставшуюся порцию пополам. Но частицы дискретны, и число их убывает поштучно, так что каждую в принципе можно
сосчитать.

При такой концентрации молекулы практически не мешают друг другу. Средняя длина свободного пробега, то есть расстояние между двумя последовательными
столкновениями молекулы, достигает здесь сотен километров. До стенки камеры молекула
долетает скорее, чем встречает другую молекулу. Благодаря этому пучок сохраняется в кольце часами,
проходя за это время миллиарды километров.

```{admonition} Копнуть глубже: сколько ударов держит стена вашей комнаты
:class: dropdown
Задача 5.4.1 из задачника Савченко звучит так: *оцените число молекул воздуха,
попадающих на 1 см² стены вашей комнаты за 1 секунду.*

Для оценки достаточно того, что мы уже знаем. Пусть в кубическом метре воздуха
$n = 2{,}7 \cdot 10^{25}$ молекул, а движутся они со средней скоростью около 500 м/с.
За секунду до площадки $S$ долетят молекулы из объёма $v \cdot S$, но лишь примерно
четверть из них движется в нужную сторону:

$$N \approx \frac{1}{4}\, n\, v\, S \approx
\frac{1}{4} \cdot 2{,}7 \cdot 10^{25} \cdot 500 \cdot 10^{-4}
\approx 3 \cdot 10^{23}\ \text{ударов в секунду}.$$

Триста секстиллионов ударов ежесекундно о площадку размером с ноготь. Даже
за одну миллиардную долю секунды по ней успевает ударить $3 \cdot 10^{14}$ молекул.

Теперь понятно, почему стена не дрожит, а броуновская частица дрожит. Ударов так много,
что для крупного тела они идеально уравновешиваются, и остаётся ровное давление воздуха.
```

## 6. Опыт своими руками: измеряем диффузию

Численная модель показала, чего следует ожидать. Теперь проверим предсказание
на опыте. Он несложен, но даёт **число**, которое можно сопоставить с теорией.

```{admonition} Опыт 1. Диффузия в горячей и холодной воде
:class: opyt
**Что нужно:** два одинаковых прозрачных стакана, холодная и горячая вода,
пакетик чая (или несколько кристалликов марганцовки, или капля пищевого красителя),
часы с секундной стрелкой или секундомер в телефоне, линейка.

**Что делать:**

1. Налейте в первый стакан холодную воду из-под крана, во второй — горячую.
   Уровень воды сделайте одинаковым и измерьте его линейкой.
2. Одновременно опустите в оба стакана по пакетику чая (или бросьте по кристаллику
   марганцовки). **Не размешивайте и не двигайте стаканы** — иначе вы будете изучать
   перемешивание, а не диффузию.
3. Включите секундомер. Каждые 30 секунд отмечайте, до какой высоты опустилась
   окрашенная область в каждом стакане.
4. Засеките время, за которое вода окрасится равномерно.

**Что записать:** таблицу «время — высота окрашенного слоя» для обоих стаканов
и итоговое время полного окрашивания.
```

```{admonition} Осторожно
:class: tb
Горячую воду наливайте из чайника, а не кипяток. Стакан должен быть термостойким.
Марганцовку берите пинцетом и не трогайте руками.
```

Теперь обработайте результаты. Впишите **свои** числа, и расчёт выполнится сам.

````python
# ↓ впишите сюда результаты своего опыта
t_holodnaya = 25 * 60      # время полного окрашивания в холодной воде, с
t_goryachaya = 6 * 60      # то же в горячей воде, с
vysota = 0.08              # высота слоя воды, м

print(f"В холодной воде диффузия заняла {t_holodnaya / 60:.0f} мин")
print(f"В горячей воде  — {t_goryachaya / 60:.0f} мин")
print(f"Горячая вода быстрее в {t_holodnaya / t_goryachaya:.1f} раза")
print()
print(f"Средняя скорость продвижения границы окраски: "
      f"{vysota / t_goryachaya * 1000:.3f} мм/с")
````

Сравните полученную скорость со скоростью самих молекул воды. Она составляет около
600 м/с, то есть 600 000 мм/с. Разница достигает сотен тысяч раз.

Молекула движется быстро, а продвигается медленно. Противоречия здесь нет: именно
это и показал численный опыт со случайным блужданием.

```{admonition} Опыт 2. Увидеть броуновское движение своими глазами
:class: opyt
Эта демонстрация названа в рабочей программе первой — и она стоит того, чтобы
```

**Оцените, за какое время запах дошёл бы через комнату, если бы работала одна
только диффузия — без сквозняков и потоков воздуха.**

Примите скорость молекулы равной 400 м/с, расстояние между двумя столкновениями —
$10^{-7}$ м, ширину комнаты — 5 метров.

Подсказка. Вспомните главный вывод про случайное блуждание: расстояние растёт как
корень из числа шагов, то есть $L = \lambda\sqrt{N}$.

````python
lam = 1e-7        # длина свободного пробега, м
v = 400           # скорость молекулы, м/с
L = 5             # ширина комнаты, м

nu = v / lam                 # столкновений в секунду
N = (L / lam) ** 2           # сколько шагов нужно, чтобы уйти на L
t = N / nu                   # время

print(f"Столкновений в секунду: {nu:.0e}")
print(f"Нужно шагов: {N:.0e}")
print(f"Время: {t:.0e} с = {t / 86400:.0f} суток")
````

Больше недели.

В действительности запах распространяется за минуты, а значит, переносит его не диффузия.
Основную работу выполняет **конвекция**, то есть потоки воздуха от отопительных
приборов, от людей, от приоткрытой двери. Диффузия в чистом виде идёт крайне медленно и заметна
лишь там, где потоков нет: в стоячей жидкости и в твёрдом теле.
