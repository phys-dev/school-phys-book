## 1. Откуда взялась мысль о молекулах

Мысль о том, что все тела состоят из мельчайших частиц, появилась около двух с половиной
тысяч лет назад — задолго до любых опытов. Древнегреческий философ Демокрит рассуждал так:
если делить кусок вещества пополам снова и снова, то либо это можно делать бесконечно,
либо однажды мы дойдём до частицы, которую разделить нельзя. Такую частицу он назвал
**атомом** (греч. *atomos* — неделимый).

Это была догадка, а не научный вывод: проверить её Демокрит не мог. Учение о молекулярном
строении вещества стало наукой лишь во второй половине XIX века, когда его подтвердили
опытами. Сегодня оно называется **молекулярно-кинетической теорией** (МКТ) — «кинетической»
потому, что главную роль в ней играет движение частиц (греч. *kinetikos* — приводящий
в движение).

```{admonition} Как устроена физическая теория
:class: note
Теория в физике — это не «мнение» и не «версия». Это утверждение, из которого следуют
проверяемые предсказания, и все они до сих пор подтверждались опытом. Если хотя бы одно
предсказание не сбудется — теорию придётся менять. Поэтому дальше мы не просто выучим
три положения МКТ, а для каждого посмотрим, **какой опыт** его подтверждает.
```

## 2. Три положения МКТ

```{admonition} Основные положения молекулярно-кинетической теории
:class: important
1. **Все вещества состоят из частиц** (молекул, атомов), между которыми существуют промежутки.
2. **Все частицы вещества находятся в непрерывном хаотическом** (беспорядочном) **движении**.
3. **Частицы вещества взаимодействуют друг с другом** силами притяжения и отталкивания.
```

Три коротких утверждения — а объясняют они и то, почему нагретый воздух поднимается вверх,
и почему сахар растворяется в чае, и почему сталь прочная, а вода течёт. Весь курс физики
8 класса, по сути, — развёрнутое следствие этих трёх фраз.

## 3. Опытные подтверждения

### Первое положение: вещество состоит из частиц с промежутками между ними

Молекулы очень малы: они не видны не только невооружённым глазом, но и в оптический
микроскоп. Поэтому долгое время о них судили **косвенно** — по явлениям, которые иначе
объяснить не получается.

| Опыт или явление | Что из него следует |
|---|---|
| **Тепловое расширение**: нагретое тело увеличивается в объёме | Между частицами есть промежутки, которые могут меняться |
| **Сжатие газа** в шприце с закрытым отверстием | В газе промежутки между молекулами велики, их можно уменьшить |
| **Растворение**: 100 мл спирта + 100 мл воды дают не 200, а около 195 мл | Молекулы одного вещества занимают промежутки между молекулами другого |
| **Диффузия**: капля чернил окрашивает всю воду | Молекулы проникают в промежутки между чужими молекулами |

Опыт со спиртом и водой стоит того, чтобы его запомнить: объём буквально «теряется».
Никакого вещества при этом не исчезает — масса сохраняется точно. Просто молекулы
упаковываются плотнее, чем были.

### Второе положение: частицы непрерывно и хаотически движутся

Здесь два главных доказательства — **диффузия** и **броуновское движение**.

**Диффузия** — это самопроизвольное взаимное проникновение веществ друг в друга.
Никто не размешивает воду с чернилами — они перемешиваются сами. Единственное объяснение:
молекулы обоих веществ всё время движутся и залетают в промежутки между чужими молекулами.

Проверьте на модели: за перегородкой два газа, молекулы движутся и упруго сталкиваются.

Вот почему диффузия идёт только в одну сторону: не потому, что обратное невозможно,
а потому, что оно невероятно.

### Броуновское движение

В 1827 году английский ботаник Роберт Броун рассматривал в микроскоп взвешенные в воде
частички пыльцы и увидел, что они непрерывно дёргаются и движутся по ломаным линиям.
Сначала он решил, что это признак жизни, но потом обнаружил то же самое у частиц
растёртого камня, который жить точно не мог.

Объяснение дал только через 80 лет **Альберт Эйнштейн** (1905), а проверил на опыте
**Жан Перрен**: видимая частица получает удары со всех сторон от невидимых молекул воды.
Ударов очень много, и в среднем они уравновешивают друг друга — но лишь **в среднем**.
В какой-то момент слева частицу толкнёт чуть сильнее, чем справа, — и она сдвинется.

````python
sim.brownian()
````

Красная линия — след частицы: её положение отмечается через равные промежутки времени,
как это делал Перрен, зарисовывая наблюдения. Обратите внимание на два признака,
по которым броуновское движение узнаётся сразу:

* оно **никогда не прекращается** — сколько ни жди, частица не остановится;
* оно тем **быстрее, чем выше температура** и чем **меньше сама частица**.

Именно поэтому броуновское движение — прямое доказательство того, что тепловое движение
молекул существует и не затухает.

```{admonition} Почему броуновское движение видно только у мелких частиц
:class: dropdown
Число ударов о частицу пропорционально её площади поверхности, то есть растёт как $R^2$.
А масса частицы растёт как $R^3$. Разница между числом ударов слева и справа —
величина случайная, и она составляет примерно корень из полного числа ударов $N$,
то есть $\sqrt{N}$.

Получается, что ускорение, которое частица получает от «лишних» ударов, ведёт себя как
$a \sim \sqrt{N}/m \sim R/R^3 = R^{-2}$. Увеличили частицу в 10 раз — дрожание ослабло
в 100 раз. Поэтому пылинка размером в микрометр пляшет, а песчинка в миллиметр
уже неподвижна.
```

### Третье положение: частицы притягиваются и отталкиваются

Если бы молекулы только двигались и не взаимодействовали, любое вещество мгновенно
разлетелось бы: не было бы ни твёрдых тел, ни жидкостей. Значит, притяжение есть.

Опыты, которые это показывают:

* два свинцовых цилиндра со свежесрезанными и прижатыми друг к другу гладкими торцами
  слипаются так, что выдерживают гирю;
* капля воды не распадается, а собирается в шарик — молекулы стягивают её поверхность;
* чтобы разорвать проволоку, нужна сила, и тем большая, чем толще проволока.

А отталкивание? Оно проявляется, когда молекулы сближаются слишком сильно. Именно
поэтому твёрдые тела и жидкости почти невозможно сжать: попробуйте сжать воду
в закрытом шприце — не выйдет, хотя воздух сжимается легко.

```{admonition} Итог: одна таблица на весь урок
:class: important
| Положение МКТ | Опыт-подтверждение |
|---|---|
| 1. Частицы и промежутки между ними | тепловое расширение, сжатие газа, смешивание спирта с водой |
| 2. Непрерывное хаотическое движение | диффузия, броуновское движение |
| 3. Притяжение и отталкивание | слипание свинцовых цилиндров, прочность тел, несжимаемость жидкости |
```

## 4. Численный опыт: почему след частицы выглядит именно так

Соберём модель броуновского движения на компьютере. Правило будет всего одно: каждый шаг
частица получает толчок в случайную сторону. Больше никакой физики не закладываем —
и посмотрим, похоже ли получится на то, что видел Броун.

Разбирать код необязательно: смотрите на рисунки и на выводы под ними.

````python
import numpy as np
import matplotlib.pyplot as plt
from phys import style

style.use()
rng = np.random.default_rng(8)          # одно и то же «случайное» — для повторяемости опыта

def blujdanie(shagov, chastic=1):
    """Случайное блуждание: каждый шаг — толчок в случайную сторону."""
    ugly = rng.uniform(0, 2 * np.pi, size=(chastic, shagov))
    x = np.cumsum(np.cos(ugly), axis=1)
    y = np.cumsum(np.sin(ugly), axis=1)
    return x, y

fig, ax = plt.subplots(figsize=(5.6, 5.2))
for nomer in range(4):
    x, y = blujdanie(400)
    ax.plot(x[0], y[0], linewidth=1.2, alpha=0.9)
ax.plot(0, 0, "ko", markersize=6)
ax.set_title("Четыре броуновские частицы, вышедшие из одной точки")
ax.set_xlabel("x, условные единицы")
ax.set_ylabel("y, условные единицы")
ax.set_aspect("equal")
plt.show()
````

Получилось похоже на зарисовки Перрена — а ведь мы не закладывали ничего, кроме случайных
толчков. Значит, объяснение Эйнштейна работает.

Теперь главный вопрос: **как далеко** уходит частица за время $t$? Кажется, что раз частица
всё время движется, то и удаляться она должна пропорционально времени: за вдвое большее
время — вдвое дальше. Проверим. Запустим сразу 2000 частиц и посмотрим, на каком расстоянии
от старта они оказываются в среднем.

````python
x, y = blujdanie(500, chastic=2000)
rasstoyanie = np.sqrt(x**2 + y**2)        # расстояние каждой частицы от старта
srednee = rasstoyanie.mean(axis=0)        # усредняем по всем 2000 частицам
shagi = np.arange(1, 501)

fig, ax = style.axes("Число шагов (время)", "Среднее расстояние от старта",
                     "Частица удаляется всё медленнее")
ax.plot(shagi, srednee, label="результат моделирования")
ax.plot(shagi, 0.886 * np.sqrt(shagi), "--", label="корень из числа шагов")
ax.legend()
plt.show()

for shag in (100, 200, 400):
    print(f"после {shag:3d} шагов частица в среднем удалилась на {srednee[shag - 1]:.1f}")
````

Посмотрите на числа: время выросло в 4 раза (со 100 шагов до 400), а расстояние —
только в 2 раза. Результат неожиданный и очень важный:

```{admonition} Главный вывод
:class: important
При случайном блуждании расстояние от старта растёт **не как время, а как корень
из времени**. Чтобы уйти вдвое дальше, нужно вчетверо больше времени.
```

Вот почему чернила расходятся по стакану минутами, а не мгновенно, хотя молекулы воды
летают со скоростью сотен метров в секунду. Молекула не летит по прямой: она всё время
сталкивается с соседями, отскакивает и, по сути, топчется почти на месте.

```{admonition} Копнуть глубже: как из этого получили число Авогадро
:class: dropdown
Эйнштейн в 1905 году получил формулу для среднего квадрата смещения броуновской частицы:

$$\langle x^2 \rangle = \frac{RT}{N_A} \cdot \frac{t}{3\pi \eta r},$$

где $\eta$ — вязкость жидкости, $r$ — радиус частицы, $R$ — универсальная газовая
постоянная. Всё в правой части, кроме $N_A$, можно измерить. Значит, наблюдая под
микроскопом дрожание частички и измеряя её смещения, **можно сосчитать число Авогадро** —
узнать, сколько молекул в моле вещества, ни разу не увидев ни одной молекулы.

Именно это и проделал Жан Перрен в 1908 году, получив $N_A \approx 6{,}5 \cdot 10^{23}$
(современное значение $6{,}02 \cdot 10^{23}$). За эту работу он получил Нобелевскую
премию 1926 года, а спор о том, существуют ли атомы на самом деле, был окончательно
закрыт.
```

## 5. Физика рядом: вакуум в ускорителе

Ускоритель заряженных частиц — это кольцо или прямая труба, по которой летит пучок
электронов или протонов. Чтобы пучок не рассеивался на молекулах воздуха, из трубы
откачивают газ до **сверхвысокого вакуума**.

Возьмём не выдуманный, а самый настоящий пример. В новосибирском Академгородке
Институт ядерной физики строит **СКИФ** — Сибирский кольцевой источник фотонов.
В его основном накопительном кольце нужен вакуум $2 \cdot 10^{-9}$ Торр: именно такое
давление позволяет пучку жить в кольце как можно дольше.

```{admonition} Что такое Торр
:class: note
Это привычные вам миллиметры ртутного столба из 7 класса. Один Торр — давление
столбика ртути высотой 1 мм. Нормальное атмосферное давление равно 760 Торр,
то есть 760 мм рт. ст.
```

Вопрос: сколько молекул остаётся в таком «пустом» кубическом сантиметре?

Считать умеем уже сейчас. Опытом установлено, что при обычных условиях в одном
кубическом сантиметре воздуха находится примерно $2{,}7 \cdot 10^{19}$ молекул.
А число молекул в сосуде тем меньше, чем меньше давление: во сколько раз откачали
давление, во столько же раз стало меньше молекул.

````python
n_vozduh = 2.7e19        # молекул в 1 см³ воздуха при обычных условиях
p_atm = 760              # атмосферное давление, Торр
p_skif = 2e-9            # вакуум в накопительном кольце СКИФ, Торр

vo_skolko_raz = p_atm / p_skif        # во столько раз откачали воздух
n_skif = n_vozduh / vo_skolko_raz     # во столько же раз меньше стало молекул

print(f"Воздух откачали в {vo_skolko_raz:.1e} раз")
print(f"В 1 см³ воздуха:        {n_vozduh:.1e} молекул")
print(f"В 1 см³ кольца СКИФ:    {n_skif:.1e} молекул")
````

Даже в лучшем вакууме, который умеет делать техника, в каждом кубическом сантиметре
остаются **десятки миллионов молекул**. Полностью «убрать вещество» не получается —
это ещё одно свидетельство в пользу первого положения МКТ: вещество состоит из
отдельных частиц, и их можно пересчитать.

Зато молекул так мало, что они почти не мешают друг другу. Средняя длина свободного
пробега — расстояние, которое молекула пролетает между двумя столкновениями, —
в таком вакууме составляет сотни километров. Молекула в вакуумной камере скорее
долетит до её стенки, чем встретит другую молекулу. Именно поэтому пучок частиц может
циркулировать в кольце ускорителя часами, проходя миллиарды километров.

```{admonition} Копнуть глубже: сколько ударов держит стена вашей комнаты
:class: dropdown
Задача 5.4.1 из задачника Савченко звучит так: *оцените число молекул воздуха,
попадающих на 1 см² стены вашей комнаты за 1 секунду.*

Для оценки достаточно того, что мы уже знаем. Пусть в кубическом метре воздуха
$n = 2{,}7 \cdot 10^{25}$ молекул, а движутся они со средней скоростью около 500 м/с.
За секунду до площадки $S$ долетят молекулы из объёма $v \cdot S$, но лишь примерно
четверть из них движется в нужную сторону:

$$N \approx \frac{1}{4}\, n\, v\, S \approx
\frac{1}{4} \cdot 2{,}7 \cdot 10^{25} \cdot 500 \cdot 10^{-4}
\approx 3 \cdot 10^{23}\ \text{ударов в секунду}.$$

Триста секстиллионов ударов ежесекундно о площадку размером с ноготь. Даже
за одну миллиардную долю секунды по ней успевает ударить $3 \cdot 10^{14}$ молекул.

Теперь понятно, почему стена не дрожит, а броуновская частица дрожит: ударов так много,
что для крупного тела они идеально уравновешиваются, и остаётся ровное давление воздуха.
```

## 6. Опыт своими руками: измеряем диффузию

Численную модель мы посмотрели — теперь проверим её в реальности. Опыт простой,
но он даёт **число**, которое можно сравнить с предсказанием теории.

```{admonition} Опыт 1. Диффузия в горячей и холодной воде
:class: opyt
**Что нужно:** два одинаковых прозрачных стакана, холодная и горячая вода,
пакетик чая (или несколько кристалликов марганцовки, или капля пищевого красителя),
часы с секундной стрелкой или секундомер в телефоне, линейка.

**Что делать:**

2. Одновременно опустите в оба стакана по пакетику чая (или бросьте по кристаллику

3. Включите секундомер. Каждые 30 секунд отмечайте, до какой высоты опустилась

4. Засеките время, за которое вода окрасится равномерно.

**Что записать:** таблицу «время — высота окрашенного слоя» для обоих стаканов
```

````python
# ↓ впишите сюда результаты своего опыта
t_holodnaya = 25 * 60      # время полного окрашивания в холодной воде, с
t_goryachaya = 6 * 60      # то же в горячей воде, с
vysota = 0.08              # высота слоя воды, м

print(f"В холодной воде диффузия заняла {t_holodnaya / 60:.0f} мин")
print(f"В горячей воде  — {t_goryachaya / 60:.0f} мин")
print(f"Горячая вода быстрее в {t_holodnaya / t_goryachaya:.1f} раза")
print()
print(f"Средняя скорость продвижения границы окраски: "
      f"{vysota / t_goryachaya * 1000:.3f} мм/с")
````

**Оцените, за какое время запах дошёл бы через комнату, если бы работала одна
только диффузия — без сквозняков и потоков воздуха.**

Что для этого нужно знать: молекула летит со скоростью около 400 м/с и пролетает
между столкновениями примерно $10^{-7}$ м. Комната — 5 метров.

Подсказка: вспомните главный вывод про случайное блуждание — расстояние растёт
как корень из числа шагов, то есть $L = \lambda\sqrt{N}$.

````python
lam = 1e-7        # длина свободного пробега, м
v = 400           # скорость молекулы, м/с
L = 5             # ширина комнаты, м

nu = v / lam                 # столкновений в секунду
N = (L / lam) ** 2           # сколько шагов нужно, чтобы уйти на L
t = N / nu                   # время

print(f"Столкновений в секунду: {nu:.0e}")
print(f"Нужно шагов: {N:.0e}")
print(f"Время: {t:.0e} с = {t / 86400:.0f} суток")
````

Больше недели! А в жизни запах доходит за минуты. Значит, разносит его вовсе
не диффузия, а **конвекция** — потоки воздуха от батарей, от людей, от открытой двери.
Диффузия в чистом виде — процесс крайне медленный, и заметен он только там,
где потоков нет: в стоячей воде, в твёрдом теле.
