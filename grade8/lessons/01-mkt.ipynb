## 1. Откуда взялась мысль о молекулах

Возьмите кусочек мела и разломите пополам. Потом ещё раз. И ещё.

```{admonition} Как устроена физическая теория
Поэтому рядом с каждым положением МКТ дальше будет стоять опыт, который его

```

## 2. Три положения МКТ

```{admonition} Основные положения молекулярно-кинетической теории
:class: important
1. **Все вещества состоят из частиц** (молекул, атомов), между которыми существуют промежутки.
2. **Все частицы вещества находятся в непрерывном хаотическом** (беспорядочном) **движении**.
3. **Частицы вещества взаимодействуют друг с другом** силами притяжения и отталкивания.
```

Три фразы.

Из них следует, почему нагретый воздух поднимается вверх, почему сахар растворяется
в чае, почему сталь прочная, а вода течёт, почему зимой лопаются трубы, почему пахнет
из кухни в дальней комнате и почему в мороз бельё всё-таки сохнет. Весь курс физики
8 класса — это, по большому счёту, разворачивание трёх утверждений, которые вы только
что прочли за десять секунд.

## 3. Опытные подтверждения

### Первое положение: вещество состоит из частиц с промежутками между ними

Молекулу не разглядеть ни глазом, ни в оптический микроскоп — она слишком мала,
и тут уже ничего не поделаешь. Значит, судить о них приходится **косвенно**,
по явлениям, которые без молекул объяснить не выходит.

| Опыт или явление | Что из него следует |
|---|---|
| **Тепловое расширение**: нагретое тело увеличивается в объёме | Между частицами есть промежутки, которые могут меняться |
| **Сжатие газа** в шприце с закрытым отверстием | В газе промежутки между молекулами велики, их можно уменьшить |
| **Растворение**: 100 мл спирта + 100 мл воды дают не 200, а около 195 мл | Молекулы одного вещества занимают промежутки между молекулами другого |
| **Диффузия**: капля чернил окрашивает всю воду | Молекулы проникают в промежутки между чужими молекулами |

На опыте со спиртом и водой стоит задержаться. Пять миллилитров исчезли. Не испарились,
не вытекли — исчезли из измерений, и всё. При этом масса смеси в точности равна сумме
масс, взвесьте и убедитесь. Вещество никуда не делось, просто молекулы упаковались
плотнее, чем стояли до смешивания. Примерно как если высыпать стакан гороха в стакан
пшена: общий объём выйдет меньше двух стаканов, потому что пшено забьётся в промежутки
между горошинами.

### Второе положение: частицы непрерывно и хаотически движутся

Здесь два главных доказательства — **диффузия** и **броуновское движение**.

**Диффузия** — самопроизвольное взаимное проникновение веществ друг в друга.
Ложкой никто не работает, чернила расходятся по воде сами. Объяснение тут одно.
Молекулы обоих веществ непрерывно движутся и залетают в промежутки между чужими.

В модели ниже за перегородкой заперты два газа. Уберите её и посмотрите, что будет.
Потом попробуйте то же самое при другой температуре.

Так что диффузия идёт в одну сторону не потому, что обратное запрещено. Оно просто
невероятно редко.

### Броуновское движение

1827 год. Шотландский ботаник Роберт Броун разглядывает в микроскоп пыльцу, взвешенную
в воде, и видит странное. Частички дёргаются. Не плывут по течению, а именно дёргаются,
вычерчивая ломаные линии, и остановиться не думают.

Броун решил, что нашёл признак жизни. Потом растёр в порошок камень, посмотрел на него
в тот же микроскоп — и увидел ровно то же самое. Камень жить не мог.

Объяснение пришло только через восемьдесят лет. **Альберт Эйнштейн** в 1905 году
показал, откуда берётся это дрожание, а **Жан Перрен** подтвердил расчёты опытом.
Видимую частицу со всех сторон бомбардируют невидимые молекулы воды. Ударов очень много,
и в среднем они друг друга гасят. В среднем — но не в каждый момент. Стоит слева
толкнуть чуть сильнее, чем справа, и частица сдвинется.

````python
sim.brownian()
````

Красная линия — след частицы. Положение отмечается через равные промежутки времени,
точно так же зарисовывал свои наблюдения Перрен.

По двум признакам броуновское движение узнают сразу. Оно **никогда не прекращается** —
сколько ни жди, частица не остановится. И оно тем **заметнее, чем выше температура**
и чем **мельче сама частица**.

Так и получилось прямое доказательство. Тепловое движение молекул существует
и не затухает.

```{admonition} Почему броуновское движение видно только у мелких частиц
:class: dropdown
Число ударов о частицу пропорционально её площади поверхности, то есть растёт как $R^2$.
А масса частицы растёт как $R^3$. Разница между числом ударов слева и справа —
величина случайная, и она составляет примерно корень из полного числа ударов $N$,
то есть $\sqrt{N}$.

Получается, что ускорение, которое частица получает от «лишних» ударов, ведёт себя как
$a \sim \sqrt{N}/m \sim R/R^3 = R^{-2}$. Увеличили частицу в 10 раз — дрожание ослабло
в 100 раз. Поэтому пылинка размером в микрометр пляшет, а песчинка в миллиметр
уже неподвижна.
```

### Третье положение: частицы притягиваются и отталкиваются

Не будь притяжения между молекулами, любое вещество мгновенно разлетелось бы. Ни
твёрдых тел, ни жидкостей просто не существовало бы. Значит, притяжение есть.

Опыты, которые это показывают:

* два свинцовых цилиндра со свежесрезанными и прижатыми друг к другу гладкими торцами
  слипаются так, что выдерживают гирю;
* капля воды не распадается, а собирается в шарик — молекулы стягивают её поверхность;
* чтобы разорвать проволоку, нужна сила, и тем большая, чем толще проволока.

А отталкивание? Оно включается, когда молекулы сходятся слишком близко. Попробуйте
сжать воду в закрытом шприце — ничего не выйдет, хотя воздух в том же шприце сжимается
легко.

```{admonition} Итог: одна таблица на весь урок
:class: important
| Положение МКТ | Опыт-подтверждение |
|---|---|
| 1. Частицы и промежутки между ними | тепловое расширение, сжатие газа, смешивание спирта с водой |
| 2. Непрерывное хаотическое движение | диффузия, броуновское движение |
| 3. Притяжение и отталкивание | слипание свинцовых цилиндров, прочность тел, несжимаемость жидкости |
```

## 4. Численный опыт: почему след частицы выглядит именно так

Объяснение Эйнштейна можно проверить начисто. Зададим компьютеру одно-единственное
правило. На каждом шаге частица получает толчок в случайную сторону — и всё, больше
никакой физики. Ни молекул, ни воды, ни температуры. Только случайность.

Если получившаяся картинка окажется похожей на то, что Броун видел в микроскоп,
объяснение верное.

Код разбирать не нужно, смотрите на рисунки.

````python
import numpy as np
import matplotlib.pyplot as plt
from phys import style

style.use()
rng = np.random.default_rng(8)          # одно и то же «случайное» — для повторяемости опыта

def blujdanie(shagov, chastic=1):
    """Случайное блуждание: каждый шаг — толчок в случайную сторону."""
    ugly = rng.uniform(0, 2 * np.pi, size=(chastic, shagov))
    x = np.cumsum(np.cos(ugly), axis=1)
    y = np.cumsum(np.sin(ugly), axis=1)
    return x, y

fig, ax = plt.subplots(figsize=(5.6, 5.2))
for nomer in range(4):
    x, y = blujdanie(400)
    ax.plot(x[0], y[0], linewidth=1.2, alpha=0.9)
ax.plot(0, 0, "ko", markersize=6)
ax.set_title("Четыре броуновские частицы, вышедшие из одной точки")
ax.set_xlabel("x, условные единицы")
ax.set_ylabel("y, условные единицы")
ax.set_aspect("equal")
plt.show()
````

Похоже на зарисовки Перрена. А заложена была одна случайность.

Теперь вопрос поинтереснее. **Как далеко** частица уходит за время $t$?

Ответ вроде бы очевиден. Частица движется без остановки, стало быть, за вдвое большее
время уйдёт вдвое дальше. Проверим — запустим две тысячи частиц разом и посмотрим,
где они окажутся в среднем.

````python
x, y = blujdanie(500, chastic=2000)
rasstoyanie = np.sqrt(x**2 + y**2)        # расстояние каждой частицы от старта
srednee = rasstoyanie.mean(axis=0)        # усредняем по всем 2000 частицам
shagi = np.arange(1, 501)

fig, ax = style.axes("Число шагов (время)", "Среднее расстояние от старта",
                     "Частица удаляется всё медленнее")
ax.plot(shagi, srednee, label="результат моделирования")
ax.plot(shagi, 0.886 * np.sqrt(shagi), "--", label="корень из числа шагов")
ax.legend()
plt.show()

for shag in (100, 200, 400):
    print(f"после {shag:3d} шагов частица в среднем удалилась на {srednee[shag - 1]:.1f}")
````

Посмотрите на числа. Время выросло вчетверо (со 100 шагов до 400), а расстояние —
только в 2 раза. Результат неожиданный и очень важный:

```{admonition} Главный вывод
:class: important
При случайном блуждании расстояние от старта растёт **не как время, а как корень
из времени**. Чтобы уйти вдвое дальше, нужно вчетверо больше времени.
```

Вот вам и разгадка медлительности диффузии.

Молекулы воды летают со скоростью сотен метров в секунду — за такое время им бы стакан
пересечь тысячу раз. А чернила расходятся минутами. Всё потому, что по прямой молекула
не летит никогда: пролетела десятую долю микрона, столкнулась, отскочила, снова
столкнулась. Тысячи столкновений в секунду, и каждое сбивает с пути. В итоге частица
не путешествует, а топчется — и продвигается вперёд куда медленнее, чем можно
было ожидать от её скорости.

```{admonition} Копнуть глубже: как из этого получили число Авогадро
:class: dropdown
Эйнштейн в 1905 году получил формулу для среднего квадрата смещения броуновской частицы:

$$\langle x^2 \rangle = \frac{RT}{N_A} \cdot \frac{t}{3\pi \eta r},$$

где $\eta$ — вязкость жидкости, $r$ — радиус частицы, $R$ — универсальная газовая
постоянная. Всё в правой части, кроме $N_A$, можно измерить. Значит, наблюдая под
микроскопом дрожание частички и измеряя её смещения, **можно сосчитать число Авогадро** —
узнать, сколько молекул в моле вещества, ни разу не увидев ни одной молекулы.

Именно это и проделал Жан Перрен в 1908 году, получив $N_A \approx 6{,}5 \cdot 10^{23}$
(современное значение $6{,}02 \cdot 10^{23}$). За эту работу он получил Нобелевскую
премию 1926 года, а спор о том, существуют ли атомы на самом деле, был окончательно
закрыт.
```

## 5. Физика рядом: вакуум в ускорителе

Ускоритель — это труба, по которой летит пучок электронов или протонов. Свёрнута она
в кольцо или вытянута в линию, сейчас неважно. Важно другое. Воздуха внутри быть
не должно, иначе пучок рассеется на молекулах и погибнет, не долетев. Поэтому газ
из трубы откачивают до **сверхвысокого вакуума**.

Слово «сверхвысокий» здесь не для красоты.

В новосибирском Академгородке Институт ядерной физики строит **СКИФ** — Сибирский
кольцевой источник фотонов. Вакуум в его основном накопительном кольце должен быть
$2 \cdot 10^{-9}$ Торр; при таком давлении пучок живёт дольше всего.

```{admonition} Что такое Торр
:class: note
Это привычные вам миллиметры ртутного столба из 7 класса. Один Торр — давление
столбика ртути высотой 1 мм. Нормальное атмосферное давление равно 760 Торр,
то есть 760 мм рт. ст.
```

Вопрос: сколько молекул остаётся в таком «пустом» кубическом сантиметре?

Считать умеем уже сейчас, никакой новой физики не нужно.

Опытом установлено, что в кубическом сантиметре воздуха при обычных условиях
примерно $2{,}7 \cdot 10^{19}$ молекул. Дальше работает простая пропорция: во сколько
раз откачали давление, во столько же раз убавилось и молекул.

````python
n_vozduh = 2.7e19        # молекул в 1 см³ воздуха при обычных условиях
p_atm = 760              # атмосферное давление, Торр
p_skif = 2e-9            # вакуум в накопительном кольце СКИФ, Торр

vo_skolko_raz = p_atm / p_skif        # во столько раз откачали воздух
n_skif = n_vozduh / vo_skolko_raz     # во столько же раз меньше стало молекул

print(f"Воздух откачали в {vo_skolko_raz:.1e} раз")
print(f"В 1 см³ воздуха:        {n_vozduh:.1e} молекул")
print(f"В 1 см³ кольца СКИФ:    {n_skif:.1e} молекул")
````

Семьдесят миллионов молекул в кубическом сантиметре — и это лучшее, что умеет
современная техника.

Полностью «убрать вещество» не выходит. Сколько ни откачивай, что-то остаётся, и это,
между прочим, ещё одно свидетельство в пользу первого положения МКТ. Будь вещество
сплошным, его можно было бы убирать бесконечно, деля порции пополам. А частицы кончаются
не плавно — их просто становится всё меньше штук, и каждую можно, в принципе, сосчитать.

Зато уж молекул так мало, что друг другу они почти не мешают. Расстояние, которое
молекула пролетает между двумя столкновениями, в таком вакууме измеряется сотнями
километров: до стенки камеры она долетит скорее, чем встретит соседку. Потому пучок
и живёт в кольце часами, наматывая миллиарды километров.

```{admonition} Копнуть глубже: сколько ударов держит стена вашей комнаты
:class: dropdown
Задача 5.4.1 из задачника Савченко звучит так: *оцените число молекул воздуха,
попадающих на 1 см² стены вашей комнаты за 1 секунду.*

Для оценки достаточно того, что мы уже знаем. Пусть в кубическом метре воздуха
$n = 2{,}7 \cdot 10^{25}$ молекул, а движутся они со средней скоростью около 500 м/с.
За секунду до площадки $S$ долетят молекулы из объёма $v \cdot S$, но лишь примерно
четверть из них движется в нужную сторону:

$$N \approx \frac{1}{4}\, n\, v\, S \approx
\frac{1}{4} \cdot 2{,}7 \cdot 10^{25} \cdot 500 \cdot 10^{-4}
\approx 3 \cdot 10^{23}\ \text{ударов в секунду}.$$

Триста секстиллионов ударов ежесекундно о площадку размером с ноготь. Даже
```

## 6. Опыт своими руками: измеряем диффузию

Модель на экране — вещь хорошая, но проверять физику полагается в реальности.
Опыт ниже простой, зато даёт **число**, которое можно сравнить с предсказанием теории.

```{admonition} Опыт 1. Диффузия в горячей и холодной воде
:class: opyt
**Что нужно:** два одинаковых прозрачных стакана, холодная и горячая вода,
пакетик чая (или несколько кристалликов марганцовки, или капля пищевого красителя),
часы с секундной стрелкой или секундомер в телефоне, линейка.

**Что делать:**

1. Налейте в первый стакан холодную воду из-под крана, во второй — горячую.
   Уровень воды сделайте одинаковым и измерьте его линейкой.
2. Одновременно опустите в оба стакана по пакетику чая (или бросьте по кристаллику
   марганцовки). **Не размешивайте и не двигайте стаканы** — иначе вы будете изучать
   перемешивание, а не диффузию.
3. Включите секундомер. Каждые 30 секунд отмечайте, до какой высоты опустилась
   окрашенная область в каждом стакане.
4. Засеките время, за которое вода окрасится равномерно.

**Что записать:** таблицу «время — высота окрашенного слоя» для обоих стаканов
и итоговое время полного окрашивания.
```

```{admonition} Осторожно
:class: tb
Горячую воду наливайте из чайника, а не кипяток. Стакан должен быть термостойким.
Марганцовку берите пинцетом и не трогайте руками.
```

Теперь обработаем результаты. Подставьте **свои** числа — код пересчитает всё сам.

````python
# ↓ впишите сюда результаты своего опыта
t_holodnaya = 25 * 60      # время полного окрашивания в холодной воде, с
t_goryachaya = 6 * 60      # то же в горячей воде, с
vysota = 0.08              # высота слоя воды, м

print(f"В холодной воде диффузия заняла {t_holodnaya / 60:.0f} мин")
print(f"В горячей воде  — {t_goryachaya / 60:.0f} мин")
print(f"Горячая вода быстрее в {t_holodnaya / t_goryachaya:.1f} раза")
print()
print(f"Средняя скорость продвижения границы окраски: "
      f"{vysota / t_goryachaya * 1000:.3f} мм/с")
````

А теперь сравните полученную скорость со скоростью самих молекул воды. Она около
600 м/с, то есть 600 000 мм/с — разница выходит в сотни тысяч раз.

Молекула летит быстро. Продвигается медленно. Между этими двумя утверждениями нет

**Оцените, за какое время запах дошёл бы через комнату, если бы работала одна
только диффузия — без сквозняков и потоков воздуха.**

Примите скорость молекулы равной 400 м/с, расстояние между двумя столкновениями —
$10^{-7}$ м, ширину комнаты — 5 метров.

Подсказка. Вспомните главный вывод про случайное блуждание: расстояние растёт как
корень из числа шагов, то есть $L = \lambda\sqrt{N}$.

````python
lam = 1e-7        # длина свободного пробега, м
v = 400           # скорость молекулы, м/с
L = 5             # ширина комнаты, м

nu = v / lam                 # столкновений в секунду
N = (L / lam) ** 2           # сколько шагов нужно, чтобы уйти на L
t = N / nu                   # время

print(f"Столкновений в секунду: {nu:.0e}")
print(f"Нужно шагов: {N:.0e}")
print(f"Время: {t:.0e} с = {t / 86400:.0f} суток")
````

Больше недели.

А в жизни запах доходит за минуты — значит, разносит его не диффузия. Работает
**конвекция**: потоки воздуха от батарей, от людей, от приоткрытой двери. Диффузия
в чистом виде идёт крайне медленно и заметна только там, где потоков нет, —
в стоячей воде да в твёрдом теле.
