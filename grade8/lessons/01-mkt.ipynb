## 1. Откуда взялась мысль о молекулах

Разломим кусочек мела пополам, затем ещё раз и ещё.

Вопрос о том, долго ли можно продолжать такое деление, поставил Демокрит две с половиной
тысячи лет назад. Возможных ответов он видел два: либо деление продолжается неограниченно,
либо однажды остаётся частица, которую разделить нельзя. Такую частицу он назвал **атомом**
(по-гречески *atomos* означает «неделимый»).

Проверить данную догадку Демокрит не мог, поскольку подходящих опытов не существовало. В
течение двух тысяч лет она оставалась чистым умозрением.

Наукой учение о молекулах стало только во второй половине XIX века, когда оно было
подтверждено опытами. Данное учение называют **молекулярно-кинетической теорией**,
сокращённо МКТ. Слово «кинетическая» происходит от греческого *kinetikos*, «приводящий в
движение», и указывает на главную роль движения частиц в данной теории.

```{admonition} Как устроена физическая теория
:class: note
В обычной речи слово «теория» означает «предположение». В физике оно имеет противоположный
смысл: теорией называют утверждение, из которого следуют проверяемые предсказания, причём
все они до настоящего времени подтверждались. Если хотя бы одно предсказание не
подтвердится, теорию потребуется изменить.

По этой причине рядом с каждым положением МКТ далее приводится опыт, который его
подтверждает.
```

## 2. Три положения МКТ

```{admonition} Основные положения молекулярно-кинетической теории
:class: important
1. **Все вещества состоят из частиц** (молекул, атомов), между которыми существуют промежутки.
2. **Все частицы вещества находятся в непрерывном хаотическом** (беспорядочном) **движении**.
3. **Частицы вещества взаимодействуют друг с другом** силами притяжения и отталкивания.
```

Из всего трёх утверждений следует многое: почему нагретый воздух поднимается вверх, почему
сахар растворяется в чае, почему сталь прочна, а вода текуча, почему зимой разрывает
водопроводные трубы, почему запах из кухни доходит до дальней комнаты и почему бельё
высыхает на морозе. Весь курс физики 8 класса представляет собой развёрнутое следствие
данных трёх положений.

## 3. Опытные подтверждения

### Первое положение: вещество состоит из частиц с промежутками между ними

Молекула не видна ни невооружённым глазом, ни в оптический микроскоп, поскольку её размеры
для этого слишком малы. Судить о молекулах приходится **косвенно**, по явлениям, которые без
них объяснения не получают.

| Опыт или явление | Что из него следует |
|---|---|
| **Тепловое расширение**: нагретое тело увеличивается в объёме | Между частицами есть промежутки, которые могут меняться |
| **Сжатие газа** в шприце с закрытым отверстием | В газе промежутки между молекулами велики, их можно уменьшить |
| **Растворение**: 100 мл спирта + 100 мл воды дают не 200, а около 195 мл | Молекулы одного вещества занимают промежутки между молекулами другого |
| **Диффузия**: капля чернил окрашивает всю воду | Молекулы проникают в промежутки между чужими молекулами |

На опыте со спиртом и водой следует остановиться подробнее. Пять миллилитров объёма
исчезают. Вещество при этом не испаряется и не вытекает: масса смеси в точности равна сумме
масс исходных жидкостей, что проверяется взвешиванием. Уменьшается только объём, поскольку
молекулы упаковываются плотнее, чем были упакованы до смешивания.

Наглядной аналогией служит стакан пшена, высыпанный в стакан гороха. Смеси получается меньше
двух стаканов, поскольку мелкие зёрна занимают промежутки между крупными.

### Второе положение: частицы непрерывно и хаотически движутся

Здесь имеются два главных доказательства: **диффузия** и **броуновское движение**.

**Диффузией** называют самопроизвольное взаимное проникновение веществ друг в друга.
Размешивать чернила не требуется, поскольку они расходятся по воде самостоятельно.
Объяснение здесь единственное: молекулы обоих веществ непрерывно движутся и проникают в
промежутки между чужими молекулами.

В модели ниже за перегородкой заперты два газа. Уберите перегородку и проследите за
происходящим, затем повторите то же самое при другой температуре.

Законы механики этого не запрещают.

Каждая молекула движется по сосуду независимо от остальных, и вероятность оказаться в левой
половине у неё примерно такая же, как вероятность выпадения орла у монеты. Для одной

«орлом» окажутся сто молекул, составляет около $10^{-30}$: ожидание такого события заняло бы
больше времени, чем существует Вселенная.

Диффузия необратима не потому, что обратный процесс запрещён, а потому, что он неизмеримо
менее вероятен.

### Броуновское движение

В 1827 году шотландский ботаник Роберт Броун наблюдал в микроскоп пыльцу, взвешенную в воде.
Частицы вели себя необычно: они не плыли по течению, а совершали беспорядочные скачки,
вычерчивая ломаные линии, и данное движение не прекращалось.

Сначала Броун счёл его признаком жизни. Затем он растёр в порошок камень и рассмотрел взвесь
в том же микроскопе. Картина оказалась той же самой, хотя камень живым не являлся.

Объяснение появилось лишь через восемьдесят лет. В 1905 году **Альберт Эйнштейн** показал
происхождение данного движения, а **Жан Перрен** подтвердил его расчёты опытом.

Видимую частицу со всех сторон бомбардируют невидимые молекулы воды. Ударов очень много, и в
среднем они компенсируют друг друга, однако именно в среднем, а не в каждый отдельный
момент. Достаточно, чтобы слева удары оказались несколько сильнее, чем справа, и частица
сместится.


````python
sim.brownian()
````

Красная линия представляет собой след частицы. Положение отмечается через равные промежутки
времени, точно так же зарисовывал свои наблюдения Перрен.

Броуновское движение узнаётся по двум признакам. Оно **никогда не прекращается**, сколь бы
долго за частицей ни наблюдали, и оно тем **заметнее, чем выше температура** и чем **мельче
сама частица**.

Броуновское движение стало прямым доказательством того, что тепловое движение молекул
существует и не затухает.

```{admonition} Почему броуновское движение видно только у мелких частиц
:class: dropdown
Число ударов о частицу пропорционально её площади поверхности, то есть растёт как $R^2$,
тогда как масса частицы растёт как $R^3$. Разница между числом ударов слева и справа
является случайной величиной и составляет примерно корень из полного числа ударов $N$, то
есть $\sqrt{N}$.

Следовательно, ускорение, которое частица получает от «лишних» ударов, ведёт себя как
$a \sim \sqrt{N}/m \sim R/R^3 = R^{-2}$. При увеличении частицы в 10 раз дрожание ослабевает
в 100 раз. По этой причине пылинка размером в микрометр совершает заметные скачки, а
песчинка размером в миллиметр уже неподвижна.
```

### Третье положение: частицы притягиваются и отталкиваются

При отсутствии притяжения между молекулами любое вещество мгновенно разлетелось бы, и ни
твёрдых тел, ни жидкостей не существовало бы. Следовательно, притяжение существует.

Его демонстрируют следующие опыты:

* два свинцовых цилиндра со свежесрезанными и прижатыми друг к другу гладкими торцами
  слипаются так, что выдерживают гирю;
* капля воды не распадается, а собирается в шарик, поскольку молекулы стягивают её поверхность;
* для разрыва проволоки требуется сила, тем большая, чем толще проволока.

Отталкивание проявляется, когда молекулы сходятся слишком близко. Воду в закрытом шприце
сжать не удаётся, хотя воздух в том же шприце сжимается легко.

```{admonition} Итог: одна таблица на весь урок
:class: important
| Положение МКТ | Опыт-подтверждение |
|---|---|
| 1. Частицы и промежутки между ними | тепловое расширение, сжатие газа, смешивание спирта с водой |
| 2. Непрерывное хаотическое движение | диффузия, броуновское движение |
| 3. Притяжение и отталкивание | слипание свинцовых цилиндров, прочность тел, несжимаемость жидкости |
```

## 4. Численный опыт: почему след частицы выглядит именно так

Объяснение Эйнштейна допускает прямую проверку. Зададим в модели одно-единственное правило:
на каждом шаге частица получает толчок в случайном направлении. Ни молекул, ни воды, ни
температуры в модель не заложено, одна лишь случайность.

Если результат воспроизведёт картину, которую Броун наблюдал в микроскоп, объяснение следует
признать верным.

Разбирать код не обязательно, достаточно рассматривать рисунки и выводы под ними.


````python
import numpy as np
import matplotlib.pyplot as plt
from phys import style

style.use()
rng = np.random.default_rng(8)          # одно и то же «случайное» — для повторяемости опыта

def blujdanie(shagov, chastic=1):
    """Случайное блуждание: каждый шаг представляет собой толчок в случайную сторону."""
    ugly = rng.uniform(0, 2 * np.pi, size=(chastic, shagov))
    x = np.cumsum(np.cos(ugly), axis=1)
    y = np.cumsum(np.sin(ugly), axis=1)
    return x, y

fig, ax = plt.subplots(figsize=(5.6, 5.2))
for nomer in range(4):
    x, y = blujdanie(400)
    ax.plot(x[0], y[0], linewidth=1.2, alpha=0.9)
ax.plot(0, 0, "ko", markersize=6)
ax.set_title("Четыре броуновские частицы, вышедшие из одной точки")
ax.set_xlabel("x, условные единицы")
ax.set_ylabel("y, условные единицы")
ax.set_aspect("equal")
plt.show()
````

Результат воспроизводит зарисовки Перрена, хотя в модель заложена одна лишь случайность.

Следующий вопрос состоит в том, **как далеко** частица уходит за время $t$.

Ответ кажется очевидным: частица движется непрерывно, следовательно, за вдвое большее время
она должна уйти вдвое дальше. Проверим данное предположение на двух тысячах частиц, измерив
среднее расстояние от точки старта.


````python
x, y = blujdanie(500, chastic=2000)
rasstoyanie = np.sqrt(x**2 + y**2)        # расстояние каждой частицы от старта
srednee = rasstoyanie.mean(axis=0)        # усредняем по всем 2000 частицам
shagi = np.arange(1, 501)

fig, ax = style.axes("Число шагов (время)", "Среднее расстояние от старта",
                     "Частица удаляется всё медленнее")
ax.plot(shagi, srednee, label="результат моделирования")
ax.plot(shagi, 0.886 * np.sqrt(shagi), "--", label="корень из числа шагов")
ax.legend()
plt.show()

for shag in (100, 200, 400):
    print(f"после {shag:3d} шагов частица в среднем удалилась на {srednee[shag - 1]:.1f}")
````

Из полученных чисел видно, что время выросло вчетверо, со 100 шагов до 400, тогда как
расстояние увеличилось лишь вдвое. Данный результат неожидан и важен:

```{admonition} Главный вывод
:class: important
При случайном блуждании расстояние от старта растёт **не как время, а как корень из
времени**. Для того чтобы уйти вдвое дальше, требуется вчетверо большее время.
```

Отсюда становится понятной и медлительность диффузии.

Молекулы воды движутся со скоростью в сотни метров в секунду. Такой скорости было бы
достаточно, чтобы пересечь стакан тысячи раз за секунду. Тем не менее чернила расходятся
минутами. Причина заключается в том, что прямолинейно молекула не движется практически
никогда: пролетев около десятой доли микрометра, она сталкивается с соседней молекулой и
меняет направление. Столкновения происходят миллиарды раз в секунду, и каждое сбивает
частицу с прежнего пути. Продвижение вперёд оказывается несравнимо меньше пройденного пути.

```{admonition} Копнуть глубже: почему смещение растёт именно как корень
:class: dropdown
Численный опыт показал закон, но не доказал его. Между тем доказательство укладывается в
несколько строк и требует только формулы квадрата суммы.

Пусть частица движется по прямой и на каждом шаге сдвигается на расстояние $a$ вправо или
влево, причём оба направления равновероятны. Обозначим направление $i$-го шага через
$\xi_i$, где $\xi_i = +1$ при шаге вправо и $\xi_i = -1$ при шаге влево. После $N$ шагов
частица окажется в точке

$$x = a\,(\xi_1 + \xi_2 + \ldots + \xi_N).$$

Само по себе $x$ не даёт полезной информации: вправо и влево частица уходит одинаково часто,
и среднее значение $x$ равно нулю. Требуется величина, которая не зависит от знака.
Подходящих кандидатов два, модуль смещения $|x|$ и его квадрат $x^2$, и выбор между ними не
случаен; он рассматривается отдельно ниже. Возведём $x$ в квадрат по формуле квадрата суммы:

$$x^2 = a^2\,\bigl(\underbrace{\xi_1^2 + \xi_2^2 + \ldots + \xi_N^2}_{\text{квадраты}}
 + \underbrace{2\xi_1\xi_2 + 2\xi_1\xi_3 + \ldots}_{\text{попарные произведения}}\bigr).$$

Далее всё определяют два наблюдения.

**Квадраты.** Каждое $\xi_i$ равно $+1$ или $-1$, а квадрат и того и другого равен единице.
Следовательно, квадратов ровно $N$ штук и сумма их равна $N$.

**Попарные произведения.** Каждое произведение $\xi_i \xi_j$ также равно $+1$ или $-1$:
плюс, если оба шага направлены в одну сторону, минус, если в разные. Направления шагов не
зависят друг от друга, поэтому оба случая встречаются одинаково часто, и при усреднении по
многим частицам каждое такое слагаемое даёт ноль.

Остаются только квадраты:

$$\langle x^2 \rangle = a^2 N.$$

Число шагов пропорционально времени, $N \sim t$, откуда

$$\sqrt{\langle x^2 \rangle} = a\sqrt{N} \sim \sqrt{t}.$$

На этом вывод завершается. Следует отметить, где именно заключена физика: в утверждении о
независимости направлений шагов. Молекула действительно не сохраняет памяти о том, куда она
двигалась до столкновения, и потому попарные произведения гасятся. Если бы память
сохранялась (например, частицу сносило бы течением), слагаемые не гасились бы, и смещение
росло бы пропорционально времени, а не корню из него.
```

Доказательство можно проверить непосредственно здесь. Запустим одномерное блуждание, в
котором частица шагает по прямой вправо или влево, и проверим, действительно ли средний
квадрат смещения равен числу шагов.


````python
rng2 = np.random.default_rng(3)

print("шагов N     <x²> измеренное    N (предсказание)   √<x²>    √N")
for N in (100, 400, 1600, 6400):
    shagi = rng2.choice([-1, 1], size=(20000, N))   # 20 000 частиц по N шагов
    x = shagi.sum(axis=1)                            # где оказалась каждая
    sredniy_kvadrat = (x**2).mean()
    print(f"{N:7d} {sredniy_kvadrat:16.0f} {N:18d} {np.sqrt(sredniy_kvadrat):9.1f} {np.sqrt(N):6.1f}")
````

Совпадение наблюдается с точностью до случайного разброса. Увеличение числа шагов вчетверо
удваивает смещение, и никакие другие законы данному набору случайных толчков не
соответствуют.

:::{admonition} Почему берут именно квадрат смещения, а не модуль
:class: dropdown
Данный вопрос является закономерным. Модуль $|x|$ также не зависит от знака и также годился
бы в качестве меры удаления. Более того, он растёт по тому же закону: измерения дают
$\langle |x| \rangle \approx 0{,}8\, a\sqrt{N}$, то есть опять корень из времени.

Причин у выбора две, и обе существенны.

**Первая: квадрат раскрывается алгебраически.** Для суммы существует формула квадрата,
которая использовалась выше, тогда как для модуля суммы формулы нет. Величину $|a + b|$
нельзя выразить через $|a|$ и $|b|$, поэтому приведённый вывод в три строки с модулем не
проходит.

**Вторая, более глубокая: квадраты складываются.** Разобьём путь частицы на два одинаковых
участка. Средний квадрат смещения за весь путь окажется в точности равен сумме средних
квадратов за каждый участок. Для модуля такое свойство не выполняется.


````python
rng3 = np.random.default_rng(11)
pervye = rng3.choice([-1, 1], size=(40000, 800)).sum(axis=1)    # первые 800 шагов
vtorye = rng3.choice([-1, 1], size=(40000, 800)).sum(axis=1)    # следующие 800

print("КВАДРАТЫ")
print(f"  за первые 800 шагов:  {(pervye**2).mean():8.1f}")
print(f"  за вторые 800 шагов:  {(vtorye**2).mean():8.1f}")
print(f"  сумма:                {(pervye**2).mean() + (vtorye**2).mean():8.1f}")
print(f"  за все 1600 шагов:    {((pervye + vtorye)**2).mean():8.1f}   ← то же самое")
print()
print("МОДУЛИ")
print(f"  за первые 800 шагов:  {abs(pervye).mean():8.1f}")
print(f"  за вторые 800 шагов:  {abs(vtorye).mean():8.1f}")
print(f"  сумма:                {abs(pervye).mean() + abs(vtorye).mean():8.1f}")
print(f"  за все 1600 шагов:    {abs(pervye + vtorye).mean():8.1f}   ← совсем другое")
````

Складываемость квадратов не является случайным совпадением, а представляет собой следствие
того, что попарные произведения гасятся. Для независимых участков пути это выполняется
всегда, и именно поэтому средний квадрат растёт строго пропорционально числу шагов.

Существует и третья причина, физическая. Квадрат смещения входит в определение коэффициента
диффузии, а квадрат скорости входит в кинетическую энергию. Величины, с которыми природа
имеет дело в тепловых явлениях, оказываются связаны именно с квадратами, а не с модулями.
:::

```{admonition} Копнуть глубже: как из этого получили число Авогадро
:class: dropdown
Эйнштейн в 1905 году получил формулу для среднего квадрата смещения броуновской частицы:

$$\langle x^2 \rangle = \frac{RT}{N_A} \cdot \frac{t}{3\pi \eta r},$$

где $\eta$ — вязкость жидкости, $r$ — радиус частицы, $R$ — универсальная газовая
постоянная. Все величины в правой части, кроме $N_A$, можно измерить. Следовательно,
наблюдая под микроскопом дрожание частички и измеряя её смещения, **можно сосчитать число
Авогадро**, то есть узнать, сколько молекул содержится в моле вещества, ни разу не увидев ни
одной молекулы.

Именно это и проделал Жан Перрен в 1908 году, получив $N_A \approx 6{,}5 \cdot 10^{23}$
(современное значение $6{,}02 \cdot 10^{23}$). За данную работу он получил Нобелевскую
премию 1926 года, а спор о том, существуют ли атомы в действительности, был окончательно
закрыт.
```

## 5. Физика рядом: вакуум в ускорителе

Ускоритель заряженных частиц представляет собой трубу, по которой движется пучок электронов
или протонов; свёрнута она в кольцо или вытянута в линию, для дальнейшего несущественно.
Существенно другое: воздуха внутри быть не должно. Столкновения с молекулами остаточного
газа рассеивают пучок, и он гибнет, не выполнив своей задачи. По этой причине газ из трубы
откачивают до **сверхвысокого вакуума**.

Насколько высокого, показывает конкретный пример.

В наукограде Кольцово под Новосибирском строится **СКИФ** (Сибирский кольцевой источник
фотонов), для которого Институт ядерной физики изготавливает ускорительное оборудование.
Давление в его основном накопительном кольце должно составлять $2 \cdot 10^{-9}$ Торр; при
таком вакууме время жизни пучка максимально.

```{admonition} Что такое Торр
:class: note
Это известные из курса 7 класса миллиметры ртутного столба. Один Торр представляет собой
давление столбика ртути высотой 1 мм. Нормальное атмосферное давление равно 760 Торр, то
есть 760 мм рт. ст.
```

Определим, сколько молекул остаётся в таком «пустом» кубическом сантиметре.

Для расчёта достаточно уже известных сведений, новой физики не требуется.

Опытом установлено, что в кубическом сантиметре воздуха при обычных условиях содержится
примерно $2{,}7 \cdot 10^{19}$ молекул. Число молекул в единице объёма пропорционально
давлению: во сколько раз понижено давление, во столько же раз меньше остаётся молекул.


````python
n_vozduh = 2.7e19        # молекул в 1 см³ воздуха при обычных условиях
p_atm = 760              # атмосферное давление, Торр
p_skif = 2e-9            # вакуум в накопительном кольце СКИФ, Торр

vo_skolko_raz = p_atm / p_skif        # во столько раз откачали воздух
n_skif = n_vozduh / vo_skolko_raz     # во столько же раз меньше стало молекул

print(f"Воздух откачали в {vo_skolko_raz:.1e} раз")
print(f"В 1 см³ воздуха:        {n_vozduh:.1e} молекул")
print(f"В 1 см³ кольца СКИФ:    {n_skif:.1e} молекул")
````

Даже лучшая современная вакуумная техника оставляет в кубическом сантиметре около семидесяти
миллионов молекул.

Полностью удалить вещество не удаётся: сколько бы ни откачивали газ, часть молекул остаётся.
Это служит ещё одним доводом в пользу первого положения МКТ. Если бы вещество было сплошным,
его можно было бы разрежать неограниченно, деля оставшуюся порцию пополам. Однако частицы
дискретны, и число их убывает поштучно, так что каждую в принципе можно сосчитать.

При такой концентрации молекулы практически не мешают друг другу. Средняя длина свободного
пробега, то есть расстояние между двумя последовательными столкновениями молекулы, достигает
здесь десятков километров. До стенки камеры молекула долетает раньше, чем встречает другую
молекулу. Благодаря этому пучок сохраняется в кольце часами, проходя за это время миллиарды
километров.

```{admonition} Копнуть глубже: сколько ударов держит стена вашей комнаты
:class: dropdown
Задача 5.4.1 из задачника Савченко формулируется так: *оцените число молекул воздуха,
попадающих на 1 см² стены вашей комнаты за 1 секунду.*

Для оценки достаточно уже известных сведений. Пусть в кубическом метре воздуха содержится
$n = 2{,}7 \cdot 10^{25}$ молекул, а движутся они со средней скоростью около 500 м/с. За
секунду до площадки $S$ долетят молекулы из объёма $v \cdot S$, однако лишь примерно
четверть из них движется в нужную сторону:

$$N \approx \frac{1}{4}\, n\, v\, S \approx
\frac{1}{4} \cdot 2{,}7 \cdot 10^{25} \cdot 500 \cdot 10^{-4}
\approx 3 \cdot 10^{23}\ \text{ударов в секунду}.$$

Триста секстиллионов ударов ежесекундно приходится на площадку размером с ноготь. Даже за
одну миллиардную долю секунды по ней успевает ударить $3 \cdot 10^{14}$ молекул.

Отсюда видно, почему стена не дрожит, тогда как броуновская частица дрожит. Ударов настолько
много, что для крупного тела они полностью уравновешиваются, и остаётся ровное давление
воздуха.
```

## 6. Опыт своими руками: измеряем диффузию

Численная модель показала, чего следует ожидать. Далее предсказание проверяется на опыте.
Опыт несложен, однако даёт **число**, которое можно сопоставить с теорией.

```{admonition} Опыт 1. Диффузия в горячей и холодной воде
:class: opyt
**Что нужно:** два одинаковых прозрачных стакана, холодная и горячая вода, пакетик чая (или
несколько кристалликов марганцовки, или капля пищевого красителя), часы с секундной стрелкой
или секундомер в телефоне, линейка.

**Что делать:**

1. Налейте в первый стакан холодную воду из-под крана, во второй горячую.
   Уровень воды сделайте одинаковым и измерьте его линейкой.
2. Одновременно опустите в оба стакана по пакетику чая (или бросьте по кристаллику
   марганцовки). **Не размешивайте и не двигайте стаканы**, иначе изучаться будет
   перемешивание, а не диффузия.
3. Включите секундомер. Каждые 30 секунд отмечайте, до какой высоты опустилась
   окрашенная область в каждом стакане.
4. Засеките время, за которое вода окрасится равномерно.

**Что записать:** таблицу «время, высота окрашенного слоя» для обоих стаканов и итоговое
время полного окрашивания.
```

```{admonition} Осторожно
:class: tb
Горячую воду наливайте из чайника, а не кипяток. Стакан должен быть термостойким.
Марганцовку берите пинцетом и не трогайте руками.
```

Далее следует обработать результаты. Впишите **свои** числа, и расчёт выполнится
автоматически.


````python
# ↓ впишите сюда результаты своего опыта
t_holodnaya = 25 * 60      # время полного окрашивания в холодной воде, с
t_goryachaya = 6 * 60      # то же в горячей воде, с
vysota = 0.08              # высота слоя воды, м

print(f"В холодной воде диффузия заняла {t_holodnaya / 60:.0f} мин")
print(f"В горячей воде диффузия заняла {t_goryachaya / 60:.0f} мин")
print(f"Горячая вода быстрее в {t_holodnaya / t_goryachaya:.1f} раза")
print()
print(f"Средняя скорость продвижения границы окраски: "
      f"{vysota / t_goryachaya * 1000:.3f} мм/с")
````

Сравним полученную скорость со скоростью самих молекул воды. Она составляет около 600 м/с,
то есть 600 000 мм/с. Разница составляет миллионы раз.

Молекула движется быстро, а продвигается медленно. Противоречия здесь нет: именно это и
показал численный опыт со случайным блужданием.

```{admonition} Опыт 2. Увидеть броуновское движение своими глазами
:class: opyt
Данная демонстрация названа в рабочей программе первой, и показать её следует. Требуется
школьный микроскоп с увеличением не менее чем в 300 раз.

**Что обсудить:** капельки жира не являются живыми. Что же их толкает? Почему движение не

```

$10^{-7}$ м, ширину комнаты равной 5 метрам.

Подсказка. Следует вспомнить главный вывод о случайном блуждании: расстояние растёт как
корень из числа шагов, то есть $L = \lambda\sqrt{N}$.


````python
lam = 1e-7        # длина свободного пробега, м
v = 400           # скорость молекулы, м/с
L = 5             # ширина комнаты, м

nu = v / lam                 # столкновений в секунду
N = (L / lam) ** 2           # сколько шагов нужно, чтобы уйти на L
t = N / nu                   # время

print(f"Столкновений в секунду: {nu:.0e}")
print(f"Нужно шагов: {N:.0e}")
print(f"Время: {t:.0e} с = {t / 86400:.0f} суток")
````

Полученное время составляет около недели.

В действительности запах распространяется за минуты, следовательно, переносит его не
диффузия. Основную работу выполняет **конвекция**, то есть потоки воздуха от отопительных
приборов, от людей, от приоткрытой двери. Диффузия в чистом виде идёт крайне медленно и
заметна лишь там, где потоков нет: в стоячей жидкости и в твёрдом теле.
