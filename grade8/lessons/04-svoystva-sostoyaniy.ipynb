## 2. Разбираем свойства по очереди

### Почему газ занимает весь сосуд

Молекулы газа далеко друг от друга — в воздухе примерно в 11 раз дальше собственного
размера. На таком расстоянии притяжение практически не действует: удержать молекулы
вместе нечему. А движутся они быстро и беспорядочно. Значит, ничто не мешает молекуле
улететь в любую сторону, пока она не наткнётся на стенку.

**Вывод:** газ не сохраняет ни формы, ни объёма и занимает всё, что ему предоставили.

### Почему жидкость сохраняет объём, но не форму

Молекулы жидкости стоят вплотную — притяжение велико и не даёт им разлететься.
Поэтому объём сохраняется. Но молекула не привязана к одному месту: поколебавшись
рядом с соседями, она перескакивает дальше.

**Вывод:** объём есть, формы нет — слои свободно скользят друг по другу, жидкость течёт
и принимает форму сосуда.

### Почему твёрдое тело сохраняет и форму, и объём

В твёрдом теле молекулы сидят в узлах решётки и только колеблются около своих мест,
не покидая их. Стоит попытаться раздвинуть молекулы — возникает притяжение,
стоит сблизить — отталкивание.

**Вывод:** и форма, и объём сохраняются.

### Почему газ легко сжать, а жидкость и твёрдое тело — нет

Сжать вещество — значит сблизить молекулы. В газе между ними много пустого места,
и сжатие просто уменьшает промежутки. В жидкости и твёрдом теле молекулы уже стоят
на расстоянии $r_0$, где отталкивание резко растёт при любом сближении.

**Вывод:** газ сжимается легко, жидкость и твёрдое тело — практически нет.

```{admonition} Сводка: одно и то же вещество, разные свойства
:class: important
| Свойство | Газ | Жидкость | Твёрдое тело |
|---|---|---|---|
| Сохраняет объём | нет | **да** | **да** |
| Сохраняет форму | нет | нет | **да** |
| Легко сжимается | **да** | нет | нет |
| Течёт | **да** | **да** | нет |
| Причина | молекулы далеко и почти не взаимодействуют | молекулы вплотную, но перескакивают | молекулы в узлах решётки, только колеблются |
```

## 4. Численный опыт: во сколько раз газ «пустее» жидкости

Посчитаем, какую долю объёма занимают сами молекулы в воде и в воздухе.

````python
import numpy as np
import matplotlib.pyplot as plt
from phys import style

style.use()
m_H = 1.66e-27
razmer = 3e-10                      # характерный размер молекулы, м

def dolya_zanyatogo_obyoma(plotnost, otn_massa):
    """Какую часть объёма занимают сами молекулы."""
    m0 = otn_massa * m_H
    V0 = m0 / plotnost                          # объём, приходящийся на молекулу
    V_molekuly = razmer ** 3                    # объём самой молекулы
    return V_molekuly / V0

for nazvanie, rho, otn in [("вода", 1000, 18), ("лёд", 900, 18),
                           ("воздух", 1.29, 29), ("водяной пар (100 °C)", 0.6, 18)]:
    dolya = dolya_zanyatogo_obyoma(rho, otn)
    print(f"{nazvanie:22s} молекулы занимают {dolya * 100:6.2f} % объёма")
````

В воде и во льду молекулы занимают почти весь объём — сжимать нечего. В воздухе
на молекулы приходится меньше десятой доли процента: всё остальное — пустота.
Вот числовое выражение того, что мы объясняли словами.
