## 2. Разбираем свойства по очереди

### Почему газ занимает весь сосуд

Молекулы газа разбросаны далеко: в воздухе расстояние между ними примерно в одиннадцать
раз больше их собственного размера. Притяжение на такой дистанции почти не работает,
и удерживать молекулы нечем. Движутся же они быстро и беспорядочно. Поэтому каждая
летит в произвольном направлении, пока не столкнётся со стенкой.

**Вывод:** газ не сохраняет ни формы, ни объёма и занимает весь предоставленный объём.

### Почему жидкость сохраняет объём, но не форму

Молекулы жидкости расположены вплотную, притяжение велико и не даёт им разлететься.
Поэтому объём сохраняется. К своему месту молекула при этом не привязана: она колеблется рядом с соседями и перескакивает дальше.

**Вывод:** объём сохраняется, форма нет. Слои свободно скользят друг по другу, жидкость
течёт и принимает форму сосуда.

### Почему твёрдое тело сохраняет и форму, и объём

В твёрдом теле молекулы находятся в узлах решётки и колеблются, не покидая своих мест.
При попытке их раздвинуть включается притяжение, при попытке сблизить они отталкиваются.

**Вывод:** и форма, и объём сохраняются.

### Почему газ легко сжать, а жидкость и твёрдое тело — нет

Сжать вещество означает сблизить молекулы. В газе между ними много пустоты, которая
и убирается при сжатии. В жидкости и твёрдом теле молекулы уже находятся на расстоянии
$r_0$, где любое сближение вызывает резкий рост отталкивания.

**Вывод:** газ сжимается легко, жидкость и твёрдое тело практически не сжимаются.

```{admonition} Сводка: одно и то же вещество, разные свойства
:class: important
| Свойство | Газ | Жидкость | Твёрдое тело |
|---|---|---|---|
| Сохраняет объём | нет | **да** | **да** |
| Сохраняет форму | нет | нет | **да** |
| Легко сжимается | **да** | нет | нет |
| Течёт | **да** | **да** | нет |
| Причина | молекулы далеко и почти не взаимодействуют | молекулы вплотную, но перескакивают | молекулы в узлах решётки, только колеблются |
```

## 4. Численный опыт: во сколько раз газ «пустее» жидкости

Определим, какую долю объёма занимают сами молекулы в воде и в воздухе.


````python
import numpy as np
import matplotlib.pyplot as plt
from phys import style

style.use()
m_H = 1.66e-27
razmer = 3e-10                      # характерный размер молекулы, м

def dolya_zanyatogo_obyoma(plotnost, otn_massa):
    """Какую часть объёма занимают сами молекулы."""
    m0 = otn_massa * m_H
    V0 = m0 / plotnost                          # объём, приходящийся на молекулу
    V_molekuly = razmer ** 3                    # объём самой молекулы
    return V_molekuly / V0

for nazvanie, rho, otn in [("вода", 1000, 18), ("лёд", 900, 18),
                           ("воздух", 1.29, 29), ("водяной пар (100 °C)", 0.6, 18)]:
    dolya = dolya_zanyatogo_obyoma(rho, otn)
    print(f"{nazvanie:22s} молекулы занимают {dolya * 100:6.2f} % объёма")
````

В воде и во льду молекулы занимают почти весь объём, поэтому сжимать нечего. В воздухе
на молекулы приходится меньше десятой доли процента, а всё остальное составляет пустота.
Таково числовое выражение того, что ранее объяснялось словами.

:::{admonition} Копнуть глубже: откуда берётся атмосферное давление
:class: dropdown
Проверим теорию на прочность. Давление воздуха известно из опыта и равно примерно
$10^5$ Па. Выясним, способна ли молекулярная картина дать это число, не используя
его заранее.

Давление создают удары молекул о стенку. Пусть в единице объёма их $n$ штук, масса
каждой $m$, скорость $v$. Точный подсчёт даёт формулу

$$p = \frac{1}{3}\,n\,m\,v^2 .$$

Тройка в знаменателе появляется потому, что движение идёт по трём направлениям,
а на выбранную стенку давит только одно из них. Все три величины справа известны:
число молекул в кубометре вычислялось на предыдущих уроках, масса молекулы воздуха равна
29 массам водорода, скорость теплового движения при комнатной температуре составляет
около 500 м/с.


````python
n = 2.5e25            # молекул в кубометре воздуха
m = 29 * 1.66e-27     # масса молекулы воздуха, кг
v = 500               # скорость теплового движения, м/с

p = n * m * v**2 / 3

print(f"Расчёт по молекулярной картине: {p:.2e} Па")
print(f"Измеренное атмосферное давление: 1.0e+05 Па")
````

Совпадение полное, притом что ни одна величина в расчёте к давлению отношения
не имела. Взяты число молекул, их масса и скорость их движения, а получена сила,
с которой воздух давит на ладонь.

Так и проверяют теорию: не объясняют задним числом уже известный факт,
а независимо вычисляют число, которое могло бы и не совпасть.
:::
