4. Решите в общем виде, затем подставьте числа. Проверьте единицы: должны получиться

## 2. Четыре типа задач

### Тип 1. Найти количество теплоты

**Марон, № 97.** Трансформаторное масло массой 200 г нагрели от 24 до 40 °C, затратив 5,4 кДж.
Найдите удельную теплоёмкость масла.

Здесь неизвестна $c$: $c = Q / (m \Delta t) = 5400 / (0{,}2 \cdot 16) \approx 1690$ Дж/(кг·°C).

### Тип 2. Найти массу или изменение температуры

**Марон, № 102.** Достаточно ли передать стальной болванке массой 60 кг 12,6 МДж, чтобы нагреть
её на 800 °C?

Определим, какое количество теплоты требуется: $Q = 500 \cdot 60 \cdot 800 = 24 \cdot 10^6$ Дж = 24 МДж.
Следовательно, переданной энергии недостаточно: её хватит только на нагрев
на $12{,}6 \cdot 10^6 / (500 \cdot 60) = 420$ °C.

### Тип 3. Несколько тел нагреваются вместе

**Марон, № 105.** Сколько теплоты нужно, чтобы нагреть 20 л воды в железном котле массой 5 кг
от 10 до 50 °C?

Нагреваются оба тела, на одинаковые 40 °C: $Q = 4200 \cdot 20 \cdot 40 + 460 \cdot 5 \cdot 40 =
3\,360\,000 + 92\,000 = 3\,452\,000$ Дж $\approx 3{,}45$ МДж. Вклад котла составляет менее трёх
процентов, поэтому в грубых оценках посудой пренебрегают.

### Тип 4. Тепловой баланс

**Марон, № 109** была решена на прошлом уроке: резец, вода, закалка. Схема одна: теплота,
отданная горячим телом, равна теплоте, полученной холодным.

## 3. Задачи с графиками

Учебник (упражнение 9, задание 3) и Марон (№ 70) содержат задачи, в которых данные требуется
прочитать с графика. Построим такой график и рассмотрим, как его читать.

````python
import numpy as np
import matplotlib.pyplot as plt
from phys import style

style.use()
c, m = 400, 0.5                                  # медное тело массой 500 г
Q = np.linspace(0, 40_000, 100)
t = 20 + Q / (c * m)                             # температура растёт линейно с теплотой

fig, ax = style.axes("Переданное количество теплоты Q, кДж", "Температура t, °C",
                     "График нагревания: чем круче, тем меньше теплоёмкость тела")
ax.plot(Q / 1000, t)
ax.plot([0, 20], [120, 120], ":", color="gray"); ax.plot([20, 20], [20, 120], ":", color="gray")
ax.annotate("20 кДж → 120 °C", (20, 120), textcoords="offset points", xytext=(10, -14))
plt.show()
print(f"по графику: 20 кДж нагрели тело от 20 до 120 °C, значит cm = 20 000 / 100 = {20_000 / 100:.0f} Дж/°C")
print(f"для меди c = 400, откуда m = 200 / 400 = {200 / 400} кг")
````

Наклон графика равен $1/(cm)$: чем круче поднимается линия, тем меньше теплоёмкость тела.
По двум точкам графика находят $cm$, а при известном веществе и массу.


````python
from phys import quiz
quiz.reveal(
    "Марон, № 70. На одинаковых горелках греют равные массы воды, меди и железа. Как расположатся "
    "три графика температуры от времени?",
    "Горелки одинаковые, следовательно, теплота поступает с одинаковой скоростью, и ось времени можно "
    "читать как ось теплоты. Наклон каждого графика обратно пропорционален c: круче всех медь "
    "(400), чуть ниже железо (460), почти полого вода (4200). Все три прямые выходят из одной "
    "точки начальной температуры.",
)
````

## 5. Задача на оценку

**Сколько времени электрический чайник мощностью 2 кВт греет литр воды от 20 °C до кипения,
если вся его энергия идёт в воду?** Мощность 2 кВт означает, что каждую секунду передаётся
2000 Дж.


````python
quiz.numeric("Время нагревания, секунды:", answer=168, unit="с", tol=0.03,
             explain="Q = 4200 · 1 · 80 = 336 000 Дж, и при 2000 Дж/с это 336 000 / 2000 = 168 с, около трёх "
                     "минут. Настоящий чайник греет дольше, так как часть энергии уходит на нагрев самого "
                     "чайника и в воздух.")
````
