# Урок 26. Закон сохранения и превращения энергии в тепловых процессах

## 1. Механическая энергия сохраняется не всегда

В 7 классе мы формулировали закон сохранения механической энергии: сумма кинетической
и потенциальной энергий тела остаётся постоянной, если на него не действуют силы трения
и сопротивления. Оговорка существенна. Мяч, брошенный вверх, вернётся с прежней скоростью
только в пустоте; в воздухе часть энергии «пропадёт». Санки, скатившись с горы, остановятся
на ровном месте — их кинетическая энергия исчезнет.

Куда? Ответ у нас уже есть с урока 9: во внутреннюю. Санки и снег под ними нагрелись, воздух
за мячом тоже. Свинцовый шар, упавший на плиту, потерял механическую энергию — и ровно
на столько выросла внутренняя энергия шара и плиты.

## 2. Полная энергия и изолированная система

```{admonition} Закон сохранения и превращения энергии
:class: important
**Полной энергией** системы тел называют сумму их механической энергии и внутренней энергии.

Систему, на которую не действуют внешние силы ($A = 0$) и которая не участвует в теплообмене
с окружающими телами ($Q = 0$), называют **изолированной**.

Во всех явлениях, происходящих в природе, энергия не возникает из ничего и не исчезает:
она лишь **переходит от одного тела к другому или превращается из одного вида в другой**.
Полная энергия изолированной системы остаётся постоянной.
```

Если же система не изолирована, её внутренняя энергия меняется ровно на столько, сколько
ей передали теплотой и работой:

$$\Delta U = Q + A_{\text{внеш}},$$

где $Q$ — полученное количество теплоты, $A_{\text{внеш}}$ — работа внешних сил над системой.
Когда работу совершает сама система (газ толкает поршень), её внутренняя энергия на эту
работу уменьшается. Это и есть закон сохранения энергии для тепловых процессов, и все
формулы темы — его частные случаи.

## 3. Закон в действии

Пройдём по теме и увидим, что каждый урок был об этом.

| Явление | Что во что переходит |
|---|---|
| Трение, удар (урок 9) | механическая энергия → внутренняя |
| Теплопередача (уроки 10, 13) | внутренняя энергия горячего тела → внутренняя энергия холодного; $Q_{\text{отд}} = Q_{\text{пол}}$ |
| Сгорание топлива (урок 17) | химическая энергия топлива → внутренняя энергия продуктов горения |
| Плавление, испарение (уроки 18, 21) | подводимая теплота → потенциальная энергия взаимодействия молекул |
| Тепловой двигатель (уроки 24–25) | внутренняя энергия газа → механическая работа плюс теплота холодильнику: $Q_1 = A + Q_2$ |

Уравнение теплового баланса, КПД, расчёт теплоты плавления — всё это одна и та же
бухгалтерия энергии, записанная для разных случаев.

```{admonition} Кто открыл закон
:class: note
К закону сохранения энергии пришли в 1840-х годах независимо несколько учёных: немецкий
врач Роберт Майер, английский физик Джеймс Джоуль, измеривший, сколько работы нужно на
единицу теплоты, и Герман Гельмгольц, давший общую формулировку. Ломоносов за столетие
до них писал о сохранении «движения» в природе. Джоуль, между прочим, свои первые измерения
делал на пивоварне, которой владел, — там были и точные термометры, и большие ёмкости с водой.
```

## 4. Численный опыт: энергия не пропадает, а перетекает

Проследим за энергией стального шарика, брошенного с высоты на стальную плиту. Считаем, что
всё тепло от удара остаётся в шарике.

````python
import numpy as np
import matplotlib.pyplot as plt
from phys import style

style.use()
m, h, g, c = 0.1, 20, 9.8, 500              # шарик 100 г, высота 20 м, сталь
E0 = m * g * h
t_pad = np.sqrt(2 * h / g)
t = np.linspace(0, t_pad, 100)
Ep = m * g * (h - g * t**2 / 2)                # потенциальная во время падения
Ek = E0 - Ep                                    # кинетическая

fig, ax = style.axes("Время, с", "Энергия, Дж", "Падение шарика: сумма энергий постоянна, а после удара — становится внутренней")
ax.plot(t, Ep, label="потенциальная")
ax.plot(t, Ek, label="кинетическая")
ax.plot(t, Ep + Ek, "--", label="полная механическая")
ax.plot([t_pad, t_pad + 0.3], [0, E0], color="#e07a5f", label="внутренняя после удара")
ax.legend()
plt.show()
dt = E0 / (c * m)
print(f"механическая энергия {E0:.1f} Дж целиком переходит во внутреннюю; шарик нагревается на {dt:.2f} °C")
````

Шарик нагрелся на четыре десятых градуса — измерить трудно, но энергия никуда не делась.
Так что «потери» в физике — всегда потери только механической энергии; полная сохраняется.

## 6. Задачи

**Задача 1 (учебник, упражнение 11).** Два тела массой 1 кг каждое движутся навстречу
со скоростями 2 м/с, сталкиваются и останавливаются. На сколько изменилась внутренняя
энергия системы?

````python
from phys import quiz
quiz.numeric("Изменение внутренней энергии, Дж:", answer=4, unit="Дж", tol=0.02,
             explain="Кинетическая энергия каждого тела mv²/2 = 1 · 4 / 2 = 2 Дж, обоих — 4 Дж. После "
                     "остановки механической энергии нет: все 4 Дж стали внутренней энергией тел.")
````

**Задача 2 (Марон, № 18).** При теплопередаче газ получил 40 кДж и совершил работу 10 кДж.
Чему равно изменение его внутренней энергии?

````python
quiz.numeric("Изменение внутренней энергии, кДж:", answer=30, unit="кДж", tol=0.02,
             explain="Из полученных 40 кДж 10 кДж газ потратил на работу над окружающими телами; остальные "
                     "30 кДж остались в нём: ΔU = 40 − 10 = 30 кДж. Ответ сборника — 30 кДж.")
````

**Задача 3 (Марон, № 19).** Внутренняя энергия газа выросла на 100 кДж, а передано ему
400 кДж. Какую работу совершил газ?

````python
quiz.numeric("Работа газа, кДж:", answer=300, unit="кДж", tol=0.02,
             explain="A = Q − ΔU = 400 − 100 = 300 кДж.")
````

**Задача 4 (Марон, № 20).** Внутренняя энергия газа увеличилась на 300 кДж, при этом он
совершил работу 800 кДж. Сколько энергии передано газу?

````python
quiz.numeric("Передано газу, МДж:", answer=1.1, unit="МДж", tol=0.02,
             explain="Q = ΔU + A = 300 + 800 = 1100 кДж = 1,1 МДж. Ответ сборника — 1,1 МДж.")
````
