## 1. Одно вещество, три состояния

Вода бывает льдом, жидкостью и паром, и это привычно. Железо при 1500 °C плавится,
при 2860 °C кипит и превращается в газ. Воздух, наоборот, при сильном охлаждении становится жидким, а жидкий азот кипит при −196 °C, и его наливают в сосуд, как воду.

Следовательно, дело не в веществе. **Любое** вещество бывает в трёх состояниях, вопрос
только в условиях.

Свойства состояний известны ещё из курса 7 класса:

| Состояние | Сохраняет форму | Сохраняет объём | Легко ли сжать |
|---|---|---|---|
| **Твёрдое** | да | да | почти невозможно |
| **Жидкое** | нет, принимает форму сосуда | да | почти невозможно |
| **Газообразное** | нет | нет, занимает весь сосуд | легко |

```{admonition} Вопрос на подумать
:class: seealso
Во всех трёх состояниях вода состоит из **одинаковых** молекул воды. Молекула не тает, не замерзает и не испаряется, она остаётся всегда одной и той же. Возникает вопрос,
откуда берётся такая разница в свойствах.
```

## 2. Два фактора: движение и взаимодействие

О молекулах известно две вещи. Они **непрерывно движутся** и они **взаимодействуют**
друг с другом. Действуют эти два свойства прямо противоположно.

Отсюда следуют два мысленных опыта, приведённые вслед за учебником.

**Опыт первый: оставим только движение.** Пусть молекулы беспорядочно движутся, но совсем
не притягиваются. Тогда ничто не удержит их вместе, и они разлетятся по всему сосуду,
заняв весь предоставленный объём. Единственно возможным состоянием было бы
**газообразное**.

**Опыт второй: оставим только взаимодействие.** Пусть молекулы притягиваются, но не
движутся. Тогда каждая займёт своё положение равновесия и останется в нём навсегда.
Вещество было бы только **твёрдым**.

```{admonition} Главная мысль урока
:class: important
В природе присутствует и то и другое одновременно. Силы притяжения стремятся собрать
молекулы вместе, а хаотическое движение препятствует этому.

**Состояние вещества (твёрдое, жидкое или газообразное) зависит от того, какой из двух
факторов преобладает.** Управляет этим температура: чем она выше, тем быстрее движение.
```

## 3. Как взаимодействуют две молекулы

Прежде чем рассматривать всё вещество, следует разобраться с двумя молекулами. Слово «притягиваются» здесь оказывается слишком простым, так как на разных расстояниях взаимодействие происходит по-разному.

В учебнике это показано на двух тележках. Магниты изображают притяжение, пружинки
отталкивание. Пока тележки далеко, работают магниты и тянут их друг к другу. Как только
тележки сходятся и касаются пружинами, появляется отталкивание.

Меняйте расстояние ползунком и следите за стрелками сил и за точкой на графике.

````python
from phys import sim
sim.molecule_pair()
````

Из модели видно следующее:

* существует особое **расстояние равновесия $r_0$**, на котором силы притяжения
  и отталкивания равны, а суммарная сила равна нулю;
* если молекулы **ближе** $r_0$, они отталкиваются, поэтому жидкость и твёрдое тело
  почти не сжимаются;
* если **дальше** $r_0$, они притягиваются, поэтому тела не распадаются сами;
* если развести молекулы на расстояние в несколько их размеров, взаимодействие
  практически исчезает, поэтому молекулы газа почти свободны.

Расстояние $r_0$ примерно равно размеру самой молекулы, тем самым $10^{-10}$ м, о которых шла речь на предыдущем уроке.

:::{admonition} Копнуть глубже: где эта модель верна, а где нет

Закон, по которому в данной модели взаимодействуют молекулы, называется потенциалом
Леннард-Джонса. Он строился для инертных газов, у которых притяжение возникает
от лёгкого искажения электронных оболочек. У такого притяжения два параметра:
равновесное расстояние $r_0$ и глубина ямы $\varepsilon$ (энергия, которую необходимо затратить, чтобы развести две молекулы).

Если модель верна, температура плавления должна быть пропорциональна глубине ямы:

для которых модель и создавалась.

````python
print("вещество    ε/k, К   T плавления, К   отношение")
for nazvanie, eps_k, T_pl in [("неон", 36, 24.6), ("аргон", 120, 83.8),
                              ("криптон", 171, 115.8), ("ксенон", 221, 161.4)]:
    print(f"{nazvanie:10s} {eps_k:6d} {T_pl:14.1f} {T_pl / eps_k:12.2f}")
````

Отношение держится около 0,7 у всех четырёх веществ, хотя сами температуры различаются
в семь раз. Следовательно, модель работает.

Рассмотрим теперь то, о чём модель не говорит. Ртуть плавится при 234 К, лёд при 273 К,
вольфрам при 3695 К. Подставлять сюда ту же формулу нельзя, и промах вышел бы в разы.
Причина в том, что связь между частицами здесь имеет другую природу: у металлов её
создают свободные электроны, у льда водородные связи между молекулами воды. Это уже
не леннард-джонсовское притяжение, и параметры $r_0$ и $\varepsilon$ его не описывают.

Отсюда следует вывод, важный не только для данной темы. Рассмотренная модель верно передаёт **общую картину** и объясняет, почему вещество бывает твёрдым, жидким
и газообразным и почему переходы между состояниями определяются соотношением движения
и притяжения. Конкретные же числа для конкретного вещества по ней считать нельзя.

## 5. Три состояния подробно

### Газ

Расстояние между молекулами газа **намного больше их размеров**, в воздухе примерно
в 11 раз. Взаимодействие на такой дистанции почти не ощущается.

Молекула большую часть времени летит по прямой, затем сталкивается с соседней и меняет
направление. Удерживать её нечему, поэтому газ не сохраняет ни формы, ни объёма
и занимает весь предоставленный объём. По той же причине его легко сжать благодаря большому количеству пустого места между молекулами.

### Жидкость

Здесь молекулы располагаются **почти вплотную**. Притяжения достаточно, чтобы удержать
объём, но недостаточно, чтобы удержать форму.

Молекула жидкости ведёт себя следующим образом. Она несколько десятков, а то и сотен
раз колеблется около одного положения, после чего **перескакивает** на новое место,
к новым соседям. Благодаря этим перескокам жидкость и течёт.

### Твёрдое тело

Молекулы находятся на расстояниях, близких к их размерам, и колеблются около положений
равновесия, не покидая их. При попытке сблизить или раздвинуть молекулы возникают
большие силы, возвращающие их обратно. Поэтому твёрдое тело сохраняет и форму, и объём.

У большинства твёрдых веществ молекулы расположены в строгом порядке, образуя
повторяющуюся структуру, которую называют **кристаллической решёткой**. Места, где
находятся частицы, называют **узлами** решётки. Кристаллическое строение имеют
поваренная соль, лёд, алмаз, железо. Такая решётка и наблюдалась в модели при низкой температуре.

```{admonition} Сводка: сравните три состояния
:class: important
| | Газ | Жидкость | Твёрдое тело |
|---|---|---|---|
| Расстояние между молекулами | много больше размеров молекул | порядка размеров молекул | порядка размеров молекул |
| Взаимодействие | практически отсутствует | заметное | сильное |
| Движение молекул | свободный полёт между столкновениями | колебания и перескоки | колебания около узлов |
| Порядок расположения | полный беспорядок | порядок только среди ближайших соседей | строгий порядок (решётка) |
```

## 6. Численный опыт: почему газ легко сжать, а воду — нет

Оценим, сколько «пустого места» приходится на одну молекулу в воде и в воздухе.
Способ расчёта известен из предыдущего урока: объём делится на число молекул.

````python
m_H = 1.66e-27

def rasstoyanie_mezhdu_molekulami(plotnost, otnositelnaya_massa):
    """Среднее расстояние между молекулами: ребро кубика, приходящегося на одну молекулу."""
    m0 = otnositelnaya_massa * m_H         # масса молекулы
    V0 = m0 / plotnost                     # объём на одну молекулу
    return V0 ** (1 / 3)

razmer_molekuly = 3e-10                    # примерный размер молекулы, м

for nazvanie, plotnost, otn in [("вода", 1000, 18), ("воздух", 1.29, 29)]:
    d = rasstoyanie_mezhdu_molekulami(plotnost, otn)
    print(f"{nazvanie:8s}: расстояние между молекулами {d:.1e} м, "
          f"это {d / razmer_molekuly:.1f} размера молекулы")
````

Отсюда следует ответ. В воде молекулы расположены вплотную, сжимать нечего, и при
попытке сблизить их сразу возникает сильное отталкивание. В воздухе расстояние между
молекулами примерно в десять раз больше их размеров, а значит, на каждую молекулу
приходится в тысячу раз больший объём, и почти всё пространство остаётся пустым.
Оно и «выбирается» при сжатии газа.

## 7. Опыт своими руками: три состояния на столе

Модель на экране убедительна, но проверить её необходимо руками. Приведённые ниже
три опыта перечислены и в рабочей программе, поэтому их следует провести в классе.

```{admonition} Опыт 1. Почему газ сжимается, а жидкость — нет
:class: opyt
**Что нужно:** два одинаковых шприца (20 мл, без иглы), вода.

**Что делать:**

1. Наберите в первый шприц 20 мл воздуха. Плотно закройте отверстие пальцем
   и нажимайте на поршень. До какого деления удалось сжать воздух? Отпустите
   поршень и опишите, что произошло.
2. Наберите во второй шприц 20 мл воды, выгоните пузырьки воздуха. Закройте
   отверстие пальцем и нажимайте на поршень так же сильно.

**Результат:** воздух сжимается в два-три раза и упруго возвращает поршень
обратно. Вода практически не сжимается, как бы сильно ни давить на поршень.

**Почему так:** в газе расстояние между молекулами примерно в 10 раз больше их
размеров, поэтому при сжатии уменьшаются пустые промежутки. В воде молекулы
уже расположены вплотную, и при попытке сблизить их резко возрастают силы отталкивания.
```

Определим, насколько сблизились молекулы воздуха при сжатии в шприце.

````python
# ↓ результаты вашего опыта
V_do = 20.0        # было воздуха, мл
V_posle = 8.0      # стало после сжатия, мл

szhatie = V_do / V_posle
rasstoyanie = szhatie ** (1 / 3)     # объём на молекулу уменьшился во столько же раз

print(f"Воздух сжат в {szhatie:.1f} раза")
print(f"Расстояние между молекулами уменьшилось в {rasstoyanie:.2f} раза")
print(f"Было около 11 размеров молекулы, стало около {11 / rasstoyanie:.0f}")
````

Даже при сжатии воздуха в 2,5 раза молекулы сблизились лишь незначительно и по-прежнему
находятся далеко друг от друга. Поэтому воздух остаётся газом и не превращается
в жидкость.

```{admonition} Опыт 2. Обнаружить притяжение молекул

разъединить их, потянув перпендикулярно поверхностям.

**Что вы почувствуете:** мокрые стёкла разъединить намного труднее, их удерживает притяжение молекул воды в тонком слое между ними. Сдвинуть же стёкла вдоль

Данный опыт соответствует опыту из рабочей программы «обнаружение действия сил
молекулярного притяжения».
```

```{admonition} Опыт 3. Вырастить кристалл (проект на две недели)

**Что делать:**
```

**Какую долю объёма комнаты занимают сами молекулы воздуха?**

Ранее отмечалось, что газ состоит в основном из пустоты. Выясним, насколько «в основном».
Для расчёта примем, что в кубическом сантиметре воздуха содержится $2{,}7 \cdot 10^{19}$
молекул, а размер молекулы составляет $3 \cdot 10^{-10}$ м.


````python
n = 2.7e19 * 1e6          # молекул в 1 м³ воздуха
d = 3e-10                 # размер молекулы, м

V_odnoy = 4 / 3 * 3.14 * (d / 2) ** 3     # объём одной молекулы как шарика
V_vseh = V_odnoy * n                       # объём всех молекул в 1 м³

print(f"Объём одной молекулы: {V_odnoy:.1e} м³")
print(f"Все молекулы из 1 м³ воздуха занимают: {V_vseh:.1e} м³")
print(f"Это {V_vseh * 100:.2f} % объёма")
````

Получается всего четыре сотых процента. Если бы все молекулы воздуха из комнаты объёмом
60 м³ собрать вплотную, они заняли бы около **23 литров**, примерно ведро.
Вся остальная комната представляет собой пустоту, в которой они летают.

````python
V_komnaty = 60                          # объём комнаты, м³
V_molekul = V_vseh * V_komnaty          # объём всех молекул воздуха комнаты

print(f"Молекулы всего воздуха комнаты заняли бы {V_molekul * 1000:.0f} л")
````

Отсюда следуют два вывода:

* газ можно сжимать в сотни раз благодаря свободному пространству между молекулами;
* воду же сжать не удастся, так как её молекулы уже занимают почти весь объём.
