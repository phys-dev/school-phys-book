## 1. Одно вещество — три состояния

Вода бывает льдом, жидкостью и паром. Железо при 1500 °C плавится, а при 2860 °C кипит
и превращается в газ. Воздух, наоборот, при сильном охлаждении становится жидким —
жидкий азот кипит при −196 °C, и его наливают в сосуд, как воду.

Значит, дело не в веществе: **любое** вещество может быть в трёх состояниях.

Свойства состояний мы знаем ещё из 7 класса:

|---|---|---|---|
| **Твёрдое** | да | да | почти невозможно |
| **Жидкое** | нет, принимает форму сосуда | да | почти невозможно |
| **Газообразное** | нет | нет, занимает весь сосуд | легко |

## 2. Два фактора: движение и взаимодействие

Из первого урока мы знаем о молекулах две вещи: они **непрерывно движутся** и они
**взаимодействуют** друг с другом. Эти два свойства противоположны по своему действию.

Проведём два мысленных опыта — так же, как это делает учебник.

**Опыт первый: оставим только движение.** Пусть молекулы беспорядочно движутся, но совсем
не притягиваются. Тогда ничто не удержит их вместе: они разлетятся по всему сосуду,
и вещество займёт весь предоставленный объём. Единственно возможным состоянием
было бы **газообразное**.

**Опыт второй: оставим только взаимодействие.** Пусть молекулы притягиваются, но не
движутся. Тогда каждая займёт своё положение равновесия и останется в нём навсегда.
Вещество было бы только **твёрдым**.

```{admonition} Главная мысль урока
:class: important
В природе есть и то и другое одновременно. Силы притяжения стремятся собрать молекулы
вместе, а хаотическое движение мешает этому.

**Каким будет вещество — твёрдым, жидким или газообразным — зависит от того,
что из двух пересиливает.** А управляет этим температура: чем она выше,
тем быстрее движение.
```

## 3. Как взаимодействуют две молекулы

Прежде чем смотреть на всё вещество, разберёмся с двумя молекулами. Взаимодействие
устроено хитрее, чем просто «притягиваются»: на разных расстояниях всё по-разному.

В учебнике это показано на модели из двух тележек: магниты на них изображают притяжение,
а пружинки — отталкивание. Пока тележки далеко, магниты тянут их друг к другу.
Когда пружины коснутся и начнут сжиматься — появляется отталкивание.

Подвигайте ползунок и следите за стрелками сил и за точкой на графике.

````python
from phys import sim
sim.molecule_pair()
````

Что видно на модели:

* есть особое **расстояние равновесия $r_0$**, на котором силы притяжения и отталкивания
  равны, а суммарная сила равна нулю;
* если молекулы **ближе** $r_0$ — они отталкиваются (поэтому жидкость и твёрдое тело
  почти не сжимаются);
* если **дальше** $r_0$ — притягиваются (поэтому тела не распадаются сами);
* если развести молекулы на расстояние в несколько их размеров — взаимодействие
  практически исчезает (поэтому молекулы газа почти свободны).

Расстояние $r_0$ примерно равно размеру самой молекулы — тем самым $10^{-10}$ м,
которые мы обсуждали на прошлом уроке.

## 5. Три состояния подробно

### Газ

Среднее расстояние между молекулами **значительно больше их размеров** — в воздухе
примерно в 10 раз. На таких расстояниях молекулы почти не взаимодействуют.

Каждая молекула большую часть времени летит прямолинейно, а затем сталкивается с другой
и меняет направление. Поэтому газ не сохраняет ни формы, ни объёма и занимает всё,
что ему предоставили. И поэтому же газ легко сжать: между молекулами много пустого места.

### Жидкость

Молекулы расположены **близко друг к другу**, почти вплотную. Притяжение уже достаточно
велико, чтобы удержать объём, но недостаточно, чтобы удержать форму.

Вот как ведёт себя молекула жидкости: она совершает несколько десятков или сотен колебаний
около одного положения, а потом **перескакивает** на новое место и колеблется рядом
с новыми соседями. Из-за этих перескоков жидкость и течёт.

### Твёрдое тело

Молекулы находятся на расстояниях, близких к их размерам, и колеблются около положений
равновесия, не покидая их. Стоит попытаться сблизить или раздвинуть молекулы —
возникают большие силы, возвращающие их обратно. Поэтому твёрдое тело сохраняет
и форму, и объём.

У большинства твёрдых веществ молекулы расположены в строгом порядке, образуя
повторяющуюся структуру — **кристаллическую решётку**. Места, где находятся частицы,
называют **узлами** решётки. Кристаллическое строение имеют поваренная соль, лёд,
алмаз, железо. Как раз такую решётку вы и видели в модели при низкой температуре.

```{admonition} Сводка: сравните три состояния
:class: important
| | Газ | Жидкость | Твёрдое тело |
|---|---|---|---|
| Расстояние между молекулами | много больше размеров молекул | порядка размеров молекул | порядка размеров молекул |
| Взаимодействие | практически отсутствует | заметное | сильное |
| Движение молекул | свободный полёт между столкновениями | колебания и перескоки | колебания около узлов |
| Порядок расположения | полный беспорядок | порядок только среди ближайших соседей | строгий порядок — решётка |
```

## 6. Численный опыт: почему газ легко сжать, а воду — нет

Оценим, сколько «пустого места» приходится на одну молекулу в воде и в воздухе.
Считать умеем ещё с прошлого урока: делим объём на число молекул.

````python
m_H = 1.66e-27

def rasstoyanie_mezhdu_molekulami(plotnost, otnositelnaya_massa):
    """Среднее расстояние между молекулами: ребро кубика, приходящегося на одну молекулу."""
    m0 = otnositelnaya_massa * m_H         # масса молекулы
    V0 = m0 / plotnost                     # объём на одну молекулу
    return V0 ** (1 / 3)

razmer_molekuly = 3e-10                    # примерный размер молекулы, м

for nazvanie, plotnost, otn in [("вода", 1000, 18), ("воздух", 1.29, 29)]:
    d = rasstoyanie_mezhdu_molekulami(plotnost, otn)
    print(f"{nazvanie:8s}: расстояние между молекулами {d:.1e} м, "
          f"это {d / razmer_molekuly:.1f} размера молекулы")
````

Вот и ответ. В воде молекулы стоят вплотную — сжимать нечего, и при попытке сблизить
их сразу возникает сильное отталкивание. В воздухе расстояние между молекулами примерно
в десять раз больше их размеров, а значит, на каждую молекулу приходится в тысячу раз
больший объём — почти всё пространство пустое. Его-то и «выбирают», когда сжимают газ.

```{admonition} Опыт 1. Почему газ сжимается, а жидкость — нет
:class: opyt
**Что нужно:** два одинаковых шприца (20 мл, без иглы), вода.

**Что делать:**

1. Наберите в первый шприц 20 мл воздуха. Плотно закройте отверстие пальцем
   и нажимайте на поршень. До какого деления удалось сжать воздух? Отпустите
   поршень — что произошло?
2. Наберите во второй шприц 20 мл воды, выгоните пузырьки воздуха. Закройте
   отверстие пальцем и нажимайте на поршень так же сильно.

**Что получится:** воздух сжимается в два-три раза и упруго возвращает поршень
обратно. Вода не сжимается практически совсем, как ни дави.

**Почему так:** в газе расстояние между молекулами примерно в 10 раз больше их
размеров — сжимать есть что, ведь уменьшаются пустые промежутки. В воде молекулы
уже стоят вплотную, и при попытке сблизить их резко возрастают силы отталкивания.
```

Занесите результат в расчёт и посмотрите, как изменилось расстояние между молекулами
воздуха, когда вы сжали его в шприце.

````python
# ↓ результаты вашего опыта
V_do = 20.0        # было воздуха, мл
V_posle = 8.0      # стало после сжатия, мл

szhatie = V_do / V_posle
rasstoyanie = szhatie ** (1 / 3)     # объём на молекулу уменьшился во столько же раз

print(f"Воздух сжат в {szhatie:.1f} раза")
print(f"Расстояние между молекулами уменьшилось в {rasstoyanie:.2f} раза")
print(f"Было около 11 размеров молекулы, стало около {11 / rasstoyanie:.0f}")
````

Даже сжав воздух в 2,5 раза, мы лишь чуть сблизили молекулы — они по-прежнему далеко
друг от друга. Именно поэтому воздух остаётся газом и не превращается в жидкость.

```{admonition} Опыт 2. Обнаружить притяжение молекул
:class: opyt

потянув перпендикулярно поверхностям.

**Что получится:** мокрые стёкла разъединить намного труднее — их удерживает

Это и есть тот самый опыт из рабочей программы — «обнаружение действия сил
молекулярного притяжения».
```

```{admonition} Опыт 3. Вырастить кристалл (проект на две недели)

**Что делать:**
```

**Какую долю объёма комнаты занимают сами молекулы воздуха?**

Мы говорили, что газ — это в основном пустота. Но насколько «в основном»? Посчитаем:
в кубическом сантиметре воздуха $2{,}7 \cdot 10^{19}$ молекул, размер молекулы
$3 \cdot 10^{-10}$ м.

````python
n = 2.7e19 * 1e6          # молекул в 1 м³ воздуха
d = 3e-10                 # размер молекулы, м

V_odnoy = 4 / 3 * 3.14 * (d / 2) ** 3     # объём одной молекулы как шарика
V_vseh = V_odnoy * n                       # объём всех молекул в 1 м³

print(f"Объём одной молекулы: {V_odnoy:.1e} м³")
print(f"Все молекулы из 1 м³ воздуха занимают: {V_vseh:.1e} м³")
print(f"Это {V_vseh * 100:.2f} % объёма")
````

Всего четыре сотых процента! Если бы все молекулы воздуха из комнаты объёмом 60 м³
собрать вплотную, они заняли бы около **23 литров** — примерно ведро. Вся остальная
комната — пустота, в которой они летают.

````python
V_komnaty = 60                          # объём комнаты, м³
V_molekul = V_vseh * V_komnaty          # объём всех молекул воздуха комнаты

print(f"Молекулы всего воздуха комнаты заняли бы {V_molekul * 1000:.0f} л")
````

Отсюда сразу два вывода:

* газ можно сжимать в сотни раз — есть куда;
* а вот воду сжать не удастся: там молекулы уже занимают почти весь объём.
