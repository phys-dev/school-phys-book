# Урок 22. Влажность воздуха. Лабораторная работа «Определение относительной влажности воздуха»

В воздухе всегда есть водяной пар — он испаряется с океанов, рек, почвы, листьев, с нашей
кожи. От того, сколько его в воздухе, зависит, сохнет ли бельё, выпадет ли роса, легко ли
дышать в жару. Сегодня научимся измерять содержание пара в воздухе — сначала на бумаге,
потом руками.

## 1. Абсолютная и относительная влажность

```{admonition} Абсолютная влажность
:class: important
Массу водяного пара, содержащегося в 1 м³ воздуха, называют **абсолютной влажностью**
воздуха. Обозначение $\rho$, единица — г/м³ (это плотность пара).
```

Но одно и то же количество пара летом и зимой ощущается по-разному. Дело в том, что при
каждой температуре в воздухе может находиться лишь определённое наибольшее количество пара —
пар при этом насыщенный. Чем теплее, тем больше пара нужно для насыщения.

| $t$, °C | $\rho_0$, г/м³ | $t$, °C | $\rho_0$, г/м³ |
|---:|---:|---:|---:|
| −10 | 2,1 | 15 | 12,8 |
| −5 | 3,2 | 20 | 17,3 |
| 0 | 4,8 | 25 | 23,0 |
| 5 | 6,8 | 30 | 30,4 |
| 10 | 9,4 | 50 | 83,0 |

Плотность насыщенного водяного пара по таблице учебника (§ 20). При 20 °C кубометр воздуха
вмещает до 17,3 г пара, при 0 °C — только 4,8 г. Поэтому важно не количество пара само по себе,
а его отношение к предельно возможному.

```{admonition} Относительная влажность и точка росы
:class: important
**Относительной влажностью** воздуха $\varphi$ называют отношение абсолютной влажности
к плотности насыщенного пара при той же температуре, выраженное в процентах:

$$\varphi = \frac{\rho}{\rho_0} \cdot 100\,\%.$$

Температуру, при которой находящийся в воздухе пар становится насыщенным, называют
**точкой росы**. Если воздух охладить до точки росы, пар начинает конденсироваться:
выпадает роса, появляется туман, запотевают стёкла.
```

Пример. При 20 °C в кубометре воздуха 11,8 г пара (учебник, задание 13).

````python
import numpy as np
T  = np.array([-10, -5, 0, 5, 10, 15, 20, 25, 30, 50])
R0 = np.array([2.1, 3.2, 4.8, 6.8, 9.4, 12.8, 17.3, 23.0, 30.4, 83.0])
rho0 = lambda t: float(np.interp(t, T, R0))          # плотность насыщенного пара при t
tochka_rosy = lambda rho: float(np.interp(rho, R0, T))

rho = 11.8
print(f"относительная влажность: {rho / rho0(20) * 100:.0f} %")
print(f"точка росы: около {tochka_rosy(rho):.0f} °C — при такой температуре стекла начнут запотевать")
````

Человек чувствует себя комфортно при относительной влажности 40–60 %. Ниже — сохнут
слизистые и кожа, статическое электричество трещит в одежде; выше — трудно дышать в жару,
плохо сохнет бельё, сыреют стены.

## 2. Как измеряют влажность

**Конденсационный гигрометр** — металлическая коробочка с полированной стенкой, которую
охлаждают, испаряя внутри эфир. Как только стенка запотевает, термометр показывает точку
росы; по таблице находят $\rho$, а зная температуру воздуха — и $\varphi$.

**Волосной гигрометр** использует свойство обезжиренного волоса удлиняться во влажном
воздухе. Первые римские гигрометры были устроены так же — из натянутой верёвки (Марон,
№ 265).

**Психрометр** — два одинаковых термометра, сухой и влажный: резервуар влажного обёрнут
тканью, конец которой опущен в воду. С ткани испаряется вода, и влажный термометр
показывает меньше сухого. Чем суше воздух, тем интенсивнее испарение и тем больше разность
показаний. По этой разности и температуре сухого термометра относительную влажность находят
по психрометрической таблице.

|---:|---:|---:|---:|---:|---:|---:|---:|---:|---:|---:|---:|
| 6 | 100 | 86 | 73 | 60 | 47 | 35 | 23 | 10 | — | — | — |
| 8 | 100 | 87 | 75 | 63 | 51 | 40 | 28 | 18 | 7 | — | — |
| 10 | 100 | 88 | 76 | 65 | 54 | 44 | 34 | 24 | 14 | 5 | — |
| 12 | 100 | 89 | 78 | 68 | 57 | 48 | 38 | 29 | 20 | 11 | — |
| 14 | 100 | 89 | 79 | 70 | 60 | 51 | 42 | 34 | 25 | 17 | 9 |
| 16 | 100 | 90 | 81 | 71 | 62 | 54 | 46 | 37 | 30 | 22 | 15 |
| 18 | 100 | 91 | 82 | 73 | 65 | 56 | 49 | 41 | 34 | 27 | 20 |
| 20 | 100 | 91 | 83 | 74 | 66 | 59 | 51 | 44 | 37 | 30 | 24 |
| 22 | 100 | 92 | 83 | 76 | 68 | 61 | 54 | 47 | 40 | 34 | 28 |
| 24 | 100 | 92 | 84 | 77 | 69 | 62 | 56 | 49 | 43 | 37 | 31 |
| 26 | 100 | 92 | 85 | 78 | 71 | 64 | 58 | 51 | 46 | 40 | 34 |

Психрометрическая таблица учебника (таблица 6): в столбцах — разность показаний сухого
и влажного термометров, в клетках — относительная влажность в процентах.

**Приборы и материалы:** термометр, кусочек ткани (марля, бинт), сосуд с водой комнатной

```{admonition} Техника безопасности

```

### Ход работы

1. Измерьте температуру воздуха в классе — это показание сухого термометра $t$.
2. Оберните резервуар термометра тканью и смочите её водой комнатной температуры.
3. Следите за показанием: оно начнёт падать. Как только изменение прекратится, запишите
   показание влажного термометра $t_1$. Обычно на это уходит 3–5 минут; можно слегка
   помахать термометром — так делают в настоящем психрометре.
4. Найдите разность $\Delta t = t - t_1$ и по психрометрической таблице определите
   относительную влажность $\varphi$.
5. Если в кабинете есть психрометр, снимите его показания и сравните.

### Погрешности

Термометр даёт ±1 °C. Разность двух показаний имеет погрешность до 2 °C, а это по таблице —
две соседние клетки, то есть ±8–10 % влажности. Поэтому результат записывают с округлением
до пяти процентов, например «около 45 %». Это грубое измерение, и таково оно и у настоящего
психрометра.

### Таблица измерений

| $t$, °C (сухой) | $t_1$, °C (влажный) | $\Delta t$, °C | $\varphi$, % по таблице | $\varphi$, % по психрометру |
|---|---|---|---|---|
| | | | | |

### Обработка результатов

````python
import numpy as np
# психрометрическая таблица учебника: строки — t сухого, столбцы — разность 0..10 °C
psy = {6: [100,86,73,60,47,35,23,10], 8: [100,87,75,63,51,40,28,18,7], 10: [100,88,76,65,54,44,34,24,14,5],
       12: [100,89,78,68,57,48,38,29,20,11], 14: [100,89,79,70,60,51,42,34,25,17,9],
       16: [100,90,81,71,62,54,46,37,30,22,15], 18: [100,91,82,73,65,56,49,41,34,27,20],
       20: [100,91,83,74,66,59,51,44,37,30,24], 22: [100,92,83,76,68,61,54,47,40,34,28],
       24: [100,92,84,77,69,62,56,49,43,37,31], 26: [100,92,85,78,71,64,58,51,46,40,34]}

def vlazhnost(t_suh, t_vl):
    """Относительная влажность по психрометрической таблице (с интерполяцией)."""
    d = t_suh - t_vl
    rows = sorted(psy)
    t_row = min(rows, key=lambda r: abs(r - t_suh))
    stolb = psy[t_row]
    return float(np.interp(d, range(len(stolb)), stolb))

# ↓ ваши измерения
t_suh, t_vl = 21.0, 15.0
phi = vlazhnost(t_suh, t_vl)
print(f"сухой {t_suh} °C, влажный {t_vl} °C, разность {t_suh - t_vl:.0f} °C → влажность около {phi:.0f} %")
print(f"с учётом погрешности термометров: от {vlazhnost(t_suh, t_vl + 1):.0f} до {vlazhnost(t_suh, t_vl - 1):.0f} %")
norma = "в норме (40–60 %)" if 40 <= phi <= 60 else ("суховато" if phi < 40 else "слишком влажно")
print(f"для школьного класса это {norma}")
````

### Вывод

> Относительная влажность воздуха в классе составляет около … %, с учётом погрешности —
> от … до … %. Санитарной норме 40–60 % это … . Погрешность метода велика, потому что
> результат определяется разностью двух показаний термометра, а каждое известно с точностью
> до градуса.

**Задача 2 (Марон, № 268).** В 5 м³ воздуха содержится 80 г пара. Найдите абсолютную влажность.
Насыщен ли этот пар при 10 °C?

````python
quiz.reveal("Показать разбор",
            "ρ = 80 / 5 = 16 г/м³. При 10 °C насыщенный пар имеет плотность лишь 9,4 г/м³, значит, 16 г/м³ "
            "в кубометре удержаться не могут: пар не просто насыщенный — лишние 6,6 г с каждого кубометра "
            "сконденсируются в туман или росу. Ответ сборника — 16 г/м³.")
````

**Задача 3 (Марон, № 269).** Температура воздуха 18 °C, точка росы 8 °C. Найдите абсолютную
и относительную влажность.

````python
quiz.numeric("Относительная влажность, %:", answer=54, unit="%", tol=0.05,
             explain="Точка росы 8 °C означает, что пар стал бы насыщенным при 8 °C: ρ = ρ₀(8 °C) ≈ 8,4 г/м³ "
                     "(между 6,8 при 5 °C и 9,4 при 10 °C). При 18 °C насыщенный пар — около 15,4 г/м³. "
                     "φ = 8,4 / 15,4 ≈ 54 %. Ответ сборника: 8,3 г/м³ и 54 %.")
````
