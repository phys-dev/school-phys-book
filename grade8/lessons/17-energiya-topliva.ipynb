# Урок 17. Энергия топлива. Удельная теплота сгорания

Чайник, батарея отопления, двигатель автомобиля, электростанция: почти вся теплота,
которая используется в быту и технике, получена сжиганием топлива. В настоящем уроке
рассматривается, откуда берётся эта энергия и как рассчитать количество теплоты,
выделяющееся при сгорании килограмма дров или бензина.

## 1. Откуда у топлива энергия

Дрова, уголь, нефть, газ содержат углерод. При горении атомы углерода соединяются
с атомами кислорода воздуха: образуется углекислый газ, а разность энергий связей
выделяется в виде теплоты. Сжигая топливо, мы получаем энергию, запасённую когда-то
растениями из солнечного света. Таким образом, уголь и нефть представляют собой солнечный
свет, законсервированный сотни миллионов лет назад.

Различные виды топлива дают различное количество теплоты. Для их сравнения вводится
специальная величина.

```{admonition} Удельная теплота сгорания
:class: important
Физическую величину, показывающую, какое количество теплоты выделяется при полном
сгорании топлива массой 1 кг, называют **удельной теплотой сгорания** топлива.
Обозначается буквой $q$, единицей измерения служит джоуль на килограмм, Дж/кг.

При полном сгорании топлива массой $m$ выделяется количество теплоты

$$Q = q\,m.$$
```

| Топливо | $q$, 10⁷ Дж/кг | Топливо | $q$, 10⁷ Дж/кг |
|---|---:|---|---:|
| Порох | 0,38 | Древесный уголь | 3,4 |
| Дрова сухие | 1,0 | Природный газ | 4,4 |
| Торф | 1,4 | Нефть | 4,4 |
| Каменный уголь | 2,7 | Бензин | 4,6 |
| Спирт | 2,7 | Керосин | 4,6 |
| Антрацит | 3,0 | Водород | 12 |

В таблице приведены значения из учебника (§ 13). Рассмотрим смысл одного из значений:
удельная теплота сгорания бензина $4{,}6 \cdot 10^7$ Дж/кг означает, что килограмм
бензина при полном сгорании выделяет 46 миллионов джоулей, то есть 46 мегаджоулей.

Для оценки этой величины сравним её с задачей предыдущих уроков.


````python
q_benzin = 4.6e7            # Дж/кг
Q = q_benzin * 1.0          # килограмм бензина
c, dt = 4200, 80            # воду греем от 20 до 100 °C
m_vody = Q / (c * dt)
print(f"1 кг бензина даёт {Q / 1e6:.0f} МДж, чего достаточно для доведения до кипения {m_vody:.0f} кг воды")
````

Таким образом, один килограмм бензина позволяет вскипятить почти сто сорок литров воды.
Именно по этой причине цивилизация в значительной степени зависит от ископаемого топлива:
концентрация энергии в нём чрезвычайно велика.

## 2. Численный опыт: сравниваем топливо


````python
import matplotlib.pyplot as plt
from phys import style

style.use()
toplivo = {"порох": 0.38, "дрова": 1.0, "торф": 1.4, "каменный уголь": 2.7, "спирт": 2.7,
           "антрацит": 3.0, "древесный уголь": 3.4, "природный газ": 4.4, "нефть": 4.4,
           "бензин": 4.6, "керосин": 4.6, "водород": 12}
fig, ax = style.axes("Удельная теплота сгорания, 10⁷ Дж/кг", "",
                     "Сколько энергии даёт килограмм топлива", figsize=(7.5, 4.8))
ax.barh(list(toplivo), list(toplivo.values()))
plt.show()
print(f"водород даёт больше бензина в {12 / 4.6:.1f} раза, а дров в {12 / 1.0:.0f} раз")
````

Водород обладает наибольшей удельной теплотой сгорания, и его рассматривают в качестве
топлива будущего. Тем не менее до настоящего времени он не получил широкого применения
в качестве топлива, поскольку килограмм водорода при обычных условиях занимает одиннадцать
кубометров: хранить его трудно, а хранение в баллоне под давлением или в жидком виде
является дорогим и опасным. Данному вопросу посвящено задание 5 из упражнения 10 учебника.

## 3. Топливо и вода: две формулы вместе

Чаще всего теплота сгорания используется для нагревания какого-либо тела. В этом случае
совместно применяются две формулы: $Q = qm_{\text{т}}$ для топлива и $Q = cm\Delta t$
для нагреваемого тела.

**Марон, № 133.** При сгорании 1 м³ природного газа выделяется $3{,}3 \cdot 10^7$ Дж.
Хватит ли этого, чтобы нагреть 200 л воды от 10 до 60 °C, если потерь нет?


````python
Q_gaz = 3.3e7
Q_voda = 4200 * 200 * (60 - 10)
print(f"газ даёт {Q_gaz / 1e6:.1f} МДж, воде нужно {Q_voda / 1e6:.1f} МДж")
print(f"недостаточно; потребуется {Q_voda / Q_gaz:.2f} м³ газа")
````

Ответ сборника составляет 1,27 м³, что совпадает с полученным результатом.

В реальных условиях потери никогда не являются нулевыми: часть теплоты уходит с горячими
продуктами горения в трубу, часть уходит в воздух и стенки. Долю теплоты, использованную
по назначению, называют **коэффициентом полезного действия нагревателя**. У газовой плиты
он составляет около 30–40 %, у костра менее 10 %, у хорошего котла отопления выше 90 %.

## 4. Опыт своими руками

```{admonition} Опыт. Сколько энергии даёт свеча
:class: opyt
**Что нужно:** свеча, весы с точностью до 0,1 г (кухонные электронные подойдут),
жестяная банка или маленькая кастрюлька, 100 мл воды, термометр, штатив или подставка,
часы.

**Что делать:**

1. Взвесьте свечу. Налейте в банку 100 г воды, измерьте температуру.
2. Зажгите свечу и держите банку над пламенем на расстоянии 2–3 см в течение 5 минут,
   слегка помешивая воду термометром.
3. Погасите свечу, снова измерьте температуру воды и взвесьте свечу.

**Что записать:** массу свечи до и после, температуры до и после.

**Что обсудить:** сколько теплоты получила вода? Сколько парафина сгорело? Разделив
одно на другое, вы получите «удельную теплоту сгорания» парафина в вашем опыте. Табличное
значение составляет около $4{,}4 \cdot 10^7$ Дж/кг. Определите, во сколько раз полученное
значение меньше табличного и на что расходуется разница.
```

```{admonition} Осторожно
:class: tb
Свечу ставят на негорючую подставку, волосы убирают, банку держат прихваткой или
закрепляют в штативе. Опыт проводится под наблюдением взрослого.
```


````python
# ↓ ваши измерения
m_svechi_do, m_svechi_posle = 24.6, 23.9     # граммы
t_do, t_posle = 19.0, 41.0                     # °C
m_vody = 0.100                                 # кг

Q_vode = 4200 * m_vody * (t_posle - t_do)
m_sgorelo = (m_svechi_do - m_svechi_posle) / 1000
q_opyt = Q_vode / m_sgorelo
print(f"вода получила {Q_vode:.0f} Дж, сгорело {m_sgorelo * 1000:.1f} г парафина")
print(f"в опыте: q = {q_opyt:.2e} Дж/кг, это {q_opyt / 4.4e7 * 100:.0f} % от табличного")
print("остальная энергия унесена горячим воздухом и светом пламени, часть пошла на нагрев банки")
````

## 5. Задачи

**Задача 1 (учебник, упражнение 10).** Какое количество теплоты выделится при полном
сгорании 5 кг торфа?

````python
from phys import quiz
quiz.numeric("Ответ в мегаджоулях:", answer=70, unit="МДж", tol=0.02,
             explain="Q = qm = 1,4·10⁷ · 5 = 7·10⁷ Дж = 70 МДж.")
````

**Задача 2 (учебник, упражнение 10).** При полном сгорании 6 кг топлива выделилось
$2{,}76 \cdot 10^8$ Дж. Какое это топливо?

````python
quiz.choice("Какое топливо сгорело?", ["Каменный уголь", "Природный газ", "Бензин или керосин", "Водород"],
            correct=2,
            explain="q = Q/m = 2,76·10⁸ / 6 = 4,6·10⁷ Дж/кг. В таблице такое значение у бензина "
                    "и керосина, поэтому по одной лишь теплоте сгорания их различить невозможно.")
````

**Задача 3 (Марон, № 131).** Сколько керосина нужно сжечь, чтобы получить 10 МДж теплоты?
А сколько спирта?

````python
quiz.reveal("Показать разбор",
            "m = Q/q. Керосин: 10⁷ / 4,6·10⁷ ≈ 0,22 кг. Спирт: 10⁷ / 2,7·10⁷ ≈ 0,37 кг. "
            "Спирта требуется в 1,7 раза больше, поскольку ровно во столько же раз его удельная теплота "
            "сгорания меньше. Ответ сборника: 0,22 кг и 0,37 кг.")
````

том же давлении?

````python
quiz.reveal("Показать разбор",
            "Удельная теплота сгорания относится к килограмму, а в баллонах одинаков объём. Молекула "
            "бутана в 29 раз тяжелее молекулы водорода, и при том же давлении и объёме масса бутана "
            "в баллоне во столько же раз больше. Преимущество водорода по q, составляющее всего 2,5 раза, "
            "этого не компенсирует. Следовательно, сравнивать нужно теплоту на единицу объёма, а не массы.")
````
