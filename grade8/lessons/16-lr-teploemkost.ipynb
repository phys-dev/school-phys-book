```{admonition} Цель работы
:class: important
Определить удельную теплоёмкость вещества, из которого сделан металлический цилиндр,
```

**Приборы и материалы:** металлический цилиндр на нити, калориметр, измерительный цилиндр,
термометр, весы с разновесами или электронные, фильтровальная бумага или салфетка, сосуд
с горячей водой (у учителя).

```{admonition} Техника безопасности
:class: tb
Горячий цилиндр переносят только за нить, над столом, не касаясь его руками. Сосуд
с кипящей водой стоит у учителя; к нему подходят по одному. Термометр не используют
для перемешивания с силой.
```

## Идея измерения

Горячий цилиндр опускают в холодную воду. Цилиндр отдаёт теплоту $c_{\text{ц}} m_{\text{ц}}
(t_{\text{ц}} - t)$, вода получает $c_{\text{в}} m_{\text{в}} (t - t_{\text{в}})$. По уравнению
баланса они равны, и всё, кроме $c_{\text{ц}}$, измеримо:

$$c_{\text{ц}} = \frac{c_{\text{в}}\, m_{\text{в}}\,(t - t_{\text{в}})}{m_{\text{ц}}\,(t_{\text{ц}} - t)}.$$

Полученное значение сравнивают с таблицей и называют металл.

## Ход работы

1. Взвесьте цилиндр — это $m_{\text{ц}}$.
2. Налейте в калориметр 150 мл холодной воды: $m_{\text{в}} = 0{,}150$ кг. Измерьте
   её температуру $t_{\text{в}}$.
3. Опустите цилиндр в сосуд с горячей водой у учителя и подождите две-три минуты, пока
   он прогреется. Измерьте температуру горячей воды — это и есть $t_{\text{ц}}$: цилиндр
   в тепловом равновесии с ней.
4. Быстро перенесите цилиндр за нить в калориметр. Не задерживайтесь: каждая секунда
   в воздухе — потеря теплоты. Стряхивать капли не надо, но и лишней воды на цилиндре быть
   не должно.
5. Осторожно помешивая термометром, следите за температурой. Когда она перестанет расти,
   запишите $t$.

### Таблица измерений

| $m_{\text{ц}}$, кг | $m_{\text{в}}$, кг | $t_{\text{в}}$, °C | $t_{\text{ц}}$, °C | $t$, °C | $c_{\text{ц}}$, Дж/(кг·°C) | Табличное $c$ | Вещество |
|---|---|---|---|---|---|---|---|
| | | | | | | | |

Прямые измерения записывайте с погрешностью, равной цене деления прибора: массы —
по весам, температуры — по термометру, обычно ±1 °C.

## Обработка результатов

````python
c_v = 4200
# ↓ ваши измерения
m_c = 0.150      # масса цилиндра, кг
m_v = 0.150      # масса воды, кг
t_v = 20.0       # начальная температура воды, °C
t_c = 98.0       # температура цилиндра (горячей воды), °C
t = 28.0         # установившаяся температура, °C

c_c = c_v * m_v * (t - t_v) / (m_c * (t_c - t))
print(f"c = 4200 · {m_v} · ({t} − {t_v}) / ({m_c} · ({t_c} − {t})) = {c_c:.0f} Дж/(кг·°C)")

tablica = {"свинец": 140, "олово": 230, "серебро": 250, "медь": 400, "железо": 460,
           "сталь": 500, "чугун": 540, "алюминий": 920}
blizhayshee = min(tablica, key=lambda v: abs(tablica[v] - c_c))
otkl = (c_c - tablica[blizhayshee]) / tablica[blizhayshee] * 100
print(f"ближе всего {blizhayshee} ({tablica[blizhayshee]}); отклонение {otkl:+.0f} %")
````

### Откуда берётся ошибка

Программа не требует полной теории погрешностей, но одно рассуждение провести стоит:
какое из измерений портит результат сильнее всего. Проверим, как меняется $c$, если каждую
температуру сдвинуть на цену деления термометра.

````python
def c_iz(t_v, t_c, t):
    return c_v * m_v * (t - t_v) / (m_c * (t_c - t))

baza = c_iz(20.0, 98.0, 28.0)
for opis, znach in [("температура воды t_в", c_iz(21.0, 98.0, 28.0)),
                    ("температура цилиндра t_ц", c_iz(20.0, 99.0, 28.0)),
                    ("температура смеси t", c_iz(20.0, 98.0, 29.0))]:
    print(f"ошибка в 1 °C в {opis:26s} меняет c на {(znach - baza) / baza * 100:+5.0f} %")
````

Самое чувствительное место — разность $t - t_{\text{в}}$: она мала, всего несколько градусов,
и ошибка в один градус составляет от неё десятую часть. Отсюда практический вывод:
брать меньше воды или более тяжёлый цилиндр, чтобы вода нагрелась сильнее и разность
выросла.

Кроме термометра, результат занижают потери: цилиндр остывает, пока его несут, и часть
его теплоты уходит в стенки калориметра. Обе причины действуют в одну сторону — вода
получает меньше, чем цилиндр «должен» был отдать, и вычисленное $c$ выходит **меньше**
табличного. Если ваше значение ниже табличного на 5–15 %, это ожидаемо.

## Вывод
