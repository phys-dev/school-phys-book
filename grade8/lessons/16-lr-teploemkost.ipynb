```{admonition} Цель работы

Определить удельную теплоёмкость вещества, из которого сделан металлический цилиндр,

```

**Приборы и материалы:** металлический цилиндр на нити, калориметр, измерительный цилиндр,
термометр, весы с разновесами или электронные, фильтровальная бумага или салфетка, сосуд
с горячей водой (у учителя).

```{admonition} Техника безопасности
:class: tb
Горячий цилиндр переносят только за нить, над столом, не касаясь его руками. Сосуд с кипящей водой стоит у учителя, к нему подходят по одному. Термометр не используют
для перемешивания с силой.
```

## Идея измерения

Горячий цилиндр опускают в холодную воду. Цилиндр отдаёт количество теплоты
$c_{\text{ц}} m_{\text{ц}} (t_{\text{ц}} - t)$, тогда как вода получает
$c_{\text{в}} m_{\text{в}} (t - t_{\text{в}})$. Согласно уравнению теплового баланса
эти величины равны, при этом все входящие в них величины, кроме $c_{\text{ц}}$, измеримы:

$$c_{\text{ц}} = \frac{c_{\text{в}}\, m_{\text{в}}\,(t - t_{\text{в}})}{m_{\text{ц}}\,(t_{\text{ц}} - t)}.$$

Полученное значение сравнивают с табличным и определяют металл.

## Ход работы

1. Взвесьте цилиндр и запишите его массу $m_{\text{ц}}$.
2. Налейте в калориметр 150 мл холодной воды: $m_{\text{в}} = 0{,}150$ кг. Измерьте
   её температуру $t_{\text{в}}$.
3. Опустите цилиндр в сосуд с горячей водой у учителя и подождите две-три минуты, пока
   он прогреется. Измерьте температуру горячей воды: она принимается за $t_{\text{ц}}$, так как цилиндр находится с ней в тепловом равновесии.
4. Быстро перенесите цилиндр за нить в калориметр. Не задерживайтесь, каждая секунда пребывания цилиндра в воздухе означает потерю теплоты. Стряхивать капли не нужно, но лишней воды на цилиндре быть не должно.
5. Осторожно помешивая термометром, следите за температурой. Когда она перестанет расти,
   запишите $t$.

### Таблица измерений

| $m_{\text{ц}}$, кг | $m_{\text{в}}$, кг | $t_{\text{в}}$, °C | $t_{\text{ц}}$, °C | $t$, °C | $c_{\text{ц}}$, Дж/(кг·°C) | Табличное $c$ | Вещество |
|---|---|---|---|---|---|---|---|
| | | | | | | | |

Прямые измерения записывайте с погрешностью, равной цене деления прибора: массы
по шкале весов, температуры по шкале термометра, обычно ±1 °C.

## Обработка результатов

````python
c_v = 4200
# ↓ ваши измерения
m_c = 0.150      # масса цилиндра, кг
m_v = 0.150      # масса воды, кг
t_v = 20.0       # начальная температура воды, °C
t_c = 98.0       # температура цилиндра (горячей воды), °C
t = 28.0         # установившаяся температура, °C

c_c = c_v * m_v * (t - t_v) / (m_c * (t_c - t))
print(f"c = 4200 · {m_v} · ({t} − {t_v}) / ({m_c} · ({t_c} − {t})) = {c_c:.0f} Дж/(кг·°C)")

tablica = {"свинец": 140, "олово": 230, "серебро": 250, "медь": 400, "железо": 460,
           "сталь": 500, "чугун": 540, "алюминий": 920}
blizhayshee = min(tablica, key=lambda v: abs(tablica[v] - c_c))
otkl = (c_c - tablica[blizhayshee]) / tablica[blizhayshee] * 100
print(f"ближе всего {blizhayshee} ({tablica[blizhayshee]}), отклонение {otkl:+.0f} %")
````

### Откуда берётся ошибка

Программа не требует полной теории погрешностей, но одно рассуждение провести целесообразно: следует определить, какое из измерений вносит наибольшую ошибку
в результат. Проверим, как изменяется $c$, если каждую из температур сдвинуть на цену
деления термометра.

````python
def c_iz(t_v, t_c, t):
    return c_v * m_v * (t - t_v) / (m_c * (t_c - t))

baza = c_iz(20.0, 98.0, 28.0)
for opis, znach in [("температура воды t_в", c_iz(21.0, 98.0, 28.0)),
                    ("температура цилиндра t_ц", c_iz(20.0, 99.0, 28.0)),
                    ("температура смеси t", c_iz(20.0, 98.0, 29.0))]:
    print(f"ошибка в 1 °C в {opis:26s} меняет c на {(znach - baza) / baza * 100:+5.0f} %")
````

Наиболее чувствительным местом является разность $t - t_{\text{в}}$, которая мала и составляет всего несколько градусов, так что ошибка в один градус составляет
от неё десятую часть. Отсюда следует практический вывод: целесообразно брать меньше воды
или более тяжёлый цилиндр, чтобы вода нагрелась сильнее и данная разность увеличилась.

Помимо погрешности термометра, результат занижают потери теплоты: цилиндр остывает
во время переноса, а часть его теплоты уходит в стенки калориметра. Обе причины действуют
в одну сторону: вода получает меньше теплоты, чем должен был отдать цилиндр, и вычисленное
значение $c$ оказывается **меньше** табличного. Если полученное значение ниже табличного
на 5–15 %, такой результат является ожидаемым.
