# Урок 12. Количество теплоты. Удельная теплоёмкость

С этого урока тепловые процессы начинают описываться числами. Мы знаем, что при
теплопередаче энергия переходит от горячего тела к холодному. Теперь выясним, **сколько**
её переходит и от чего это зависит.

## 1. Количество теплоты

```{admonition} Количество теплоты
:class: important
Энергию, которую тело получает или теряет в процессе теплопередачи, называют
**количеством теплоты** и обозначают буквой $Q$. Единица количества теплоты в СИ, как
и любой энергии, — **джоуль** (Дж). Используют также килоджоули и мегаджоули:
1 кДж = 1000 Дж, 1 МДж = 1 000 000 Дж.
```

Количество теплоты — не запас энергии в теле, а мера того, сколько энергии **передано**.
Поэтому нельзя сказать «в чайнике 100 кДж теплоты»; можно сказать «чайнику передано
100 кДж теплоты». Запас энергии в теле — это внутренняя энергия из урока 9.

```{admonition} Калория
:class: note
До джоуля количество теплоты измеряли в калориях: одна калория нагревает 1 г воды
на 1 °C. Связь с джоулем такова: 1 кал = 4,19 Дж. Калориями до сих пор считают
энергетическую ценность пищи, только там имеют в виду килокалории: 1 ккал = 4190 Дж.
```

## 2. От чего зависит количество теплоты

Три опыта из учебника отвечают на этот вопрос.

**Масса.** Нагреем на одинаковых горелках 100 г и 200 г воды на одно и то же число
градусов. На вторую порцию уйдёт вдвое больше времени, то есть вдвое больше энергии.
Количество теплоты пропорционально массе.

**Разность температур.** Нагреем одну и ту же воду сначала на 10 °C, потом ещё на 20 °C.
Второй раз энергии потребуется вдвое больше. Количество теплоты пропорционально изменению
температуры.

**Род вещества.** Нагреем на одинаковых горелках равные массы воды и подсолнечного масла.
Масло нагреется на 1 °C заметно быстрее — ему нужно меньше энергии. Значит, количество
теплоты зависит ещё и от того, из какого вещества тело.

## 3. Удельная теплоёмкость

Последняя зависимость описывается отдельной величиной.

```{admonition} Удельная теплоёмкость
:class: important
Физическую величину, численно равную количеству теплоты, которое необходимо сообщить
телу массой 1 кг для изменения его температуры на 1 °C, называют **удельной
теплоёмкостью** вещества. Обозначение $c$, единица — джоуль на килограмм-градус,
Дж/(кг·°C).
```

Удельная теплоёмкость воды равна 4200 Дж/(кг·°C). Это означает: чтобы нагреть один
килограмм воды на один градус, нужно передать ему 4200 Дж, а при остывании килограмма
воды на градус выделится столько же.

| Вещество | $c$, Дж/(кг·°C) | Вещество | $c$, Дж/(кг·°C) |
|---|---:|---|---:|
| Золото | 180 | Стекло | 840 |
| Ртуть | 138 | Кирпич | 880 |
| Свинец | 140 | Алюминий | 920 |
| Олово | 230 | Масло подсолнечное | 1700 |
| Серебро | 250 | Лёд | 2100 |
| Медь | 400 | Керосин | 2100 |
| Цинк, латунь | 400 | Спирт | 2500 |
| Железо | 460 | Молоко | 3900 |
| Сталь | 500 | **Вода** | **4200** |
| Чугун | 540 | | |

Значения из таблицы учебника; ими и следует пользоваться в задачах. Обратите внимание
на два обстоятельства. Металлы стоят в начале списка, а вода — в самом конце: среди
обычных веществ у неё наибольшая удельная теплоёмкость. И у одного вещества в разных
агрегатных состояниях теплоёмкость разная: у воды 4200, у льда 2100.

```{admonition} Почему у воды такая большая теплоёмкость и что из этого следует
:class: note
Вода медленно нагревается и медленно остывает, запасая много энергии. Поэтому климат
у морей мягче континентального: летом море отбирает тепло у воздуха, зимой отдаёт.
По той же причине воду используют в отоплении и для охлаждения двигателей, а грелку
наполняют водой, а не песком.
```

## 4. Формула количества теплоты

Собрав все три зависимости, получаем формулу.

```{admonition} Расчёт количества теплоты
:class: important
$$Q = c\,m\,(t_2 - t_1),$$

где $c$ — удельная теплоёмкость, $m$ — масса тела, $t_1$ и $t_2$ — начальная и конечная
температуры.

При нагревании $t_2 > t_1$ и $Q > 0$: тело получает теплоту. При охлаждении $t_2 < t_1$
и $Q < 0$: тело отдаёт теплоту. Часто говорят просто о количестве теплоты, отданном при
охлаждении, и берут его по модулю.
```

Пример. Сколько теплоты нужно, чтобы нагреть 2 кг воды от 20 до 100 °C для чая?

````python
c = 4200        # удельная теплоёмкость воды, Дж/(кг·°C)
m = 2.0         # масса воды, кг
t1, t2 = 20, 100

Q = c * m * (t2 - t1)
print(f"Q = {c} · {m} · ({t2} − {t1}) = {Q:.0f} Дж = {Q / 1000:.0f} кДж")
````

Шестьсот семьдесят килоджоулей. Для сравнения: подняв мешок цемента массой 50 кг
на второй этаж, вы совершаете работу около полутора килоджоулей. Чайник за пять минут
делает в четыреста раз больше.

## 5. Численный опыт: сравниваем вещества

Одно и то же количество теплоты, 10 кДж, передадим килограмму каждого вещества
из таблицы и посмотрим, на сколько градусов оно нагреется.

````python
import matplotlib.pyplot as plt
from phys import style

style.use()
teploemkost = {"свинец": 140, "медь": 400, "сталь": 500, "стекло": 840, "алюминий": 920,
               "масло": 1700, "лёд": 2100, "спирт": 2500, "вода": 4200}
Q, m = 10_000, 1.0
nagrev = {v: Q / (c * m) for v, c in teploemkost.items()}

fig, ax = style.axes("На сколько градусов нагреется 1 кг от 10 кДж", "",
                     "Чем больше удельная теплоёмкость, тем труднее нагреть")
ax.barh(list(nagrev), list(nagrev.values()))
plt.show()
for v in ("свинец", "вода"):
    print(f"{v:8s} нагреется на {nagrev[v]:5.1f} °C")
print(f"свинец греется легче воды в {nagrev['свинец'] / nagrev['вода']:.0f} раз")
````
