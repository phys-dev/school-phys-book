## 1. Количество теплоты

```{admonition} Количество теплоты
:class: important
Энергию, которую тело получает или теряет в процессе теплопередачи, называют
**количеством теплоты** и обозначают буквой $Q$. Единицей количества теплоты в СИ, как
и любой энергии, является **джоуль** (Дж). Используются также килоджоули и мегаджоули:
1 кДж = 1000 Дж, 1 МДж = 1 000 000 Дж.
```

Количество теплоты не является запасом энергии в теле, а служит мерой энергии,
**переданной** в процессе теплопередачи. По этой причине утверждение «в чайнике
содержится 100 кДж теплоты» некорректно, тогда как утверждение «чайнику передано
100 кДж теплоты» имеет смысл. Запас энергии в теле характеризуется внутренней энергией,
введённой в уроке 9.

```{admonition} Калория
:class: note
До введения джоуля количество теплоты измерялось в калориях. Одна калория представляет
собой количество теплоты, необходимое для нагревания 1 г воды на 1 °C, и связана
с джоулем соотношением 1 кал = 4,19 Дж. В калориях до настоящего времени выражают
энергетическую ценность пищи, при этом подразумеваются килокалории: 1 ккал = 4190 Дж.
```

## 2. От чего зависит количество теплоты

Ответ на данный вопрос дают три опыта, описанные в учебнике.

**Масса.** На одинаковых горелках нагреваются 100 г и 200 г воды на одно и то же число
градусов. Для нагревания второй порции требуется вдвое большее время, следовательно,
вдвое большее количество энергии. Следовательно, количество теплоты пропорционально массе тела.

**Разность температур.** Одна и та же порция воды нагревается сначала на 10 °C, затем
ещё на 20 °C. Во втором случае требуется вдвое большее количество энергии,
следовательно, количество теплоты пропорционально изменению температуры.

**Род вещества.** На одинаковых горелках нагреваются равные массы воды и подсолнечного
масла. Масло нагревается на 1 °C заметно быстрее, так как для этого требуется меньшее количество энергии. Следовательно, количество теплоты зависит также от вещества,
из которого состоит тело.

## 3. Удельная теплоёмкость

Последняя из перечисленных зависимостей описывается отдельной физической величиной.

```{admonition} Удельная теплоёмкость
:class: important
Физическую величину, численно равную количеству теплоты, которое необходимо сообщить
телу массой 1 кг для изменения его температуры на 1 °C, называют **удельной
теплоёмкостью** вещества. Обозначается буквой $c$, единицей измерения служит джоуль
на килограмм-градус, Дж/(кг·°C).
```

Удельная теплоёмкость воды равна 4200 Дж/(кг·°C). Это означает, что для нагревания
одного килограмма воды на один градус ему необходимо передать 4200 Дж, а при остывании
килограмма воды на один градус выделяется такое же количество теплоты.

| Вещество | $c$, Дж/(кг·°C) | Вещество | $c$, Дж/(кг·°C) |
|---|---:|---|---:|
| Золото | 130 | Стекло | 840 |
| Ртуть | 138 | Кирпич | 880 |
| Свинец | 140 | Алюминий | 920 |
| Олово | 230 | Масло подсолнечное | 1700 |
| Серебро | 250 | Лёд | 2100 |
| Медь | 400 | Керосин | 2100 |
| Цинк, латунь | 400 | Спирт | 2500 |
| Железо | 460 | Молоко | 3900 |
| Сталь | 500 | **Вода** | **4200** |
| Чугун | 540 | | |

В таблице приведены значения из учебника, которыми и следует пользоваться при решении
задач. Два обстоятельства заслуживают внимания. Металлы располагаются в начале списка, а вода, обладающая наибольшей среди распространённых веществ удельной теплоёмкостью, в самом конце. Кроме того, одно и то же вещество в различных агрегатных
состояниях имеет различную теплоёмкость: у воды она составляет 4200, у льда
2100 Дж/(кг·°C).

```{admonition} Почему у воды такая большая теплоёмкость и что из этого следует
:class: note
Вода медленно нагревается и медленно остывает, запасая при этом значительное количество
энергии. По этой причине климат вблизи морей мягче континентального: летом море отбирает
тепло у воздуха, зимой отдаёт его. По той же причине воду применяют в системах отопления
и охлаждения двигателей, а грелку наполняют водой, а не песком.
```

## 4. Формула количества теплоты

Объединяя все три зависимости, получаем расчётную формулу.

```{admonition} Расчёт количества теплоты
:class: important
$$Q = c\,m\,(t_2 - t_1),$$

где $c$ — удельная теплоёмкость, $m$ — масса тела, $t_1$ и $t_2$ — начальная и конечная
температуры.

При нагревании $t_2 > t_1$ и $Q > 0$, тело получает теплоту. При охлаждении $t_2 < t_1$ и $Q < 0$, тело отдаёт теплоту. Часто говорят просто о количестве
теплоты, отданном при охлаждении, и берут его по модулю.
```

Пример. Определим количество теплоты, необходимое для нагревания 2 кг воды
от 20 до 100 °C.

````python
c = 4200        # удельная теплоёмкость воды, Дж/(кг·°C)
m = 2.0         # масса воды, кг
t1, t2 = 20, 100

Q = c * m * (t2 - t1)
print(f"Q = {c} · {m} · ({t2} − {t1}) = {Q:.0f} Дж = {Q / 1000:.0f} кДж")
````

Полученное значение составляет 670 кДж. Для сравнения, при подъёме мешка цемента
массой 50 кг на второй этаж совершается работа около полутора килоджоулей. Следовательно, чайник за пять минут передаёт воде в четыреста раз большее количество энергии.

## 5. Численный опыт: сравниваем вещества

Передадим одно и то же количество теплоты, равное 10 кДж, килограмму каждого вещества
из таблицы и определим, на сколько градусов оно при этом нагреется.

````python
import matplotlib.pyplot as plt
from phys import style

style.use()
teploemkost = {"свинец": 140, "медь": 400, "сталь": 500, "стекло": 840, "алюминий": 920,
               "масло": 1700, "лёд": 2100, "спирт": 2500, "вода": 4200}
Q, m = 10_000, 1.0
nagrev = {v: Q / (c * m) for v, c in teploemkost.items()}

fig, ax = style.axes("На сколько градусов нагреется 1 кг от 10 кДж", "",
                     "Чем больше удельная теплоёмкость, тем труднее нагреть")
ax.barh(list(nagrev), list(nagrev.values()))
plt.show()
for v in ("свинец", "вода"):
    print(f"{v:8s} нагреется на {nagrev[v]:5.1f} °C")
print(f"свинец греется легче воды в {nagrev['свинец'] / nagrev['вода']:.0f} раз")
````
