## 1. Возвращаемся к тепловому равновесию

На уроке 8 было показано, что тела разной температуры, приведённые в контакт, приходят
к общей температуре. Теперь, когда получена формула для количества теплоты, данный вопрос
можно поставить точно. Горячее тело остывает и **отдаёт** теплоту $Q_1$, холодное
нагревается и **получает** теплоту $Q_2$. Требуется установить, как связаны эти два
количества теплоты.

Прямой ответ даёт опыт учебника. В калориметр наливают 100 мл горячей воды, добавляют 100 мл
холодной, перемешивают и измеряют температуру смеси. По формуле $Q = cm\Delta t$
вычисляют отданное и полученное количества теплоты. Они оказываются **равными**
с точностью до небольших потерь.

Иного результата и не могло быть, так как энергия при теплообмене не возникает и не исчезает, а лишь переходит от одного тела к другому.

```{admonition} Уравнение теплового баланса
:class: important
Если теплообмен идёт только между двумя телами и без потерь, то количество теплоты,
отданное горячим телом, равно количеству теплоты, полученному холодным:

$$Q_{\text{отд}} = Q_{\text{пол}}, \qquad c_1 m_1 (t_1 - t) = c_2 m_2 (t - t_2),$$

где $t_1$ и $t_2$ — начальные температуры тел, $t$ — общая конечная температура.

Данное уравнение является математическим выражением **закона сохранения энергии при
теплообмене**. Оно справедливо, когда нет потерь в окружающую среду, над телами
не совершается работа и вещество не меняет агрегатного состояния.
```

**предскажите** конечную температуру по уравнению баланса, она показана зелёным пунктиром. Затем проследите, сходятся ли к ней кривые.

````python
from phys import sim
sim.calorimeter()
````

Модель позволяет сделать три наблюдения. Во-первых, кривые сближаются сначала быстро, затем всё медленнее, так как скорость теплопередачи пропорциональна убывающей разности температур. Во-вторых, конечная температура зависит не от того, чья температура выше,
а от произведения $cm$ каждого тела: 200 г алюминия остудят воду сильнее, чем 200 г свинца
при той же температуре. В-третьих, если в модели добавить потери, вода остывает в комнату,
и общая температура оказывается **ниже** расчётной. То же самое наблюдается
в лабораторной работе.

## 3. Как найти температуру смеси

Из уравнения баланса неизвестная $t$ выражается в одну строчку. Раскроем скобки
и соберём слагаемые с $t$:

$$c_1 m_1 t_1 + c_2 m_2 t_2 = (c_1 m_1 + c_2 m_2)\,t, \qquad
t = \frac{c_1 m_1 t_1 + c_2 m_2 t_2}{c_1 m_1 + c_2 m_2}.$$

Заучивать данную формулу не требуется, в задачах удобнее записать баланс и решить уравнение. Её смысл при этом следует понимать: общая температура является
**средним** начальных температур, взвешенным по величинам $cm$. Величину $cm$ называют
теплоёмкостью тела: она показывает, какое количество теплоты необходимо всему телу
для изменения температуры на один градус.

Пример. В 200 г воды при 20 °C опустили медный цилиндр массой 200 г при 90 °C.

````python
c_v, m_v, t_v = 4200, 0.200, 20      # вода
c_m, m_m, t_m = 400, 0.200, 90       # медь

t = (c_m * m_m * t_m + c_v * m_v * t_v) / (c_m * m_m + c_v * m_v)
Q_otd = c_m * m_m * (t_m - t)
Q_pol = c_v * m_v * (t - t_v)
print(f"общая температура t = {t:.1f} °C")
print(f"медь отдала   {Q_otd:6.0f} Дж")
print(f"вода получила {Q_pol:6.0f} Дж   (баланс сходится)")
````

Медь при 90 °C нагрела воду всего на шесть градусов из-за того, что её теплоёмкость $cm$ в десять раз меньше, чем у той же массы воды.

## 4. Численный опыт: от чего зависит температура смеси

Смешаем горячую воду при 80 °C с холодной при 20 °C, изменяя их массы.

````python
import numpy as np
import matplotlib.pyplot as plt
from phys import style

style.use()
dolya = np.linspace(0, 1, 200)              # доля горячей воды в смеси
t_smesi = 80 * dolya + 20 * (1 - dolya)     # для одного вещества c сокращается

fig, ax = style.axes("Доля горячей воды", "Температура смеси, °C",
                     "Смешивание воды: температура как взвешенное среднее")
ax.plot(dolya, t_smesi)
for d, podpis in [(0.25, "1 : 3"), (0.5, "1 : 1"), (0.75, "3 : 1")]:
    ax.plot(d, 80 * d + 20 * (1 - d), "o")
    ax.annotate(f"{podpis} → {80 * d + 20 * (1 - d):.0f} °C", (d, 80 * d + 20 * (1 - d)),
                textcoords="offset points", xytext=(8, -14))
plt.show()
````

Для одного и того же вещества удельная теплоёмкость сокращается, и температура смеси
определяется только отношением масс: при равных массах она лежит ровно посередине
и составляет 50 °C.

## 5. Калориметр

Уравнение баланса требует, чтобы теплота не уходила наружу. Для этого и служит **калориметр** (сосуд, устроенный так, чтобы теплообмен с окружающей средой был как можно меньше). Школьный калориметр представляет собой два стакана, вставленных один
в другой: между ними находится воздух, плохой проводник тепла, а внутренний стакан касается
внешнего лишь в нескольких точках. На следующем уроке измеряется, насколько он лучше
обычного стакана.

## 7. Задачи

**Задача 1 (Марон, № 107).** Кипяток объёмом 8 л смешали с холодной водой такого же объёма
при 15 °C. На сколько градусов понизилась температура кипятка?

````python
from phys import quiz
quiz.numeric("На сколько градусов понизилась температура кипятка?", answer=42.5, unit="°C", tol=0.02,
             explain="Массы равны, поэтому температура смеси равна среднему арифметическому: (100 + 15) / 2 = 57,5 °C. "
                     "Кипяток остыл на 100 − 57,5 = 42,5 °C. Ответ сборника: 42,5 °C.")
````

**Задача 2 (Марон, № 108).** В аквариуме 20 л воды при 14 °C. Сколько воды при 40 °C надо
добавить, чтобы установилась температура 20 °C?

````python
quiz.numeric("Сколько литров тёплой воды добавить?", answer=6, unit="л", tol=0.02,
             explain="Баланс: добавленная вода остывает от 40 до 20 °C, аквариумная греется "
                     "от 14 до 20 °C. Удельная теплоёмкость одна и сокращается: m · 20 = 20 · 6, "
                     "откуда m = 6 кг, что соответствует 6 л, как и в ответе сборника.")
````

**Задача 3 (Марон, № 109).** Стальной резец массой 300 г раскалили и опустили для закалки
в 0,8 л воды при 20 °C. Вода нагрелась до 62 °C. До какой температуры был нагрет резец?

````python
quiz.numeric("Температура резца до закалки, °C:", answer=1003, unit="°C", tol=0.02,
             explain="Вода получила Q = 4200 · 0,8 · (62 − 20) = 141 120 Дж. Столько же отдал резец: "
                     "500 · 0,3 · (t − 62) = 141 120, откуда t − 62 = 940,8 и t ≈ 1003 °C. "
                     "Это температура жёлтого каления стали. Ответ сборника: 1003 °C.")
````

**Задача 4 (учебник, упражнение 9).** Вода массой 2 кг остыла от 95 до 25 °C. На сколько
изменилась её внутренняя энергия?

````python
quiz.numeric("На сколько уменьшилась внутренняя энергия воды? Ответ в килоджоулях.", answer=588, unit="кДж", tol=0.02,
             explain="При остывании без совершения работы внутренняя энергия уменьшается ровно "
                     "на отданное количество теплоты: Q = 4200 · 2 · 70 = 588 000 Дж = 588 кДж.")
````
