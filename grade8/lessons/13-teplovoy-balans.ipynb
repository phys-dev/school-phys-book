## 1. Возвращаемся к тепловому равновесию

На уроке 8 мы видели: тела разной температуры, приведённые в контакт, приходят
к общей температуре. Теперь у нас есть формула для количества теплоты, и вопрос можно
поставить точно. Горячее тело остывает и **отдаёт** теплоту $Q_1$, холодное нагревается
и **получает** теплоту $Q_2$. Как связаны эти два количества?

Опыт учебника отвечает прямо. В калориметр наливают 100 мл горячей воды, добавляют 100 мл
холодной, перемешивают и измеряют температуру смеси. По формуле $Q = cm\Delta t$
вычисляют отданное и полученное количества теплоты. Они оказываются **равны** — с точностью
до небольших потерь.

Иначе и быть не могло: энергия при теплообмене не возникает и не исчезает, она лишь
переходит от одного тела к другому.

```{admonition} Уравнение теплового баланса
:class: important
Если теплообмен идёт только между двумя телами и без потерь, то количество теплоты,
отданное горячим телом, равно количеству теплоты, полученному холодным:

$$Q_{\text{отд}} = Q_{\text{пол}}, \qquad c_1 m_1 (t_1 - t) = c_2 m_2 (t - t_2),$$

где $t_1$ и $t_2$ — начальные температуры тел, $t$ — общая конечная температура.

Это уравнение — математическое выражение **закона сохранения энергии при теплообмене**.
Оно справедливо, когда нет потерь в окружающую среду, над телами не совершается работа
и вещество не меняет агрегатного состояния.
```

````python
from phys import sim
sim.calorimeter()
````

Три наблюдения стоят внимания. Кривые сближаются сначала быстро, потом всё медленнее:
скорость теплопередачи пропорциональна разности температур, а она убывает. Конечная
температура зависит не от того, чья температура выше, а от произведения $cm$ каждого тела:
200 г алюминия остудят воду сильнее, чем 200 г свинца при той же температуре. И третье —
добавьте потери: вода теперь остывает в комнату, и общая температура выходит **ниже**
расчётной. Ровно это вы увидите в лабораторной работе.

## 3. Как найти температуру смеси

Из уравнения баланса неизвестную $t$ выражают в одну строчку. Раскроем скобки
и соберём слагаемые с $t$:

$$c_1 m_1 t_1 + c_2 m_2 t_2 = (c_1 m_1 + c_2 m_2)\,t, \qquad
t = \frac{c_1 m_1 t_1 + c_2 m_2 t_2}{c_1 m_1 + c_2 m_2}.$$

Заучивать эту формулу не нужно: в задачах удобнее записать баланс и решить уравнение.
Но смысл её стоит понять — общая температура есть **среднее** начальных температур,
взвешенное по величинам $cm$. Величину $cm$ называют теплоёмкостью тела: она показывает,
сколько теплоты нужно всему телу на один градус.

Пример. В 200 г воды при 20 °C опустили медный цилиндр массой 200 г при 90 °C.

````python
c_v, m_v, t_v = 4200, 0.200, 20      # вода
c_m, m_m, t_m = 400, 0.200, 90       # медь

t = (c_m * m_m * t_m + c_v * m_v * t_v) / (c_m * m_m + c_v * m_v)
Q_otd = c_m * m_m * (t_m - t)
Q_pol = c_v * m_v * (t - t_v)
print(f"общая температура t = {t:.1f} °C")
print(f"медь отдала   {Q_otd:6.0f} Дж")
print(f"вода получила {Q_pol:6.0f} Дж   — баланс сходится")
````

Медь при 90 °C нагрела воду всего на шесть с половиной градусов: её теплоёмкость $cm$
в десять раз меньше, чем у той же массы воды.

## 4. Численный опыт: от чего зависит температура смеси

Смешаем горячую воду при 80 °C с холодной при 20 °C, меняя их массы.

````python
import numpy as np
import matplotlib.pyplot as plt
from phys import style

style.use()
dolya = np.linspace(0, 1, 200)              # доля горячей воды в смеси
t_smesi = 80 * dolya + 20 * (1 - dolya)     # для одного вещества c сокращается

fig, ax = style.axes("Доля горячей воды", "Температура смеси, °C",
                     "Смешивание воды: температура — взвешенное среднее")
ax.plot(dolya, t_smesi)
for d, podpis in [(0.25, "1 : 3"), (0.5, "1 : 1"), (0.75, "3 : 1")]:
    ax.plot(d, 80 * d + 20 * (1 - d), "o")
    ax.annotate(f"{podpis} → {80 * d + 20 * (1 - d):.0f} °C", (d, 80 * d + 20 * (1 - d)),
                textcoords="offset points", xytext=(8, -14))
plt.show()
````

Для одного и того же вещества удельная теплоёмкость сокращается, и температура смеси
определяется только отношением масс: равные массы — ровно посередине, 50 °C.

## 5. Калориметр

Уравнение баланса требует, чтобы теплота не уходила наружу. Для этого и нужен
**калориметр** — сосуд, устроенный так, чтобы теплообмен с окружающей средой был как можно
меньше. Школьный калориметр — это два стакана, вставленных один в другой: между ними
воздух, плохой проводник тепла, а внутренний стакан касается внешнего лишь в нескольких
точках. На следующем уроке мы измерим, насколько он лучше обычного стакана.
