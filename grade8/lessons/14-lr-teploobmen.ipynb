## Часть 2. Теплообмен при смешивании

1. Налейте в калориметр 100 мл горячей воды, а в стакан 100 мл холодной. Отмеряйте

   в холодной воде.
3. Не вынимая термометр из горячей воды, осторожно вылейте холодную воду в калориметр.
   Перемешайте термометром и, как только показание установится, запишите температуру
   смеси.

### Погрешности: что учитывать

Погрешность прямого измерения температуры принимайте равной цене деления термометра
(обычно 1 °C), а погрешность объёма равной цене деления мензурки. Измеренные величины
записывайте в виде $t_1 = 18 \pm 1$ °C. Результаты вычислений записывайте без указания
погрешности, но с разумным округлением до трёх значащих цифр.

Сравнивая $Q_{\text{отд}}$ и $Q_{\text{пол}}$, найдите их расхождение в процентах от большего.
Если оно не превышает 10–15 %, уравнение баланса можно считать подтверждённым, так как такая разница объясняется потерями и погрешностью термометра.

### Таблица измерений

| $m_1$, кг | $t_1$, °C | $m_2$, кг | $t_2$, °C | $t$, °C | $Q_{\text{пол}}$, Дж | $Q_{\text{отд}}$, Дж |
|---|---|---|---|---|---|---|
| холодная | холодная | горячая | горячая | смесь | | |

### Обработка результатов

После ввода измерений расчёт выполняется автоматически. Это удобно для проверки арифметики, но в тетради вычисления необходимо привести полностью.

````python
c = 4200                 # удельная теплоёмкость воды, Дж/(кг·°C)
# ↓ ваши измерения
m1, t1 = 0.100, 18.0     # холодная вода: масса, кг; температура, °C
m2, t2 = 0.100, 62.0     # горячая вода
t = 39.0                 # температура смеси
delta_t = 1.0            # цена деления термометра, °C

Q_pol = c * m1 * (t - t1)
Q_otd = c * m2 * (t2 - t)
raznica = abs(Q_otd - Q_pol) / max(Q_otd, Q_pol) * 100

print(f"холодная вода получила   Q_пол = 4200 · {m1} · ({t} − {t1}) = {Q_pol:.0f} Дж")
print(f"горячая вода отдала      Q_отд = 4200 · {m2} · ({t2} − {t}) = {Q_otd:.0f} Дж")
print(f"расхождение {raznica:.0f} %")
print()
# насколько погрешность термометра сказывается на результате
Q_pol_max = c * m1 * (t - t1 + 2 * delta_t)
print(f"если бы каждая из двух температур была измерена с ошибкой в {delta_t} °C в неудачную сторону,")
print(f"Q_пол составило бы {Q_pol_max:.0f} Дж, и один термометр объясняет расхождение до "
      f"{(Q_pol_max - Q_pol) / Q_pol * 100:.0f} %")
````
