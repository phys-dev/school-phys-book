## 2. Почему очевидное объяснение не работает

Напрашивается такое рассуждение: при нагревании молекулы движутся быстрее, сильнее
колеблются — значит, им нужно больше места, и тело расширяется.

Звучит убедительно. Но давайте проверим внимательно.

Возьмём две соседние молекулы. Пусть левая закреплена, а правая колеблется около
положения равновесия $r_0$. Чем выше температура, тем **больше размах** колебаний.
Молекула уходит и влево, и вправо.

```{admonition} В чём загвоздка
:class: important
Если молекула отклоняется влево ровно настолько же, насколько вправо, то её
**среднее** положение остаётся прежним — $r_0$. Среднее расстояние между молекулами
не меняется, и тело не расширяется!

Значит, одного увеличения размаха колебаний недостаточно. Нужно что-то ещё.
```

Учебник честно предупреждает: объяснить тепловое расширение «не так просто,
как кажется на первый взгляд». Давайте разберёмся.

## 3. Разгадка: яма несимметрична

Вспомним урок 3 и график силы между молекулами. При сближении молекул отталкивание
нарастает **очень круто**. При удалении притяжение спадает **полого**.

Это и есть разгадка. Изобразим не силу, а потенциальную энергию — получится «яма»,
на дне которой находится положение равновесия $r_0$. Стенки этой ямы разные:
левая — крутая, правая — пологая.

Молекула с некоторой энергией колеблется внутри ямы, доходя до её стенок.
Слева крутая стенка останавливает её быстро, справа пологая позволяет уйти дальше.

```{admonition} Ключевая мысль урока
:class: important
Максимальное отклонение молекулы влево **меньше**, чем вправо. Поэтому среднее
положение молекулы не совпадает с $r_0$, а **смещено вправо**.

Чем выше температура, тем больше энергия колебаний, тем сильнее это смещение —
и тем больше среднее расстояние между молекулами. **Это и есть тепловое расширение.**
```

Проверьте на модели. Меняйте температуру и следите за двумя числами в подписи: насколько молекула
отходит влево и насколько вправо.

````python
from phys import sim
sim.thermal_expansion()
````

Оранжевая линия — среднее положение молекулы, серая — положение равновесия $r_0$.
При низкой температуре они почти совпадают. Чем выше температура, тем дальше
оранжевая линия уходит вправо от серой.

## 4. Численный опыт: считаем расширение

Посчитаем то же самое, но числами. Найдём границы колебаний молекулы при разных
энергиях и посмотрим, куда смещается их середина.

````python
import numpy as np
import matplotlib.pyplot as plt
from phys import style

style.use()

def energiya(r, sigma=1.0):
    """Потенциальная энергия взаимодействия двух молекул (в единицах глубины ямы)."""
    x = (sigma / r) ** 6
    return 4 * (x**2 - x)

r0 = 2 ** (1 / 6)                     # положение равновесия
r = np.linspace(0.95, 3.0, 4000)
U = energiya(r)

def granicy(E):
    """Где молекула с энергией E останавливается — слева и справа."""
    vnutri = r[U <= E]
    return vnutri.min(), vnutri.max()

energii = np.linspace(-0.95, -0.15, 9)
sredniye, vlevo, vpravo = [], [], []
for E in energii:
    a, b = granicy(E)
    sredniye.append((a + b) / 2)
    vlevo.append(r0 - a)
    vpravo.append(b - r0)

fig, ax = style.axes("Расстояние между молекулами $r/r_0$", "Потенциальная энергия",
                     "Яма несимметрична: слева круто, справа полого")
ax.plot(r / r0, U)
for E in energii[::3]:
    a, b = granicy(E)
    ax.plot([a / r0, b / r0], [E, E], "--", linewidth=1.2)
    ax.plot([(a + b) / 2 / r0], [E], "o", markersize=5)
ax.axvline(1.0, color="gray", linestyle=":", linewidth=1)
ax.set_ylim(-1.05, 0.1)
ax.set_xlim(0.85, 2.2)
plt.show()

print(" энергия   влево   вправо   среднее   расширение")
for E, v, p, s in zip(energii, vlevo, vpravo, sredniye):
    print(f"  {E:5.2f}   {v:6.3f}   {p:6.3f}   {s / r0:7.4f}   {(s / r0 - 1) * 100:6.2f} %")
````

Смотрите на два столбца в середине: молекула всегда отходит вправо дальше, чем влево,
и с ростом энергии разрыв увеличивается. Точки на графике — середины отрезков
колебаний — уползают вправо.

Если бы яма была симметричной, эти точки шли бы строго по вертикали, и никакого
расширения не было бы.

## 5. Особенность воды

Почти все вещества при нагревании расширяются. Вода ведёт себя иначе:
**при нагревании от 0 до 4 °C её объём не увеличивается, а уменьшается**.
Только начиная с 4 °C вода начинает расширяться, как все.

Значит, при 4 °C плотность воды наибольшая.

```{admonition} Почему замерзающий водоём не промерзает до дна
:class: note
Осенью вода на поверхности остывает, становится плотнее и опускается вниз —
идёт перемешивание. Но как только поверхность остынет ниже 4 °C, всё меняется:
дальнейшее охлаждение делает воду **легче**, и холодные слои остаются наверху.

Поэтому у дна круглый год держится температура около 4 °C, а лёд образуется
только сверху. Лёд же ещё легче воды и плавает, закрывая водоём, как одеяло.

Если бы вода вела себя «как все», водоёмы промерзали бы до дна, и жизнь в них
зимой была бы невозможна.
```

```{admonition} Опыт 3. Измеряем расширение воды
:class: opyt
**Что нужно:** маленькая бутылка или пузырёк, прозрачная трубочка от сока,
пластилин, вода с красителем, миска с горячей водой, линейка.

**Что делать:**

1. Наполните пузырёк подкрашенной водой до краёв.
2. Вставьте трубочку и загерметизируйте горлышко пластилином так, чтобы вода
   поднялась в трубочку на 2–3 см. Отметьте уровень.
3. Поставьте пузырёк в миску с горячей водой.
4. Каждую минуту отмечайте уровень воды в трубочке.

**Результат:** уровень поднимается — вода расширяется. Измерьте, насколько
поднялся столбик, и посчитайте, на сколько процентов увеличился объём.

**Что записать:** объём пузырька, внутренний диаметр трубочки, подъём столбика,
начальную и конечную температуру.
```

Осталось перевести подъём столбика в прирост объёма.

````python
import math

# ↓ результаты вашего опыта
V_puzyrka = 50e-6          # объём пузырька, м³ (50 мл)
d_trubochki = 0.004        # внутренний диаметр трубочки, м (4 мм)
podyom = 0.06              # на сколько поднялся столбик, м (60 мм)
dT = 60                    # на сколько нагрели, °C

S = math.pi * (d_trubochki / 2) ** 2      # площадь сечения трубочки
dV = S * podyom                            # прирост объёма
otnositelno = dV / V_puzyrka

print(f"Объём вырос на {dV * 1e9:.0f} мм³ — это {otnositelno * 100:.2f} % от объёма")
print(f"При нагревании на {dT} °C")
print(f"На каждый градус объём растёт на {otnositelno / dT * 100:.4f} %")
print()
print("Справочное значение для воды: около 0,02 % на градус при 20 °C")
````
