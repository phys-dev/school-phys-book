Такое число трудно почувствовать, поэтому попробуем сравнить.

Любопытно, где на этой шкале оказывается атом: **ровно посередине** между размером

```{admonition} Наглядное сравнение
можно сказать, что вещество вокруг нас — в основном пустота.
```

## 2. Как измерить то, чего не видно: идея Рэлея

Молекулу не видно в микроскоп. Но её размер можно оценить простым опытом,
который предложил английский физик лорд Рэлей.

Капля масла, попав на чистую поверхность воды, растекается в тончайшую плёнку.
Растекаться бесконечно она не может: тоньше, чем **один слой молекул**, плёнка не станет.
Значит, если измерить площадь пятна и знать объём капли, толщину можно вычислить —
и это будет размер молекулы.

$$d = \frac{V}{S},$$

где $V$ — объём капли, $S$ — площадь пятна.

Значение получается около $10^{-9}$ м — правда, молекулы масла длинные, поэтому
выходит чуть больше размера атома. Но проверять это мы будем не на бумаге:
через минуту проделаем опыт сами и посчитаем по своим измерениям.

```{admonition} Что такое «порядок величины» и почему физики так любят оценки
:class: note
Оценить порядок величины — значит узнать ответ с точностью «до множителя в несколько раз»:
$10^{-9}$ м, а не $10^{-6}$ и не $10^{-12}$. Часто это важнее точного значения. Если оценка
показывает, что молекула в тысячу раз меньше, чем нужно для вашего прибора, — измерять
уже не надо, прибор не подойдёт. Умение быстро оценивать порядок величины — одна из главных
привычек физика.
```

## 3. Опыт своими руками: проводим измерение

Теперь измерим размер молекулы по-настоящему. Это один из немногих школьных опытов,
где своими руками добывают число из мира, которого не видно ни в один микроскоп.

Есть одна тонкость. Обычная капля масла слишком велика: чтобы растечься до слоя
в одну молекулу, ей понадобилось бы пятно площадью в десятки квадратных метров.
Поэтому масло сначала разводят спиртом — в капле раствора масла окажется в двести
раз меньше, и пятно получится размером с таз.

```{admonition} Опыт. Определение размера молекулы по масляной плёнке
:class: opyt
**Что нужно:** подсолнечное масло (лучше — олеиновая кислота), спирт, мензурка,
шприц или пипетка, широкий таз или поднос (диаметром не меньше 40 см), тальк
или детская присыпка (подойдёт растёртый мел), линейка.

**Что делать:**

1. **Приготовьте раствор.** Отмерьте 1 мл масла и растворите в 200 мл спирта.
   Хорошо перемешайте. Теперь в каждой капле раствора масла в 200 раз меньше,
   чем в капле чистого масла.
2. **Сосчитайте объём капли.** Наберите раствор в шприц и накапайте ровно 1 мл,
   считая капли. Разделите 1 мл на число капель — получите объём одной капли.
3. **Подготовьте поверхность.** Налейте в таз воду слоем 2–3 см, дайте ей успокоиться.
   Слегка припудрите поверхность тальком — ровным тонким слоем.
4. **Капните одну каплю** раствора в центр. Спирт растворится в воде, а масло
   растечётся в плёнку и раздвинет тальк — граница пятна будет хорошо видна.
5. **Измерьте диаметр пятна** линейкой в двух направлениях и возьмите среднее.

**Что записать:** число капель в 1 мл, диаметр пятна, разведение раствора.
```

```{admonition} Осторожно
:class: tb
Спирт огнеопасен: работайте вдали от открытого огня и нагревательных приборов,
проветривайте помещение. Опыт проводится под наблюдением учителя.
```

Теперь главный шаг — расчёт. Плёнка не может быть тоньше одной молекулы, поэтому
толщина плёнки и есть искомый размер:

$$d = \frac{V_{\text{масла}}}{S}.$$

Впишите **свои** измерения — код доведёт расчёт до конца.

````python
import math

# ↓ результаты вашего опыта
kapel_v_ml = 25          # сколько капель раствора помещается в 1 мл
razvedenie = 200         # во сколько раз масло разбавлено спиртом
diametr_pyatna = 0.30    # диаметр масляного пятна, м

V_kapli = 1e-6 / kapel_v_ml            # объём капли раствора, м³ (1 мл = 10⁻⁶ м³)
V_masla = V_kapli / razvedenie         # чистого масла в этой капле
S = math.pi * (diametr_pyatna / 2) ** 2  # площадь пятна
d = V_masla / S                        # толщина плёнки = размер молекулы

print(f"Объём капли раствора:  {V_kapli:.1e} м³")
print(f"Масла в капле:         {V_masla:.1e} м³")
print(f"Площадь пятна:         {S:.3f} м²")
print(f"Размер молекулы:       {d:.1e} м = {d * 1e9:.1f} нм")
````

Около двух-трёх нанометров — и это измерение, а не теория. Вы получили размер того,
чего не видно даже в самый сильный оптический микроскоп, с помощью таза воды,
линейки и щепотки талька.

```{admonition} Почему получилось чуть больше 10⁻¹⁰ м
:class: dropdown
Размер атома — около $10^{-10}$ м, а у нас вышло примерно $3 \cdot 10^{-9}$ м,
в тридцать раз больше. Ошибки здесь нет.

Молекула растительного масла — не шарик, а длинная цепочка из десятков атомов.
На поверхности воды такие молекулы встают «частоколом»: один конец молекулы
притягивается к воде, другой торчит наружу. Поэтому толщина плёнки равна **длине**
вытянутой молекулы, а она и составляет несколько нанометров.

Так что опыт измеряет именно то, что должен, — размер конкретной молекулы масла.
А заодно показывает, что молекулы бывают очень разной формы.
```

## 4. Ещё один способ: сосчитать через плотность

Есть и другой путь. Представим, что молекулы в жидкости упакованы вплотную, и на каждую
приходится маленький кубик. Тогда ребро такого кубика — это и есть размер молекулы.

Проверим на воде. Масса молекулы воды $m_0 = 3 \cdot 10^{-26}$ кг (откуда взялось
это число — прямо в следующем разделе), плотность воды 1000 кг/м³.

````python
m0 = 3.0e-26     # масса молекулы воды, кг
rho = 1000       # плотность воды, кг/м³

V0 = m0 / rho          # объём, приходящийся на одну молекулу
d = V0 ** (1 / 3)      # ребро кубика — размер молекулы

print(f"На одну молекулу приходится объём {V0:.1e} м³")
print(f"Размер молекулы воды: {d:.1e} м")
````

Получилось $3 \cdot 10^{-10}$ м — тот же порядок величины, что и в опыте с каплей.
Два совершенно разных способа дали согласующийся ответ. Именно так и проверяют
результат в физике: если к одному числу удаётся прийти разными путями, ему можно верить.

## 5. Масса атомов и молекул

Массы атомов измерить ещё труднее, чем размеры. Впервые это удалось в начале XX века
французскому физику **Жану Перрену** — тому самому, который изучал броуновское движение.
Он вычислил массу самого лёгкого атома — атома водорода:

$$m_{\text{H}} = 1{,}66 \cdot 10^{-27}\ \text{кг}.$$

Это невообразимо мало: чтобы получить один грамм, нужно взять шестьсот тысяч миллиардов
миллиардов атомов водорода.

К тому времени химики уже знали **относительные массы** остальных атомов — то есть
во сколько раз каждый атом тяжелее атома водорода. Поэтому открытие Перрена сразу
дало массы всех атомов.

```{admonition} Точности ради
:class: note
Число $1{,}66 \cdot 10^{-27}$ кг, которым мы будем пользоваться весь год, — это
**атомная единица массы**: одна двенадцатая часть массы атома углерода. Масса самого
атома водорода чуть больше, $1{,}674 \cdot 10^{-27}$ кг, то есть отличается меньше
чем на процент.

Считать удобнее в атомных единицах: относительные массы в таблице ниже даны именно
в них, и в школьных расчётах эту разницу в доли процента не учитывают.
```

| Атом | Во сколько раз тяжелее водорода | Масса, кг |
|---|---|---|
| Водород H | 1 | $1{,}66 \cdot 10^{-27}$ |
| Углерод C | 12 | $2{,}0 \cdot 10^{-26}$ |
| Азот N | 14 | $2{,}3 \cdot 10^{-26}$ |
| Кислород O | 16 | $2{,}7 \cdot 10^{-26}$ |
| Железо Fe | 56 | $9{,}3 \cdot 10^{-26}$ |
| Медь Cu | 64 | $1{,}1 \cdot 10^{-25}$ |
| Свинец Pb | 207 | $3{,}4 \cdot 10^{-25}$ |

Чтобы найти массу **молекулы**, достаточно сложить массы атомов, из которых она состоит.
Молекула воды $\text{H}_2\text{O}$ — это два атома водорода и один кислорода:

$$m_0 = 2 \cdot 1 + 16 = 18 \text{ масс водорода} = 18 \cdot 1{,}66 \cdot 10^{-27}
= 3{,}0 \cdot 10^{-26}\ \text{кг}.$$

````python
m_H = 1.66e-27          # масса атома водорода, кг

def massa_molekuly(otnositelnaya_massa):
    """Масса молекулы по её массе в единицах массы атома водорода."""
    return otnositelnaya_massa * m_H

for nazvanie, otn in [("вода H₂O", 2 + 16),
                      ("кислород O₂", 16 * 2),
                      ("углекислый газ CO₂", 12 + 16 * 2),
                      ("атом меди Cu", 64)]:
    print(f"{nazvanie:22s} {otn:3d} масс водорода   →   {massa_molekuly(otn):.2e} кг")
````

## 6. Сколько молекул в теле

Теперь мы умеем главное. Если известна масса тела $m$ и масса одной молекулы $m_0$,
то число молекул находится делением:

```{admonition} Формула, которой будем пользоваться весь год
:class: important
$$N = \frac{m}{m_0}$$

Число молекул в теле равно массе тела, делённой на массу одной молекулы.
```

Решим задачу из учебника: **сколько молекул воды содержится в капле объёмом 50 мм³?**

````python
V = 50e-9              # объём капли: 50 мм³ = 50·10⁻⁹ м³
rho = 1000             # плотность воды, кг/м³
m0 = 18 * 1.66e-27     # масса молекулы воды, кг

m = rho * V            # масса капли
N = m / m0             # число молекул

print(f"Масса капли:     {m:.1e} кг  (это {m * 1e6:.0f} мг)")
print(f"Число молекул:   {N:.2e}")
print(f"Прописью: около {N / 1e21:.1f} секстиллиона молекул")
````

Полтора секстиллиона — это $1{,}7 \cdot 10^{21}$, то есть 1 700 000 000 000 000 000 000.
В одной капле воды. Если бы каждую молекулу можно было пересчитать со скоростью
миллион штук в секунду, на эту каплю ушло бы в сто тысяч раз больше времени,
чем существует Вселенная.

## 7. Физика рядом: сколько электронов в пучке коллайдера

В новосибирском Институте ядерной физики работает электрон-позитронный коллайдер
**ВЭПП-4М** — кольцо периметром 366 метров, по которому летят электроны и позитроны.
Сколько электронов в таком пучке?

Поштучно их, конечно, не считают. Измеряют **ток пучка**, а число частиц находят
расчётом. Ток — это заряд, проходящий через сечение трубы за секунду. Пучок летит почти
со скоростью света и успевает пролететь мимо нас столько раз в секунду, сколько делает
оборотов по кольцу. Значит, зная ток $I$ и время одного оборота $T$, получаем заряд
всего пучка $I \cdot T$, а разделив его на заряд одного электрона — число частиц:

$$N = \frac{I \cdot T}{e}, \qquad T = \frac{P}{c}.$$

````python
P = 366            # периметр кольца ВЭПП-4М, м
I = 0.8e-3         # ток пучка, А (типичное рабочее значение — около 1 мА)
c = 3.0e8          # электроны летят почти со скоростью света, м/с
e = 1.6e-19        # заряд электрона, Кл

T = P / c                  # время одного оборота по кольцу
N = I * T / e              # число электронов в пучке

print(f"Один оборот занимает {T * 1e6:.2f} мкс — это {1 / T:.1e} оборотов в секунду")
print(f"Электронов в пучке:  {N:.1e}")
````

Шесть миллиардов электронов, и каждый успевает обежать 366-метровое кольцо
почти миллион раз в секунду.

А теперь сравним пучок с той самой каплей воды:

````python
N_kaplya = 1.67e21         # молекул в капле воды объёмом 50 мм³ (посчитали выше)
N_puchok = 6.1e9           # электронов в пучке ВЭПП-4М

print(f"В капле воды частиц больше, чем в пучке, в {N_kaplya / N_puchok:.0e} раз")
````

В крошечной капле воды частиц в **сотни миллиардов раз** больше, чем во всём пучке
ускорителя. По меркам вещества пучок — это почти ничто.

Именно поэтому в вакуумной камере так важно откачать остаточный газ: даже в очень
хорошем вакууме молекул воздуха оказывается больше, чем частиц в самом пучке.

**Задача Б. Во сколько раз плотность атомного ядра больше плотности воды?**

Почти вся масса атома сосредоточена в ядре, а размер ядра в 100 000 раз меньше
размера атома.

````python
m_yadra = 1.66e-27      # масса ядра водорода (протона), кг
r_yadra = 1e-15         # размер ядра, м

V = 4 / 3 * 3.14 * r_yadra ** 3
rho = m_yadra / V

print(f"Плотность ядра: {rho:.0e} кг/м³")
print(f"Это в {rho / 1000:.0e} раз больше плотности воды")
print(f"Один кубический сантиметр такого вещества весил бы {rho * 1e-6 / 1e9:.0f} млн тонн")
````

```{admonition} А теперь взгляд в небо
:class: tip
Такая плотность существует не только в ядре. Когда массивная звезда выгорает,
её ядро сжимается до состояния, в котором вещество состоит из плотно упакованных
нейтронов — получается **нейтронная звезда**. Её плотность как раз около
$10^{17}$ кг/м³ — та же, что мы сейчас посчитали для атомного ядра.

Нейтронная звезда — атомное ядро размером с город: масса больше солнечной,
а диаметр около 20 км. Чайная ложка её вещества весила бы около 500 миллионов тонн —
примерно столько же, сколько весит всё человечество разом.
```
