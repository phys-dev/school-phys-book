# Урок 2. Масса и размер атомов и молекул

В настоящем уроке рассматриваются размеры и массы атомов и молекул. Размер молекулы измеряется в опыте с масляной плёнкой, а масса молекулы вычисляется по составу вещества, что позволяет подсчитать число молекул в произвольном теле.

## 1. Насколько малы молекулы

насколько эти частицы малы.

Размер атома составляет около $10^{-10}$ м (одна десятимиллиардная доля метра).
Поскольку такое число трудно представить наглядно, обратимся к сравнениям.

Каждый шаг ползунка вправо увеличивает размер в десять раз.

````python
from phys import sim
sim.scale_ladder()
````

Рассмотрим положение атома на данной шкале. От ядра до атома пять шагов, и ядро меньше атома в сто тысяч раз. От атома до человека ещё десять шагов, и атом меньше человека в десять миллиардов раз. Для сравнения, Солнце больше человека всего
в миллиард раз.

```{admonition} Наглядное сравнение

Ядро атома в этой вишне окажется размером в одну десятитысячную миллиметра, меньше бактерии. Почти весь объём атома занимает движение электронов, поэтому можно
сказать, что вещество вокруг нас состоит в основном из пустоты.
```

## 2. Как измерить то, чего не видно: идея Рэлея

Молекулу невозможно увидеть в микроскоп. Тем не менее её размер можно определить простым опытом, предложенным английским физиком лордом Рэлеем.

Капля масла на чистой воде растекается в тончайшую плёнку. Бесконечно растекаться она не может, так как плёнка не станет тоньше **одного слоя молекул**. Дальнейшее сводится к арифметике: деление объёма капли на измеренную площадь пятна даёт толщину слоя, равную размеру молекулы.

$$d = \frac{V}{S},$$

где $V$ — объём капли, $S$ — площадь пятна.

Значение получается около $10^{-9}$ м. Молекулы масла вытянуты, поэтому результат
выходит несколько больше размера атома. Данный результат проверяется не на бумаге:
далее опыт проводится самостоятельно, и расчёт выполняется по собственным измерениям.

```{admonition} Что такое «порядок величины» и почему физики так любят оценки
:class: note
Оценить порядок величины означает узнать ответ с точностью до множителя в несколько раз:
$10^{-9}$ м, а не $10^{-6}$ и не $10^{-12}$. Часто такая оценка важнее точного значения.
Если оценка показывает, что молекула в тысячу раз меньше, чем требуется для имеющегося
прибора, то измерение проводить не нужно, прибор заведомо не подойдёт. Умение быстро
оценивать порядок величины является одной из главных привычек физика.
```

## 3. Опыт своими руками: проводим измерение

Перейдём к реальному измерению. Школьных опытов, в которых число, характеризующее
невидимый мир, добывается собственными руками, совсем немного, и данный опыт является
одним из них.

Здесь имеется одна тонкость. Обычная капля масла для опыта слишком велика. Чтобы
растечься до одного слоя молекул, ей потребовалось бы пятно площадью в десятки
квадратных метров. По этой причине масло сначала разводят спиртом. В капле раствора
масла окажется в двести раз меньше, и пятно уместится в тазу.

```{admonition} Опыт. Определение размера молекулы по масляной плёнке
:class: opyt
**Что нужно:** подсолнечное масло (лучше олеиновая кислота), спирт, мензурка,
шприц или пипетка, широкий таз или поднос (диаметром не меньше 40 см), тальк
или детская присыпка (подойдёт растёртый мел), линейка.

**Что делать:**

1. **Приготовьте раствор.** Отмерьте 1 мл масла и растворите в 200 мл спирта.
   Хорошо перемешайте. Теперь в каждой капле раствора масла в 200 раз меньше,
   чем в капле чистого масла.
2. **Сосчитайте объём капли.** Наберите раствор в шприц и накапайте ровно 1 мл,
   считая капли. Разделите 1 мл на число капель: получится объём одной капли.
3. **Подготовьте поверхность.** Налейте в таз воду слоем 2–3 см, дайте ей успокоиться.
   Слегка припудрите поверхность тальком ровным тонким слоем.
4. **Капните одну каплю** раствора в центр. Спирт растворится в воде, а масло
   растечётся в плёнку и раздвинет тальк, так что граница пятна будет хорошо видна.
5. **Измерьте диаметр пятна** линейкой в двух направлениях и возьмите среднее.

**Что записать:** число капель в 1 мл, диаметр пятна, разведение раствора.
```

```{admonition} Осторожно
:class: tb
Спирт огнеопасен: работайте вдали от открытого огня и нагревательных приборов,
проветривайте помещение. Опыт проводится под наблюдением учителя.
```

Остаётся главное, расчёт. Плёнка не может быть тоньше одной молекулы, поэтому
её толщина и является искомым размером:

$$d = \frac{V_{\text{масла}}}{S}.$$

Впишите **свои** измерения, и расчёт будет доведён до конца.

````python
import math

# ↓ результаты вашего опыта
kapel_v_ml = 25          # сколько капель раствора помещается в 1 мл
razvedenie = 200         # во сколько раз масло разбавлено спиртом
diametr_pyatna = 0.30    # диаметр масляного пятна, м

V_kapli = 1e-6 / kapel_v_ml            # объём капли раствора, м³ (1 мл = 10⁻⁶ м³)
V_masla = V_kapli / razvedenie         # чистого масла в этой капле
S = math.pi * (diametr_pyatna / 2) ** 2  # площадь пятна
d = V_masla / S                        # толщина плёнки = размер молекулы

print(f"Объём капли раствора:  {V_kapli:.1e} м³")
print(f"Масла в капле:         {V_masla:.1e} м³")
print(f"Площадь пятна:         {S:.3f} м²")
print(f"Размер молекулы:       {d:.1e} м = {d * 1e9:.1f} нм")
````

Полученное значение составляет два-три нанометра, и это число получено самостоятельно
с помощью таза воды, линейки и щепотки талька.

```{admonition} Почему получилось чуть больше 10⁻¹⁰ м
:class: dropdown
Размер атома составляет около $10^{-10}$ м, тогда как в опыте получилось примерно $3 \cdot 10^{-9}$ м, в тридцать раз больше. Ошибки здесь нет.

Молекула растительного масла представляет собой не шарик, а длинную цепочку из десятков
атомов. На поверхности воды такие молекулы встают «частоколом»: один конец молекулы
притягивается к воде, другой обращён наружу. Поэтому толщина плёнки равна **длине**
вытянутой молекулы, которая и составляет несколько нанометров.

Следовательно, опыт измеряет то, что и должен измерять, размер конкретной молекулы масла. Кроме того, он показывает, что молекулы бывают весьма различной формы.
```

## 4. Ещё один способ: сосчитать через плотность

К тому же числу можно прийти иначе. Предположим, что молекулы в жидкости упакованы
вплотную и на каждую приходится небольшой кубик. Ребро кубика и будет размером
молекулы.

Проверим данное предположение на воде. Масса молекулы воды $m_0 = 3 \cdot 10^{-26}$ кг
(происхождение этого числа разобрано в следующем разделе), плотность воды 1000 кг/м³.

````python
m0 = 3.0e-26     # масса молекулы воды, кг
rho = 1000       # плотность воды, кг/м³

V0 = m0 / rho          # объём, приходящийся на одну молекулу
d = V0 ** (1 / 3)      # ребро кубика — размер молекулы

print(f"На одну молекулу приходится объём {V0:.1e} м³")
print(f"Размер молекулы воды: {d:.1e} м")
````

Получилось $3 \cdot 10^{-10}$ м, тот же порядок величины, что и в опыте с каплей. Два
совершенно разных пути привели к одному ответу.

Так в физике проверяют результат: числу, к которому удаётся прийти с разных сторон, можно доверять.

## 5. Масса атомов и молекул

Массы атомов измерить ещё труднее, чем размеры. Впервые это удалось в начале XX века
французскому физику **Жану Перрену**, тому самому, который изучал броуновское движение.
Он вычислил массу самого лёгкого из атомов, атома водорода:

$$m_{\text{H}} = 1{,}66 \cdot 10^{-27}\ \text{кг}.$$

Для представления о малости этой величины отметим, что для набора одного грамма
потребуется шестьсот тысяч миллиардов миллиардов атомов водорода.

К тому времени химикам уже были известны **относительные массы** остальных атомов (во сколько раз каждый атом тяжелее водородного). Поэтому открытие Перрена сразу
дало массы всех атомов.

### Как он это сделал

Взвесить атом на весах невозможно. Перрен пошёл обходным путём и воспользовался
броуновскими частицами, за которыми сам же наблюдал в микроскоп.

Рассуждение строилось следующим образом. Частицы в жидкости со всех сторон толкают
молекулы воды, и ведут себя эти частицы подобно молекулам газа. Газ, как известно,
распределяется по высоте неравномерно: у поверхности Земли воздух плотнее, с высотой
он разрежается. Плотность падает вдвое примерно каждые пять километров. Причина
заключается в соревновании двух начал: тяжесть тянет молекулы вниз, тепловое движение
разбрасывает их вверх, и равновесие между ними определяет, насколько быстро убывает
концентрация с высотой.

То же самое должно происходить и со взвешенными частицами. При этом они в триллионы раз тяжелее молекулы, а значит, их «атмосфера» будет во столько же раз ниже, не километры, а микроны.

Перрен приготовил суспензию гуммигута, отобрал частицы одинакового размера, поместил
каплю под микроскоп и подсчитывал, сколько частиц попадает в поле зрения на разной
глубине, наводя резкость то выше, то ниже. Работа была кропотливой: счёт шёл на тысячи
частиц.

Результат подтвердил предположение. Концентрация падала вдвое на каждые несколько десятков микрометров, и получалась атмосфера высотой в толщину волоса.

Далее вступает в силу арифметика. То, насколько быстро падает концентрация, зависит
от массы частицы: чем она тяжелее, тем ниже её атмосфера. Масса частицы была известна Перрену по измеренным размеру и плотности вещества. Сравнив измеренное падение
концентрации с тепловой энергией молекул, он получил **число Авогадро** (количество молекул в моле вещества).

````python
import numpy as np

# параметры опыта Перрена
r = 0.21e-6          # радиус частиц гуммигута, м
rho_chastic = 1194   # плотность гуммигута, кг/м³
rho_vody = 1000      # плотность воды, кг/м³
T = 293              # температура, К
g, R = 9.8, 8.31     # ускорение свободного падения и газовая постоянная

V = 4 / 3 * np.pi * r**3                  # объём частицы
m = (rho_chastic - rho_vody) * V          # масса с поправкой на выталкивающую силу

# наблюдение: на сколько падает концентрация при подъёме на 30 мкм
h = 30e-6
otnoshenie = 1.73                          # во столько раз меньше частиц наверху

N_A = R * T * np.log(otnoshenie) / (m * g * h)

print(f"Масса частицы (с учётом воды): {m:.2e} кг")
print(f"Число Авогадро из опыта:       {N_A:.2e} 1/моль")
print(f"Масса атома водорода:          {0.001 / N_A:.2e} кг")
````

Из числа Авогадро масса атома получается делением: если моль водорода имеет массу один
грамм и содержит $N_A$ атомов, то на каждый атом приходится $0{,}001/N_A$ килограмма.

Перрен получил значения от $6{,}5$ до $7 \cdot 10^{23}$, что при кропотливости счёта
частиц под микроскопом весьма близко к принятому сегодня $6{,}02 \cdot 10^{23}$.

### Второй способ: по дрожанию частиц

Одного измерения Перрену было недостаточно. Спор о существовании атомов продолжался
не первое десятилетие, и результат следовало подтвердить независимо. Второй способ он
построил на том самом броуновском движении, с которого начинался предыдущий урок.

Там было доказано, что средний квадрат смещения растёт пропорционально времени.
Эйнштейн пошёл дальше и вычислил, чему равен множитель:

$$\langle x^2 \rangle = 2Dt, \qquad D = \frac{RT}{N_A \cdot 6\pi\eta a}.$$

Здесь $\eta$ — вязкость жидкости, $a$ — радиус частицы, $D$ — коэффициент диффузии.
Данная формула относится к программе десятого класса, но существенно в ней другое.
В правой части стоит $N_A$, а всё остальное измеримо: вязкость воды известна, радиус
частицы виден в микроскоп, температура определяется по термометру. Следовательно,
измерив смещения, можно вычислить число Авогадро.

Перрен так и поступил. Он отмечал положение одной и той же частицы через каждые
30 секунд, зарисовывая её путь по клеткам окулярной сетки. Получалась ломаная,
похожая на построенную ранее численно. Смещения за каждый интервал он возводил
в квадрат и усреднял по сотням наблюдений.

````python
eta = 1.0e-3         # вязкость воды при 20 °C, Па·с
a_ch = 0.37e-6       # радиус частицы, м
t_nabl = 30          # интервал между отсчётами, с
x2 = 3.48e-11        # средний квадрат смещения по наблюдениям, м²

print(f"Типичное смещение за {t_nabl} с: {np.sqrt(x2) * 1e6:.1f} мкм")
print("Столько и видно в микроскоп при увеличении в несколько сотен раз")
print()

N_A_2 = R * T * t_nabl / (3 * np.pi * eta * a_ch * x2)
print(f"Число Авогадро вторым способом: {N_A_2:.2e} 1/моль")
print(f"Первым способом получилось:     {N_A:.2e} 1/моль")
````

Два метода не имеют между собой ничего общего. В первом частицы **стоят**, и
подсчитывается, как их концентрация убывает с высотой. Во втором частицы **движутся**,
и измеряется, насколько далеко они уходят за минуту. В первом работает тяжесть,
во втором вязкость. При этом оба метода дают одно и то же число.

Такое совпадение не бывает случайным, и оно убедило современников в том, что молекулы
существуют и их можно пересчитать. Вековой спор о существовании атомов на этом
закончился, а в 1926 году Перрен получил Нобелевскую премию с формулировкой «за работу
по дискретной природе вещества».

```{admonition} Что здесь самое удивительное
:class: note
Атом не был увиден. Он был взвешен путём наблюдения за частицами, которые в тысячи раз
крупнее и которые сами по себе отношения к атомам не имеют.

Так в физике устроено измерение вообще: величину, недоступную прибору,
связывают цепочкой рассуждений с величиной, которую измерить можно. Перрен считал видимые
частицы под микроскопом, а получил массу невидимого атома.
```

```{admonition} Точности ради
:class: note
Число $1{,}66 \cdot 10^{-27}$ кг, которое используется в течение всего года, представляет собой **атомную единицу массы**, одну двенадцатую часть массы атома углерода.
Масса самого атома водорода несколько больше, $1{,}674 \cdot 10^{-27}$ кг, и отличается меньше чем на процент.

Считать удобнее в атомных единицах. Относительные массы в таблице ниже даны в этих единицах, и в школьных расчётах разницу в доли процента не учитывают.
```

| Атом | Во сколько раз тяжелее водорода | Масса, кг |
|---|---|---|
| Водород H | 1 | $1{,}66 \cdot 10^{-27}$ |
| Углерод C | 12 | $2{,}0 \cdot 10^{-26}$ |
| Азот N | 14 | $2{,}3 \cdot 10^{-26}$ |
| Кислород O | 16 | $2{,}7 \cdot 10^{-26}$ |
| Железо Fe | 56 | $9{,}3 \cdot 10^{-26}$ |
| Медь Cu | 64 | $1{,}1 \cdot 10^{-25}$ |
| Свинец Pb | 207 | $3{,}4 \cdot 10^{-25}$ |

Чтобы найти массу **молекулы**, достаточно сложить массы атомов, из которых она состоит.
Молекула воды $\text{H}_2\text{O}$ состоит из двух атомов водорода и одного атома
кислорода:

$$m_0 = 2 \cdot 1 + 16 = 18 \text{ масс водорода} = 18 \cdot 1{,}66 \cdot 10^{-27}
= 3{,}0 \cdot 10^{-26}\ \text{кг}.$$

````python
m_H = 1.66e-27          # масса атома водорода, кг

def massa_molekuly(otnositelnaya_massa):
    """Масса молекулы по её массе в единицах массы атома водорода."""
    return otnositelnaya_massa * m_H

for nazvanie, otn in [("вода H₂O", 2 + 16),
                      ("кислород O₂", 16 * 2),
                      ("углекислый газ CO₂", 12 + 16 * 2),
                      ("атом меди Cu", 64)]:
    print(f"{nazvanie:22s} {otn:3d} масс водорода   →   {massa_molekuly(otn):.2e} кг")
````

## 6. Сколько молекул в теле

Теперь имеется всё необходимое для главного расчёта. Если известна масса тела $m$
и масса одной молекулы $m_0$, то число молекул находится делением:

```{admonition} Формула, которой будем пользоваться весь год
:class: important
$$N = \frac{m}{m_0}$$

Число молекул в теле равно массе тела, делённой на массу одной молекулы.
```

Решим задачу из учебника: **сколько молекул воды содержится в капле объёмом 50 мм³?**


````python
V = 50e-9              # объём капли: 50 мм³ = 50·10⁻⁹ м³
rho = 1000             # плотность воды, кг/м³
m0 = 18 * 1.66e-27     # масса молекулы воды, кг

m = rho * V            # масса капли
N = m / m0             # число молекул

print(f"Масса капли:     {m:.1e} кг  (это {m * 1e6:.0f} мг)")
print(f"Число молекул:   {N:.2e}")
print(f"Прописью: около {N / 1e21:.1f} секстиллиона молекул")
````

В одной капле содержится 1 700 000 000 000 000 000 000 молекул. Если считать их
по миллиону в секунду, то на одну эту каплю уйдёт больше пятидесяти миллионов лет.

## 7. Физика рядом: сколько электронов в пучке коллайдера

В новосибирском Институте ядерной физики работает электрон-позитронный коллайдер
**ВЭПП-4М**, представляющий собой кольцо периметром 366 метров, по которому летят
электроны и позитроны. Определим, сколько электронов содержится в таком пучке.

Поштучно их, разумеется, не считают. Измеряют **ток пучка**, а число частиц находят
расчётом. Ток представляет собой заряд, проходящий через сечение трубы за секунду.
Пучок движется почти со скоростью света и успевает пройти мимо наблюдателя столько раз
в секунду, сколько делает оборотов по кольцу. Следовательно, зная ток $I$ и время одного
оборота $T$, получаем заряд всего пучка $I \cdot T$, а разделив на заряд одного
электрона, находим число частиц:

$$N = \frac{I \cdot T}{e}, \qquad T = \frac{P}{c}.$$


````python
P = 366            # периметр кольца ВЭПП-4М, м
I = 0.8e-3         # ток пучка, А (типичное рабочее значение — около 1 мА)
c = 3.0e8          # электроны летят почти со скоростью света, м/с
e = 1.6e-19        # заряд электрона, Кл

T = P / c                  # время одного оборота по кольцу
N = I * T / e              # число электронов в пучке

print(f"Один оборот занимает {T * 1e6:.2f} мкс, что соответствует {1 / T:.1e} оборотам в секунду")
print(f"Электронов в пучке:  {N:.1e}")
````

В пучке содержится шесть миллиардов электронов, и каждый из них успевает обойти
366-метровое кольцо почти миллион раз в секунду.

Сравним теперь пучок с рассмотренной выше каплей воды.


````python
N_kaplya = 1.67e21         # молекул в капле воды объёмом 50 мм³ (посчитали выше)
N_puchok = 6.1e9           # электронов в пучке ВЭПП-4М

print(f"В капле воды частиц больше, чем в пучке, в {N_kaplya / N_puchok:.0e} раз")
````

В крошечной капле воды частиц в **сотни миллиардов раз** больше, чем во всём пучке
ускорителя. По меркам вещества пучок представляет собой почти ничто.

По этой причине в вакуумной камере столь важно откачать остаточный газ. Даже в очень
хорошем вакууме молекул воздуха оказывается больше, чем частиц в самом пучке.

**Задача Б. Во сколько раз плотность атомного ядра больше плотности воды?**

Почти вся масса атома сосредоточена в ядре, а размер ядра в 100 000 раз меньше
размера атома.

````python
m_yadra = 1.66e-27      # масса ядра водорода (протона), кг
r_yadra = 1e-15         # размер ядра, м

V = 4 / 3 * 3.14 * r_yadra ** 3
rho = m_yadra / V

print(f"Плотность ядра: {rho:.0e} кг/м³")
print(f"Это в {rho / 1000:.0e} раз больше плотности воды")
print(f"Один кубический сантиметр такого вещества весил бы {rho * 1e-6 / 1e9:.0f} млн тонн")
````

```{admonition} А теперь взгляд в небо
:class: tip
Такая плотность существует не только в ядре. Когда массивная звезда выгорает,
её ядро сжимается до состояния, в котором вещество состоит из плотно упакованных
нейтронов, и образуется **нейтронная звезда**. Её плотность составляет около $10^{17}$ кг/м³, того же порядка, что получено выше для атомного ядра.

Нейтронная звезда представляет собой атомное ядро размером с город: масса больше
солнечной, а диаметр около 20 км. Чайная ложка её вещества весила бы около 500 миллионов тонн, примерно столько же, сколько весит всё человечество.
```
