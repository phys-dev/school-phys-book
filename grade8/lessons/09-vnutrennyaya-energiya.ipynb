## 1. Куда уходит механическая энергия

Известны два вида механической энергии. **Потенциальная** — энергия взаимодействия тел
или частей одного тела; **кинетическая** — энергия движущихся тел.

Опыт из учебника. Свинцовый шар поднимают над свинцовой плитой — он приобретает
потенциальную энергию. Шар падает, потенциальная энергия переходит в кинетическую. Затем
удар: шар лежит на плите, и механической энергии у него нет. Куда она исчезла?

Осмотрим шар и плиту. Оба деформированы, а термометр показал бы, что оба нагрелись.
Изменилось состояние тел: расположение молекул и скорость их движения. Механическая
энергия шара перешла в энергию молекул шара и плиты.

## 2. Внутренняя энергия

В любом веществе молекулы находятся в тепловом движении и потому обладают кинетической
энергией. Они взаимодействуют друг с другом и потому обладают потенциальной энергией.

```{admonition} Внутренняя энергия
:class: important
Сумму кинетической энергии теплового движения всех молекул тела и потенциальной энергии
их взаимодействия между собой называют **внутренней энергией** тела.

Внутренняя энергия не зависит от механического движения тела и его положения
относительно других тел. Она определяется внутренним состоянием: характером движения
и взаимным расположением молекул.
```

Учебник поясняет это на планёре. В полёте у него есть и кинетическая, и потенциальная
энергия; на земле обеих нет. А внутренняя энергия в обоих случаях одинакова, если
одинакова температура: молекулы движутся так же и расположены так же.

От чего зависит внутренняя энергия? От температуры: чем она выше, тем больше кинетическая
энергия молекул. И от агрегатного состояния: при переходе воды в пар молекулы расходятся,
и потенциальная энергия их взаимодействия меняется, хотя температура может оставаться
прежней.

## 3. Первый способ: совершение работы

Потрите металлический стержень тканью, и он нагреется. Над стержнем совершена работа,
его внутренняя энергия увеличилась.

Второй опыт из учебника показательнее. В толстостенную трубку наливают немного эфира
и плотно закрывают пробкой. Трубку туго обтягивают верёвкой и быстро протягивают её
взад-вперёд. Через некоторое время эфир закипает, и пар выбивает пробку. Работа сил трения
увеличила внутреннюю энергию трубки настолько, что эфир нагрелся до кипения.

Работает и обратное. В банку с узким горлом накачивают воздух. Пробка вылетает, а внутри
появляется туман — капельки воды. Воздух, выталкивая пробку, совершил работу и при этом
резко охладился, поэтому водяной пар в нём сконденсировался.

```{admonition} Работа и внутренняя энергия
:class: important
Внутреннюю энергию тела можно увеличить, **совершив над телом работу**. За счёт
уменьшения внутренней энергии **само тело может совершить работу**.
```

В модели ниже газ заперт под поршнем. Сдвиньте поршень вниз: молекулы, отскакивая
от движущегося навстречу поршня, ускоряются, и газ нагревается, хотя тепло ему никто
не передавал. Отведите поршень — молекулы замедляются.

````python
from phys import sim
sim.piston()
````

Кнопка «Нагреть дно» понадобится в четвёртом разделе.

Отчего именно ускоряются молекулы, легко проверить расчётом. Молекула, летящая навстречу
поршню, отражается от него как мяч от движущейся ракетки: к её скорости добавляется
удвоенная скорость поршня.

````python
import numpy as np
rng = np.random.default_rng(2)

def szhatie(dlina_do, dlina_posle, u=0.01, n=4000):
    # одномерный газ между дном и поршнем; поршень медленно движется к дну
    x = rng.uniform(0, dlina_do, n)
    v = rng.normal(0, 1, n)
    L, dt = dlina_do, 0.05
    while L > dlina_posle:
        x += v * dt
        L -= u * dt
        otskok = x < 0
        x[otskok] = -x[otskok]; v[otskok] = -v[otskok]                 # дно
        udar = x > L
        x[udar] = 2 * L - x[udar]; v[udar] = -v[udar] - 2 * u          # движущийся поршень
    return np.mean(v**2 / 2)

e0 = 0.5
for szhat in (2, 3):
    e1 = szhatie(10.0, 10.0 / szhat)
    print(f"объём уменьшен в {szhat} раза: средняя энергия молекул выросла в {e1 / e0:.1f} раза")
````

Каждый удар о движущийся поршень добавляет молекуле скорости — так работа поршня
превращается во внутреннюю энергию газа. При медленном сжатии одномерного газа вдвое
энергия молекул растёт вчетверо, и модель это воспроизводит.

## 4. Второй способ: теплопередача

Внутренняя энергия тела меняется и без всякой работы. Поднесите к концу металлического
стержня пламя спиртовки — стержень нагреется целиком. Опустите нагретый стержень в воду:
он остынет, вода нагреется.

```{admonition} Теплопередача
:class: important
Процесс изменения внутренней энергии тела **без совершения работы** называют
**теплопередачей** (теплообменом).

Теплопередача происходит только при наличии разности температур, причём энергия всегда
передаётся от тела с большей температурой к телу с меньшей.
```

Вернитесь к модели и нажмите «Нагреть дно», не двигая поршень. Молекулы, отскакивающие
от горячего дна, приобретают скорость, и внутренняя энергия газа растёт, хотя работы над
ним никто не совершает. «Охладить дно» даёт обратный эффект.

```{admonition} Итог
:class: important
Внутреннюю энергию тела можно изменить двумя способами: **совершением работы**
и **теплопередачей**.
```

```{admonition} Копнуть глубже: теплород и опыт Румфорда
:class: dropdown
Ещё в XVIII веке учёные спорили о том, что такое теплота. Одна точка зрения: теплота —
особая невесомая жидкость, **теплород**, перетекающая от горячего тела к холодному. Она
хорошо объясняла теплопередачу. Другая, которую отстаивал М. В. Ломоносов в работе
«Размышления о причине теплоты и холода» (1744), — теплота есть движение частиц тела.

Решающий довод против теплорода дал в 1798 году Бенджамин Томпсон, граф Румфорд.
На пушечном заводе он наблюдал сверление стволов и заметил, как сильно нагреваются ствол
и стружка. Румфорд поместил заготовку в бочку с водой и сверлил тупым сверлом: через два
с половиной часа вода закипела. Теплород должен был бы исчерпаться, а тепло выделялось
сколько угодно долго, пока совершалась работа. Значит, тепло не вещество, а результат
работы, то есть движение.
```
