Третья лабораторная работа по программе. В учебнике такой работы нет, поэтому описание
составлено по образцу лабораторных работ 2 и 3 учебника: тот же калориметр, тот же метод
теплового баланса — только теперь одно из тел меняет агрегатное состояние.

```{admonition} Цель работы
:class: important
Измерить удельную теплоту плавления льда и сравнить результат с табличным значением

```

**Приборы и материалы:** калориметр, термометр, весы, измерительный цилиндр, тающий лёд
(кусочки, полежавшие в воде), бумажные салфетки, тёплая вода около 40 °C.

```{admonition} Техника безопасности

```

## Идея измерения

Кусочек **тающего** льда имеет температуру ровно 0 °C — это важно, иначе пришлось бы
учитывать его нагрев. Опускаем лёд в тёплую воду. Вода остывает от $t_1$ до $t$, отдавая
теплоту $c\,m_{\text{в}}(t_1 - t)$. Лёд эту теплоту получает и тратит на две вещи: сначала
тает при 0 °C — $\lambda m_{\text{л}}$, — потом талая вода нагревается от 0 до $t$ —
$c\,m_{\text{л}}\,t$. Уравнение баланса:

$$c\,m_{\text{в}}\,(t_1 - t) = \lambda\,m_{\text{л}} + c\,m_{\text{л}}\,t, \qquad
\lambda = \frac{c\,m_{\text{в}}\,(t_1 - t) - c\,m_{\text{л}}\,t}{m_{\text{л}}}.$$

## Ход работы

1. Налейте в калориметр 150 мл тёплой воды, взвесьте калориметр с водой: так вы узнаете
   массу воды $m_{\text{в}}$ (масса пустого калориметра известна или взвешена заранее).
2. Измерьте температуру воды $t_1$.
3. Возьмите два-три небольших кусочка тающего льда, быстро промокните их салфеткой
   (капли воды на льду исказят массу) и опустите в калориметр.
4. Аккуратно помешивая термометром, дождитесь, когда весь лёд растает и температура
   перестанет падать. Запишите $t$.
5. Снова взвесьте калориметр: прибавка массы — это масса льда $m_{\text{л}}$. Взвешивать
   лёд отдельно хуже: он тает на весах и обрастает водой.

### Погрешности

Как и раньше: прямые измерения — с ценой деления прибора (термометр ±1 °C, весы по шкале),
результат — с округлением до двух значащих цифр. Затем найдите, на сколько процентов ваше
$\lambda$ отличается от табличного. Отклонение до 15 % — нормальный результат школьного
опыта.

### Таблица измерений

| $m_{\text{в}}$, кг | $t_1$, °C | $m_{\text{л}}$, кг | $t$, °C | $\lambda$, Дж/кг | Табличное | Отклонение, % |
|---|---|---|---|---|---|---|
| | | | | | 3,4·10⁵ | |

## Обработка результатов

````python
c = 4200
# ↓ ваши измерения
m_v = 0.150      # масса воды, кг
t1 = 40.0        # начальная температура воды, °C
m_l = 0.030      # масса льда, кг
t = 21.5         # температура после таяния, °C

Q_otd = c * m_v * (t1 - t)             # отдала тёплая вода
Q_nagrev = c * m_l * t                 # ушло на нагрев талой воды от 0 до t
lam = (Q_otd - Q_nagrev) / m_l
otkl = (lam - 3.4e5) / 3.4e5 * 100
print(f"вода отдала {Q_otd:.0f} Дж; из них {Q_nagrev:.0f} Дж пошло на нагрев талой воды")
print(f"на плавление осталось {Q_otd - Q_nagrev:.0f} Дж на {m_l * 1000:.0f} г льда")
print(f"λ = {lam:.2e} Дж/кг, отклонение от табличного {otkl:+.0f} %")
````

### Что портит результат сильнее всего

````python
def lam_iz(m_v, t1, m_l, t):
    return (c * m_v * (t1 - t) - c * m_l * t) / m_l

baza = lam_iz(0.150, 40.0, 0.030, 21.5)
proby = [("температура воды t₁ на 1 °C", lam_iz(0.150, 41.0, 0.030, 21.5)),
         ("конечная температура t на 1 °C", lam_iz(0.150, 40.0, 0.030, 22.5)),
         ("масса льда на 1 г", lam_iz(0.150, 40.0, 0.031, 21.5)),
         ("масса воды на 1 г", lam_iz(0.151, 40.0, 0.030, 21.5))]
for opis, znach in proby:
    print(f"ошибка: {opis:34s} → λ меняется на {(znach - baza) / baza * 100:+5.1f} %")
````

Два измерения решают всё: температуры и масса льда. Градус ошибки в температуре — это
семь-восемь процентов результата; лишний грамм воды, взвешенный вместе со льдом, — ещё
три. Отсюда правила: сушить лёд салфеткой, ждать окончания таяния, брать льда побольше.

Куда смещён результат? Если на льду осталась вода, «лёд» окажется тяжелее, чем есть,
и $\lambda$ выйдет **меньше** табличного. Если лёд взят прямо из морозилки, холоднее 0 °C,
на его нагрев уйдёт лишняя теплота, и $\lambda$ выйдет **больше**. Потери в стенки
калориметра занижают $\lambda$. Сопоставьте знак своего отклонения с этими причинами.
