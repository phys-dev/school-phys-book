# Урок 28. Контрольная работа «Тепловые явления. Изменение агрегатных состояний вещества»

Первая контрольная работа по программе. Время — 40 минут. Шесть заданий: одно качественное
с объяснением через строение вещества, четыре расчётных, одно на влажность.

**Даны:** $c$ воды 4200, льда 2100, меди 400, стали 500 Дж/(кг·°C); $\lambda$ льда
$3{,}4 \cdot 10^5$ Дж/кг; $L$ воды $2{,}3 \cdot 10^6$ Дж/кг; $q$ бензина $4{,}6 \cdot 10^7$, дров
$1{,}0 \cdot 10^7$ Дж/кг; плотность насыщенного пара при 10 °C — 9,4 г/м³, при 20 °C — 17,3 г/м³,
при 25 °C — 23,0 г/м³.

## Вариант 1

1. Почему при плавлении льда его температура не повышается, хотя лёд получает теплоту?
   Объясните на основе представлений о строении вещества.
2. Какое количество теплоты необходимо, чтобы нагреть 3 кг воды от 20 до 70 °C?
3. Смешали 200 г воды при 80 °C и 300 г воды при 20 °C. Какая температура установится?
4. Сколько теплоты нужно, чтобы лёд массой 2 кг при 0 °C превратить в воду при 50 °C?
5. Двигатель автомобиля сжёг 500 г бензина и совершил полезную работу 5,75 МДж. Найдите КПД
   двигателя.
6. При 20 °C в 1 м³ воздуха содержится 12,1 г водяного пара. Найдите относительную влажность
   воздуха.

## Вариант 2

1. Почему при снегопаде становится теплее, а во время ледохода у реки холодно? Объясните
   с точки зрения превращений энергии.
2. Какое количество теплоты отдаст медная деталь массой 500 г, остывая на 80 °C?
3. В 400 г воды при 20 °C опустили стальную деталь массой 200 г при 120 °C. Какая температура
   установится?
4. Сколько теплоты нужно, чтобы 100 г воды при 30 °C превратить в пар при 100 °C?
5. Сколько дров нужно сжечь, чтобы нагреть 50 л воды от 10 до 60 °C, если в воду попадает 30 %
   выделившейся теплоты?
6. При 25 °C относительная влажность воздуха 40 %. Найдите абсолютную влажность и определите,
   выпадет ли роса, если ночью воздух остынет до 10 °C.
